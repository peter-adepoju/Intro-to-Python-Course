```python
def paths(steps):
    if steps == 0:
        return 1
    if steps < 0:
        return 0
    return paths(steps - 1) + paths(steps - 2)

print([paths(n) for n in range(7)])
```

*Interactive figure on the website: Each call branches into two: take 1 stair, or take 2. Calls that land exactly on 0 count as one way; calls that overshoot count as none. Notice how often the same call repeats..*

> **Key idea.** In tree recursion, each call tries every choice for one decision, recursing once per choice, and combines what comes back. The base cases say what happens when there's nothing left to decide, or when a choice has gone wrong.

```python
def ways(amount, coins):
    if amount == 0:
        return 1
    if amount < 0 or not coins:
        return 0
    use_first = ways(amount - coins[0], coins)
    skip_first = ways(amount, coins[1:])
    return use_first + skip_first

print(ways(10, [1, 2, 5]))
print(ways(100, [1, 2, 5, 10, 20, 50]))
```

> **Common mistake.** Choosing "which coin next?" from all of them at every step counts `1 + 2` and `2 + 1` as different ways. If order doesn't matter, decide about one item at a time: use it, or move past it for good.

```python
def routes(rows, cols):
    if rows == 1 or cols == 1:
        return 1
    return routes(rows - 1, cols) + routes(rows, cols - 1)

print(routes(2, 2), routes(3, 3), routes(4, 6))
```

```python
def climbs(steps):
    """Every way to climb `steps` stairs, 1 or 2 at a time, as lists of step sizes."""
    if steps == 0:
        return [[]]
    if steps < 0:
        return []
    return [[1] + rest for rest in climbs(steps - 1)] + [[2] + rest for rest in climbs(steps - 2)]

for c in climbs(4):
    print(c)
```

```python
def binary_strings(n):
    if n == 0:
        return [""]
    shorter = binary_strings(n - 1)
    return ["0" + s for s in shorter] + ["1" + s for s in shorter]

print(binary_strings(3))
```

```python
calls = 0
def paths(steps):
    global calls
    calls += 1
    if steps <= 0:
        return 1 if steps == 0 else 0
    return paths(steps - 1) + paths(steps - 2)

for n in (10, 15, 20, 25):
    calls = 0
    paths(n)
    print(n, calls)
```

```python
def can_make(amount, coins):
    """Can `amount` be made exactly from the coins, each used at most once?"""
    if amount == 0:
        return True
    if amount < 0 or not coins:
        return False
    return can_make(amount - coins[0], coins[1:]) or can_make(amount, coins[1:])

print(can_make(12, [5, 3, 9, 2]), can_make(1, [5, 3, 9, 2]))
```

### Bug hunt: Counted twice

`coin_ways(amount, coins)` should count the ways to make `amount` where order doesn't matter: for 4 with coins 1 and 2 there are 3 ways (1+1+1+1, 1+1+2, 2+2). It returns 5, because it counts 1+1+2, 1+2+1 and 2+1+1 separately. Fix it so it decides about one coin at a time.

```python
def coin_ways(amount, coins):
    if amount == 0:
        return 1
    if amount < 0:
        return 0
    return sum(coin_ways(amount - c, coins) for c in coins)
```

## Warm-up

Those are the Fibonacci numbers again. `paths(n)` counts the ways to climb n stairs taking 1 or 2 at a time: your first step is either 1 stair or 2, and after that you have a smaller staircase. Two choices, two recursive calls.

## Choices make trees

**Tree recursion** is recursion where a call makes **more than one** recursive call. Each call branches, so the calls form a tree. It fits any problem where you face a **choice**:

1. Pick one decision to make now (the first stair: 1 or 2? the first coin: use it or not?).
2. For **each** option, recurse on the smaller problem that's left.
3. Combine the results: add them up to count the ways, collect them to list the ways, or take the best to optimise.

### Making change

How many ways can you make 10 naira from coins of 1, 2 and 5 (in any amounts, where order doesn't matter)? The trick is to choose **about one coin type at a time**: either you use at least one of the first type, or you never use it.

Read it with the leap of faith: every way of making the amount either uses at least one `coins[0]` (so take one off and make the rest, still allowed to use it again) or uses none (so make the whole amount with the other coins). Those two groups don't overlap, and together they're everything, so adding them counts each way exactly once.

## Listing the ways

To **list** the ways instead of counting them, build each answer from the choice made now, plus each of the answers to the smaller problem:

Two base cases with very different answers:

- `[[]]` for 0 stairs: there is **one** way to climb no stairs, which is to do nothing. That one way is the empty list.
- `[]` for a negative number: there are **no** ways.

Getting these right is the whole trick. With `[]` for 0, every list would be empty, because there would be nothing to put a first step in front of.

### Binary strings

All the strings of n binary digits: each is either `"0"` or `"1"`, followed by every string of n - 1 digits.

Notice `shorter` is worked out **once** and used twice. Calling `binary_strings(n - 1)` twice would double the work at every level.

## How big is the tree?

The number of calls is the size of the tree. If each call makes **two** calls and the depth goes down by **one** each time, the tree has about 2 to the power n calls:

| n | 2 to the n |
|---|---|
| 10 | about a thousand |
| 20 | about a million |
| 30 | about a billion |
| 40 | about a trillion |

Each extra level doubles the work. That's why `paths(40)` takes noticeably long, and `paths(100)` would never finish. Sometimes the size really is that big: there are 2 to the n binary strings of length n, so listing them all can't be faster. But often the tree repeats the same calls over and over, and tomorrow you'll learn to make those repeats free.

Sometimes you only want to know **whether** there's a way, not how many. Then stop as soon as one branch succeeds, with `or` (which short-circuits) or an early `return`:

You can also **prune**: skip branches that can't possibly work. If the coins are sorted from smallest to largest and the smallest is already bigger than the amount, nothing in that branch can succeed, so return straight away. Pruning doesn't change the answer, only how much of the tree you visit.
