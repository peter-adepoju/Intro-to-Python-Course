````python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
````

### Exercise: Forge: a folder explorer

Write these recursive functions. All of them take a folder dictionary and none of them change it.

1. `size(folder)`: the total size of every file inside, at any depth.
2. `find(folder, suffix)`: a sorted list of the paths of every file whose name ends with `suffix`, written from the top like `src/utils/text.py`.
3. `largest_folder(folder)`: the `(path, size)` of the sub-folder (at any depth, not counting the top folder itself) with the biggest total size. Ties go to the path that comes first alphabetically. `None` if there are no sub-folders.
4. `prune(folder)`: a **new** folder with every empty sub-folder removed, including folders that only become empty once their own empty sub-folders are removed.
5. `tree(folder)`: the folder drawn as lines of text, like the `tree` command. Entries are shown in alphabetical order; each sub-folder's name ends with `/`; each line is prefixed with `├── ` for every entry but the last at that level and `└── ` for the last; and under an entry, deeper lines are indented with `│   ` if that entry wasn't the last, or four spaces if it was. It returns the lines as a list.

> **Five functions, one shape.** Look at your five functions side by side. Every one loops over the entries of one folder and, for each sub-folder, makes the **same** recursive call on it. What differs is only what each one does with a file, and how it combines the results. `find` and `tree` pass information **down** (the path, the prefix); `size` and `largest_folder` pass results **up**; `prune` builds a new tree on the way up. Once you see that shape, every new question about a tree is a small variation.

## Forge bonus: expression trees

When Python reads `1 + 2 * 3`, it first builds a **tree**: a `+` whose left side is `1` and whose right side is a `*` of `2` and `3`. Then it works out the value recursively, from the leaves up. Here's that tree written as nested lists, operator first: `["+", 1, ["*", 2, 3]]`.
