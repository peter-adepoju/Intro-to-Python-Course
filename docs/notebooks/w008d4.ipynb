### Bug hunt: Lost leftovers

`merge` loses some items: `merge([1, 5, 9], [2])` gives `[1, 2]`. Fix it.

````python
def merge(a, b):
    i = j = 0
    result = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    return result
````

### Bug hunt: The wrong position

`best_position` should return the position of the largest value, but for `[3, 9, 2]` it returns 9. Fix it.

````python
def best_position(values):
    best = 0
    for i in range(len(values)):
        if values[i] > values[best]:
            best = values[i]
    return best
````

## Filter and transform

**Filter** keeps the items that pass a test. **Transform** (also called **map**) changes every item. Both build a **new** list and leave the original alone:

They combine naturally: filter, then transform.

(Next week you'll meet **list comprehensions**, a one-line way to write exactly these two patterns.)

## Built-ins that replace loops

Several patterns are so common that Python has a function for them:

| Pattern | Built-in |
|---|---|
| accumulate a sum | `sum(items)` |
| best so far | `max(items)`, `min(items)`, with `key=` if needed |
| count | `items.count(value)`, or `len` of a filtered list |
| any of | `any(...)` |
| all of | `all(...)` |

`any` and `all` take a list of `True`/`False` values:

`any([])` is `False` (nothing passed) and `all([])` is `True` (nothing failed). Next week's comprehensions will let you write `all(m >= 50 for m in marks)` in one line.
