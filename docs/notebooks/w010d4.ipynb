````python
classes = {
    "3A": [{"name": "Ada", "mark": 93}, {"name": "Bo", "mark": 41}],
    "3B": [{"name": "Cy", "mark": 78}, {"name": "Di", "mark": 66}],
}
averages = {code: sum(s["mark"] for s in ss) / len(ss) for code, ss in classes.items()}
print(averages)
everyone = [s for ss in classes.values() for s in ss]
best = max(everyone, key=lambda s: s["mark"])
print(best["name"])
````

````python
rows = [("3A", "Ada", 93), ("3B", "Cy", 78), ("3A", "Bo", 41)]
classes = {}
for code, name, mark in rows:
    classes.setdefault(code, []).append({"name": name, "mark": mark})
print(classes)
````

````python
import copy
original = {"name": "Ada", "marks": [93, 88]}
shallow = original.copy()
deep = copy.deepcopy(original)
shallow["marks"].append(10)
deep["marks"].append(99)
print(original)
print(deep)
````

> **Key idea.** To change nested data without touching the original, use `copy.deepcopy`. A shallow `.copy()` is enough only when you'll replace whole values, not change them inside.

````python
import json
data = {"name": "Ada", "marks": [93, 88], "passed": True, "tutor": None}
text = json.dumps(data)
print(text)
print(type(text))
back = json.loads(text)
print(back["marks"][0])
````

## Warm-up

`student["marks"]` is a list, so you can index it, take its length, and append to it, like any other list. That one idea is all of today: each pair of square brackets takes you one level deeper.

## Shapes of data

There are three common shapes, and choosing the right one makes the rest of the program simpler.

**A list of dictionaries**: many records of the same kind, in order. Each dictionary has the same keys. This is the shape of a table, a spreadsheet, or rows from a database.

**A dictionary of lists**: a list of things for each key. It's what the grouping pattern produces.

**A dictionary of dictionaries**: a record for each key, found at once by that key. It's what the index pattern produces.

## Reaching in

Read a chain of square brackets from left to right, one step at a time. Each step gives a new value, and the next bracket applies to that value:

`school["classes"]` is a list; `[0]` is its first item, a dictionary; `["students"]` is a list in that; `[1]` is its second name.

Explore it yourself. Click any part to see the expression that reaches it:

### When a level might be missing

Any step can fail: a key may be missing, a list may be too short. Use `get` at the levels where something may legitimately be missing, with a default of the right **type**, so the next step still works:

`u.get("settings", {})` gives an empty dictionary when there are no settings, and `.get("theme", "light")` on an empty dictionary gives `"light"`.

## Copies

You saw in Week 8 that `.copy()` is shallow: the new outer container holds the **same** inner objects. With nested data that matters:

Changing the shallow copy's list changed the original's list, because they're the same list. `copy.deepcopy` copies every level, so the deep copy is completely separate.

## JSON

**JSON** (JavaScript Object Notation) is a text format for nested data. It looks almost exactly like Python's lists and dictionaries, and it's how most websites and apps send data to each other. The `json` module converts both ways:

`json.dumps` ("dump to string") turns data into JSON text; `json.loads` ("load from string") turns JSON text back into data. Notice the small differences in the text: JSON writes `true`, `false` and `null` where Python has `True`, `False` and `None`, and always uses double quotes.
