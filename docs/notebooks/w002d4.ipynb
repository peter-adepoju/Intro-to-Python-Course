## More than two choices: elif

Suppose a mark of 70 or more is an A, 60 or more a B, 50 or more a C, and anything lower an F. With only `if` and `else` you could nest them, but Python has a neater way, `elif` (short for "else if"):

How Python runs a chain:

1. Check the `if` condition. If it's true, run its block and **skip the rest of the chain**.
2. Otherwise check the first `elif`. If it's true, run its block and skip the rest.
3. And so on down the chain.
4. If nothing was true, run the `else` block.

So **exactly one** block runs, and it's the **first** one whose condition is true. Move the slider and watch Python work down the chain:

With 64, `mark >= 60` is true, so the grade is B. `mark >= 50` is also true, but Python never even looks at it.

## The order matters

Because the first true condition wins, the order of the conditions is part of the logic. Here is the same chain with the order mixed up:

Everyone with 50 or more gets a C, because `mark >= 50` is checked first and is true for all of them. The B and A branches can never run. Python doesn't warn you: the code is legal, just wrong.

## elif or separate ifs?

An `elif` chain picks **one** branch. Separate `if` statements are separate questions: each is checked, and any number of them can run.

With 75, all three blocks run and the program prints A, B and C. That's wrong for grades, which are one answer. But separate `if`s are right when the questions are independent:

(`.isalpha()` is `True` when every character is a letter. More methods like it in Week 5.) A weak password can fail all three tests, and the user should hear about all of them.

Deep nesting is hard to read, because you have to keep every outer condition in your head. Often `and` lets you flatten it:
