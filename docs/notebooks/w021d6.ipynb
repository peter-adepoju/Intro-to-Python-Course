### Exercise: Forge: snap, a tiny version control tool

Write `main(argv)` returning `(exit_code, output_lines)`, using `argparse` subcommands and the files in the **current folder** (only plain files directly in it, not in sub-folders, and never anything in `.snap`):

- `init`: create `.snap/store.json` with no commits; output `Initialised snap`. If it already exists: exit code 1, output `Already a snap folder`.
- `commit -m MESSAGE`: snapshot every file's text. If nothing has changed since the head commit: exit 1, output `Nothing to commit`. Otherwise store a commit with an ID from `commit_id` (in the starter), make it the head, and output `[ID] MESSAGE`.
- `log`: one line `ID MESSAGE` per commit from the head back; if there are none, `No commits yet`.
- `status`: for each file that differs from the head commit, sorted by name, a line `new: NAME`, `modified: NAME` or `deleted: NAME`; if there's nothing, `Nothing changed`.
- `diff`: for each modified file, sorted, a line `--- NAME` then the changed lines from `difflib.ndiff` (those starting `- ` or `+ `).
- `restore NAME`: write the head commit's version of the file back; output `Restored NAME`; exit 1 with `No such file in the last commit: NAME` if it isn't there.

Every command except `init` gives exit code 1 and `Not a snap folder (run snap init)` if there's no store.

````python
# Run this cell first: it creates the files used below.
with open('notes.txt', 'w', encoding='utf-8') as f:
    f.write('Buy eggs\nCall Ada\n')
with open('todo.txt', 'w', encoding='utf-8') as f:
    f.write('Finish the Forge\n')
````

````python
import argparse
import difflib
import hashlib
import json
import os

STORE = os.path.join(".snap", "store.json")

def commit_id(files, message, parent):
    lines = [f"parent {parent or 'none'}", f"message {message}"]
    for name in sorted(files):
        lines.append(f"file {name} {files[name]}")
    return hashlib.sha1("\n".join(lines).encode()).hexdigest()[:7]
````

## Forge: snap, a tiny version control tool

`snap` keeps snapshots of the text files in the current folder, in a hidden folder called `.snap`. It's used like this:

```
snap init
snap commit -m "First draft"
snap status
snap diff
snap log
snap restore notes.txt
```

It stores everything in `.snap/store.json`: a dictionary with `"commits"` (commit ID to `{"message", "parent", "files"}`) and `"head"` (the newest commit's ID, or `null`). This page has two files to start with.
