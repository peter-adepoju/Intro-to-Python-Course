````python
total = 0
best_name, best_mark = "", -1
with open("marks.txt") as f:
    for line in f:
        name, mark = line.strip().split(",")
        mark = int(mark)
        total += mark
        if mark > best_mark:
            best_name, best_mark = name, mark
print("Total:", total)
print("Best:", best_name, best_mark)
````

## What a text file is

A **text file** is a sequence of characters stored on a disk, with a name like `marks.txt`. There is nothing special about lines: a line is just the characters up to a newline character, `"\n"`. Most text files end with a newline after the last line too.

## Opening and closing

To use a file, you **open** it, which gives you a **file object**. You read from the file object, then **close** it:

`open("poem.txt")` looks for the file in the **current folder**, the folder the program is running in. `f.read()` reads the whole file into one string. `f.close()` tells the operating system you've finished, so it can free what it set aside.

Forgetting `close`, or crashing before you reach it, leaves the file open. For reading that's usually harmless; for writing (tomorrow) it can lose data. So Python has a statement that closes the file for you, however the block ends:

`with open(...) as f:` opens the file and names it `f`. When the indented block ends, normally or with an error, the file is closed. Notice `text` is still there after the block: only the file is closed.

### Encodings

A file stores **bytes**, and an **encoding** says how to turn bytes into characters. Almost every text file today uses **UTF-8**, which can store every alphabet and emoji. Python usually assumes UTF-8, but on some computers (older Windows set-ups in particular) it assumes something else, and names like "Adébáyọ̀" come out garbled. To be safe, say so:

```
with open("names.txt", encoding="utf-8") as f:
```

The examples on this site are short; in your own programs, write `encoding="utf-8"`.

## Ways to read

There are four ways to read a text file. Try them on a real file, and watch the red position marker:

### The whole file: read()

`f.read()` gives the whole file as one string, newlines included. It's simplest for small files.

`text.splitlines()` splits it into lines **without** the newline characters, which is often what you want.

### One line: readline()

`f.readline()` gives the next line, **including** its `"\n"`. At the end of the file it gives `""`.

### All lines as a list: readlines()

`f.readlines()` gives a list of every line, each still ending in `"\n"`.
