## A loop inside a loop

That prints 12 lines. Here's what Python does:

1. The outer loop sets `i` to 0.
2. The inner loop runs **completely**: `j` is 0, 1, 2, 3.
3. The outer loop moves on: `i` is 1. The inner loop starts **again from the beginning**: `j` is 0, 1, 2, 3 again.
4. And once more for `i` = 2.

So the inner variable changes fastest, like the seconds on a clock, and the outer one only ticks over when the inner loop has finished, like the minutes. Step through it:

## Shapes, row by row

Nested loops are natural for anything with **rows** and **columns**. The outer loop goes down the rows; the inner loop goes across one row.

Look at the `print()` with nothing in it. It's inside the outer loop but **outside** the inner one, so it runs once per row, after the row is complete, to start a new line. Getting that one line's indentation right is half the battle with shapes.

## How much work?

Two nested loops over `n` things do `n × n` passes of the inner body. That grows fast:

| n | n × n |
|---|---|
| 10 | 100 |
| 1,000 | 1,000,000 |
| 100,000 | 10,000,000,000 |

At 100,000 things, a computer that does 100 million simple steps a second would need well over a minute. Three nested loops are `n × n × n`. When a program is mysteriously slow, look for nested loops over large things first. In Phase 2 you'll learn to describe this precisely, and to find faster ways.
