> **Key idea.** A recursive definition has a **base case** answered directly, and a **recursive case** that uses the same definition on a smaller problem, getting closer to the base case every time.

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(4))
print(factorial(10))
````

````python
def total_to(n):
    if n == 0:
        return 0
    return n + total_to(n - 1)

print(total_to(4), total_to(100))
````

````python
def power(x, n):
    if n == 0:
        return 1
    return x * power(x, n - 1)

print(power(2, 10), power(3, 4))
````

````python
def digits(n):
    """How many digits the whole number n (0 or more) has."""
    if n < 10:
        return 1
    return 1 + digits(n // 10)

print(digits(7), digits(2026), digits(1000000))
````

> **Common mistake.** Forgetting the base case, or a recursive case that doesn't get smaller, makes the calls go on until Python gives up with `RecursionError: maximum recursion depth exceeded`. Python allows about 1,000 calls waiting at once. When you see that error, check the base case and check that every call moves towards it.

````python
def broken(n):
    return n * broken(n - 1)

print(broken(3))
````

## Warm-up

`countdown` calls `countdown`. That's allowed: a function can call any function, including itself. Each call has a smaller `n`, and when `n` reaches 0, it stops calling. That's recursion.

## Defining things in terms of themselves

Many things are most naturally described in terms of smaller versions of themselves:

- **A set of Russian dolls** is either a single solid doll, or a doll with a smaller set of Russian dolls inside it.
- **A folder** contains files, and possibly other folders, which contain files and possibly other folders...
- **Your ancestors** are your parents, and your parents' ancestors.
- **n factorial**, written n!, is the product 1 × 2 × ... × n. But it can also be defined like this:

```
0! = 1
n! = n × (n - 1)!     for n of 1 or more
```

So 4! = 4 × 3! = 4 × 3 × 2! = 4 × 3 × 2 × 1! = 4 × 3 × 2 × 1 × 0! = 24.

Every one of these definitions has two parts:

- a **base case**, the simplest version, answered directly with no reference to itself (the solid doll; 0! = 1);
- a **recursive case**, which describes the thing in terms of a **smaller** version of itself (a doll containing smaller dolls; n! = n × (n - 1)!).

The recursive case must always move **towards** the base case. Otherwise the definition never finishes.

## Your first recursive functions

A recursive definition translates into Python almost word for word:

Here's another: the sum of the numbers from 1 to n is n plus the sum from 1 to n - 1, and the sum from 1 to 0 is 0.

And powers: x to the power n is x times x to the power n - 1, and anything to the power 0 is 1.

The shape is always the same:

```
def solve(problem):
    if problem is the simplest case:
        return the answer directly
    smaller_answer = solve(a smaller problem)
    return the answer, built from smaller_answer
```

## Watching it run

How can `factorial(4)` work, when it needs `factorial(3)` before it can finish? Python keeps track of every call that's waiting. Each call gets its own **frame**, with its own `n`, stacked on top of the call that's waiting for it. Step through it:

Two phases to notice:

1. **Going down**: `factorial(4)` calls `factorial(3)`, which calls `factorial(2)`, and so on. None of them can finish yet; they're all waiting, with their own `n`, on the stack.
2. **Coming back up**: `factorial(0)` returns 1 straight away. Then `factorial(1)` can finish (1 × 1), then `factorial(2)` (2 × 1), and so on back to the first call, which returns 24.

Each call's `n` is separate. When `factorial(1)` is running, the `n` in `factorial(4)`'s frame is still 4, waiting.

## The leap of faith

You **can** follow a recursive function all the way down and back up, and the figure does exactly that. But to **write** one, don't. Instead:

1. Handle the base case directly.
2. For the recursive case, **assume the recursive call already works** for the smaller problem. Don't think about how. Just ask: if I had the answer for the smaller problem, how would I build the answer for this one?

For `factorial(n)`: if I already had `factorial(n - 1)`, I'd just multiply it by `n`. Done. That's the **leap of faith**: trusting the smaller call, the way you trust `len` or `sorted` without tracing them.

It's not really faith. It's induction: if the base case is right, and each step correctly builds on a smaller right answer, then every answer is right.

The reasoning: a number under 10 has one digit (base case). Otherwise, chopping off the last digit (`n // 10`) gives a smaller number; **if I knew how many digits that has**, the answer is one more.
