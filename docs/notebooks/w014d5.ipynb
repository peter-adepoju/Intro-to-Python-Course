```python
def is_palindrome(text):
    left, right = 0, len(text) - 1
    while left < right:
        if text[left] != text[right]:
            return False
        left += 1
        right -= 1
    return True

for t in ["level", "racecar", "python", "", "a", "abba"]:
    print(repr(t), is_palindrome(t))
```

```python
def pair_with_sum(sorted_values, target):
    left, right = 0, len(sorted_values) - 1
    while left < right:
        total = sorted_values[left] + sorted_values[right]
        if total == target:
            return sorted_values[left], sorted_values[right]
        if total < target:
            left += 1
        else:
            right -= 1
    return None

print(pair_with_sum([1, 3, 4, 6, 8, 11], 10))
print(pair_with_sum([1, 3, 4, 6, 8, 11], 13))
```

```python
def best_window(values, k):
    """The largest total of k neighbouring values, and where that window starts."""
    if not 1 <= k <= len(values):
        raise ValueError("k must be between 1 and the number of values")
    total = sum(values[:k])
    best, best_start = total, 0
    for i in range(k, len(values)):
        total += values[i] - values[i - k]
        if total > best:
            best, best_start = total, i - k + 1
    return best, best_start

sales = [4, 2, 12, 3, 8, 1, 7, 9, 2, 5]
print(best_window(sales, 3))
```

```python
values = [4, 2, 12, 3, 8, 1, 7]
prefix = [0]
for v in values:
    prefix.append(prefix[-1] + v)
print(prefix)

def range_sum(i, j):
    """The sum of values[i] to values[j], including both."""
    return prefix[j + 1] - prefix[i]

print(range_sum(2, 4), sum(values[2:5]))
print(range_sum(0, 6), sum(values))
```

### Bug hunt: The window that drifts

`best_window_sum(values, k)` should return the largest sum of `k` neighbouring values. Its sliding update is wrong. Find the bug by desk-checking `[1, 2, 3, 4]` with `k = 2` (the answer is 7).

```python
def best_window_sum(values, k):
    total = sum(values[:k])
    best = total
    for i in range(k, len(values)):
        total += values[i] - values[i - k + 1]
        best = max(best, total)
    return best
```

## The patterns you know

Here's your library so far, each with its skeleton. Every one of them has appeared in earlier weeks.

| Pattern | Use it when the problem says... | Skeleton |
|---|---|---|
| **Accumulate** | total, product, join all | `total = 0` / `for x in xs: total += x` |
| **Count** | how many | `n = 0` / `for x in xs: if cond: n += 1` |
| **Find first** | the first..., where is..., does it contain | `for i, x in enumerate(xs): if cond: return i` / `return -1` |
| **Best so far** | largest, smallest, longest, the most... | `best = xs[0]` / `for x in xs[1:]: if better(x, best): best = x` |
| **Filter** | only the ones that... | `[x for x in xs if cond]` |
| **Transform** | each one, changed | `[f(x) for x in xs]` |
| **Any / all** | is there any..., are they all... | `any(cond for x in xs)` |
| **Frequency map** | how many of each, most common | `counts[x] = counts.get(x, 0) + 1` |
| **Group** | for each X, the Ys | `groups.setdefault(key(x), []).append(x)` |
| **Seen-set** | duplicates, first repeat, unique | `if x in seen: ...` / `seen.add(x)` |
| **Neighbours** | each with the next, changes, rises | `for a, b in zip(xs, xs[1:]):` |
| **Stack** | matching, nesting, undo, most recent | `stack.append(x)` / `stack.pop()` |

`next(generator, default)` gives the first value a generator produces, or the default if there are none: a compact "find first".

## Two pointers

Some problems on a sequence are easiest with **two positions** moving towards each other from the ends. The classic is checking a palindrome without making a reversed copy:

The real power shows on **sorted** lists. To find two numbers in a sorted list that add up to a target: start at both ends. If the sum is too small, the only way to make it bigger is to move the left pointer right; if it's too big, move the right one left.

Each step moves one pointer, so the whole list is examined in one pass, instead of trying every pair. (You'll measure how much that matters in Week 17.)

## Sliding windows

Many problems ask about every run of `k` **neighbouring** items: the best 3-day total, the busiest hour, the longest stretch without a repeat. The slow way adds up each window from scratch. The **sliding window** way keeps the current window's total and updates it as the window slides: add the item coming in, subtract the item going out.

Windows can also **stretch and shrink**: to find the longest run of different characters, grow the window on the right, and shrink it from the left whenever a repeat gets in. You'll do that in the daily challenge.

## Prefix sums

Suppose you'll be asked many questions like "what's the total from position 3 to position 7?". Adding up each range is slow if there are many questions. Instead, work out once the **running totals**, where `prefix[i]` is the sum of the first `i` items. Then any range total is one subtraction:

`prefix` has one more item than `values`, starting with 0, so that the empty range and the range from the start need no special cases.

## Choosing a pattern

| The problem mentions... | Try |
|---|---|
| each item compared with the next | neighbours |
| a pair, in a sorted list, or the two ends | two pointers |
| every run of k neighbouring items, or the longest stretch with some property | sliding window |
| many questions about totals of ranges | prefix sums |
| matching, nesting, "most recent" | stack |
| "have I seen this before?" | seen-set |
| "how many of each?" | frequency map |
