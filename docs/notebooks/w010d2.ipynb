> **Key idea.** `get(key, default)` **reads** without changing the dictionary. `setdefault(key, default)` **adds** the default when the key is missing. `pop(key, default)` **removes** the key if it's there.

## Warm-up

`get` with a default, and `items` with unpacking: you'll use both constantly. Today adds the methods that make other common jobs one line.

## Removing and filling in

### pop

`d.pop(key)` removes an entry and **returns its value**. Like `del`, it raises `KeyError` if the key isn't there, unless you give a default:

`d.popitem()` removes and returns the **last** entry added, as a `(key, value)` tuple. It's handy for using a dictionary like a stack.

`d.clear()` removes everything.

### setdefault

`d.setdefault(key, default)` returns the value for `key` if it's there. If it isn't, it **adds** `key` with the value `default` first, then returns that:

That line reads "get the list for this key, making an empty one if there isn't one yet, and append to it". You'll use it on Day 3 for grouping.

## Merging

`a.update(b)` copies every entry of `b` into `a`, changing `a` in place. Where both have the same key, **`b`'s value wins**:

From Python 3.9, `a | b` makes a **new** merged dictionary and leaves both alone, and `a |= b` is the same as `a.update(b)`. Again, the right-hand side wins:

This is the usual way to combine **defaults** with someone's own choices: start with the defaults, merge their choices over the top.

### fromkeys, and its trap

`dict.fromkeys(keys, value)` gives every key the **same** value. That's fine for numbers and strings:

## Sorting and inverting

A dictionary keeps the order its entries were added; it has no `sort` method. To go through it in another order, sort its **items**:

The first sorts by key (tuples compare by their first item). The second sorts by value, highest first: each item `kv` is a `(key, value)` tuple, so `kv[1]` is the value.

`dict(sorted(...))` turns the sorted pairs back into a dictionary in that order, if you need one.

To find the key with the largest value, `max(marks, key=marks.get)` reads: of the keys, the one whose `marks.get(key)` is largest.

### Inverting

**Inverting** a dictionary swaps keys and values, so you can look up in the other direction:

That only works cleanly when the values are all different and hashable. If two keys share a value, the later one silently wins. You'll handle that properly with grouping tomorrow.

## Views and copies

`keys()`, `values()` and `items()` don't make copies. They return **views**: live windows onto the dictionary that change when it does:

Views support `len`, `in` and looping. `keys()` and `items()` also support set operations, which you'll meet on Day 5. Wrap a view in `list(...)` when you need a fixed snapshot, or need to index it.

Assignment never copies a dictionary, exactly as with lists:

`d.copy()` (or `dict(d)`) is a **shallow** copy: a new dictionary holding the same values. If the values are lists, both dictionaries share those lists, just like a shallow copy of a list of lists.
