````python
import random

def main():
    """Run a three-question times-table quiz."""
    greet()
    score = ask_questions(3)
    show_result(score, 3)

def greet():
    """Explain the quiz to the player."""
    print("Times tables! Answer three questions.")

def ask_questions(count):
    """Ask count questions; return how many were right."""
    right = 0
    for _ in range(count):
        if ask_one():
            right += 1
    return right

def ask_one():
    """Ask one random question; return True if answered correctly."""
    a = random.randint(2, 12)
    b = random.randint(2, 12)
    answer = int(input(f"{a} x {b} = "))
    if answer == a * b:
        print("Correct!")
        return True
    print(f"No, it's {a * b}.")
    return False

def show_result(score, out_of):
    """Print the score and a comment."""
    print(f"You scored {score}/{out_of}.")
    if score == out_of:
        print("Perfect!")
    elif score >= out_of / 2:
        print("Well done.")
    else:
        print("Keep practising.")

if __name__ == "__main__":
    main()
````

## Warm-up

`a` was written as if `b` and `c` already existed. You can read `a` and understand it before knowing how `b` and `c` work. That's the idea behind today: write the top first, trusting that the pieces below will be filled in.

## Big steps first

Suppose you're asked for a program that runs a times-table quiz: it asks ten random questions, tells the player whether each answer is right, and gives a score at the end with a comment.

Don't start by writing `import random`. Start by saying, in a few lines, what the program does:

1. Greet the player.
2. Ask ten questions, counting the right answers.
3. Show the score and a comment.

Each of those is a function. Step 2 is still big, so split it again: asking **one** question means making up two numbers, asking for an answer, and checking it. Keep splitting until each piece is something you could write in a few lines without thinking hard.

This is called **top-down design**, or **stepwise refinement**: you refine the big steps into smaller ones, level by level. You can draw it as a **structure chart**, with each function above the functions it calls:

## Stubs: running an unfinished program

You don't have to write every function before you can run the program. A **stub** is a function with its name, parameters and docstring, but a placeholder body: `pass`, or a simple `return` of a value of the right type.

It runs! The output is placeholders, but it proves that the **structure** works: every function is called with the right arguments, and every return value goes to the right place. Now you fill in one stub at a time, running the program after each, and the rest of the program keeps working around it.

Why work this way?

- **You never face a blank page.** There's always a next stub to fill in, and each one is small.
- **Mistakes show up early.** A badly designed step is obvious when you write `main`, before you've spent an hour on its details.
- **Each function can be tested alone.** When a stub becomes real, you check just that one function.

## main() and the __name__ guard

The top of the structure chart is usually a function called `main`. It's the program's summary, readable in a few seconds. The last line of the file calls it.

You'll often see that last line written like this:

```python
if __name__ == "__main__":
    main()
```

Python sets the special variable `__name__` to `"__main__"` when a file is run directly. When the same file is **imported** by another program as a module (you'll do this in Week 15), `__name__` is the module's name instead, so `main()` doesn't run, and the other program can use the file's functions without starting the quiz. ## Good pieces

How do you know you've split a problem well? Good functions have these qualities:

- **One job.** You can describe it in one sentence without saying "and". `ask_one` asks one question. A function that "reads the file and calculates the averages and prints the report" should be three.
- **A clear name.** A verb for functions that do something (`show_result`), a question for ones that answer yes or no (`is_finished`).
- **Short.** If it doesn't fit on your screen, it's probably doing more than one job.
- **Talks only through its parameters and return value.** No hidden global variables.
