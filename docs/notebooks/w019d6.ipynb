```python
def merge(a, b):
    out, i, j = [], 0, 0
    while i < len(a) and j < len(b):
        if b[j] < a[i]:
            out.append(b[j]); j += 1
        else:
            out.append(a[i]); i += 1
    return out + a[i:] + b[j:]

print(merge([2, 7, 9], [1, 7, 8]))
```

> **Cheap checks, big savings.** Each check in `smart_sort` costs one pass over the data, O(n), and can save far more: counting instead of comparing, or a handful of merges instead of a full sort. ## Forge: a smart sort

No single sort is best for every input. A tiny list is fastest with insertion sort. Whole numbers in a small range can be counted. Data that's already mostly in order can be merged run by run. Everything else wants a good general sort. Real libraries make choices like these all the time.

## Forge bonus: bucket sort

Counting sort needs whole numbers in a small range. **Bucket sort** handles decimal numbers spread fairly evenly over a known range, such as random numbers from 0 to 1. Make n buckets, each covering an equal slice of the range; drop each number into its bucket; sort each bucket (they're small, so insertion sort is ideal); and join the buckets in order.

If the numbers are spread evenly, each bucket gets about one number, and the whole sort is **O(n) on average**. If they all land in one bucket, it's as slow as insertion sort on everything: O(n²).
