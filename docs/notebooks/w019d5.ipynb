```python
from operator import itemgetter

orders = [("A17", "Leeds", 42.5), ("B03", "Bath", 18.0), ("C22", "Leeds", 18.0)]
print(sorted(orders, key=itemgetter(2)))          # by price
print(sorted(orders, key=itemgetter(1, 2)))       # by town, then price

people = [{"name": "Ola", "age": 31}, {"name": "Bo", "age": 25}]
print(sorted(people, key=itemgetter("age")))
```

```python
from operator import itemgetter

stock = {"tea": 12, "jam": 3, "bread": 30, "eggs": 3}
print(sorted(stock.items(), key=itemgetter(1), reverse=True))    # most first
print(sorted(stock, key=stock.get))                               # names, by quantity
print(dict(sorted(stock.items())))                                # a new dict in key order
```

```python
SIZES = {"XS": 0, "S": 1, "M": 2, "L": 3, "XL": 4, "XXL": 5}
stock = ["L", "S", "XXL", "M", "S", "XS", "XL"]
print(sorted(stock, key=SIZES.get))

DAYS = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
RANK = {day: i for i, day in enumerate(DAYS)}
shifts = [("Fri", "Ada"), ("Mon", "Ben"), ("Sun", "Cy"), ("Mon", "Dee")]
print(sorted(shifts, key=lambda s: RANK[s[0]]))
```

```python
import unicodedata

def plain(text):
    """Lower case, with accents removed: 'Émile' becomes 'emile'."""
    decomposed = unicodedata.normalize("NFD", text.casefold())
    return "".join(ch for ch in decomposed if not unicodedata.combining(ch))

names = ["Zoë", "émile", "Adebayo", "Ëkpo", "chidi"]
print(sorted(names))
print(sorted(names, key=plain))
```

```python
import re

def natural_key(text):
    parts = re.split(r"(\d+)", text.casefold())
    return [int(p) if p.isdigit() else p for p in parts]

files = ["file10.txt", "file2.txt", "File1.txt", "file1.txt", "chapter 3", "chapter 12"]
print(natural_key("file10.txt"))
print(sorted(files, key=natural_key))
```

```python
scores = [("Ada", 71), ("Ben", None), ("Cy", 85), ("Dee", None), ("Eve", 64)]
print(sorted(scores, key=lambda s: (s[1] is None, s[1] or 0)))
```

```python
runners = [("Ola", "F", 31.2), ("Ben", "M", 29.8), ("Ada", "F", 29.8), ("Cy", "M", 31.2), ("Zara", "F", 29.8)]
# fastest first; ties by name Z to A (strings can't be negated, so two passes)
step1 = sorted(runners, key=lambda r: r[0], reverse=True)
step2 = sorted(step1, key=lambda r: r[2])
print([r[0] for r in step2])
```

```python
words = ["kiwi", "banana", "fig", "cherry"]
print(max(words, key=len), min(words, key=len))
stock = {"tea": 12, "jam": 3, "bread": 30}
print(max(stock, key=stock.get))
```

```python
import heapq

scores = {"Ada": 91, "Ben": 78, "Cy": 85, "Dee": 99, "Eve": 70, "Fin": 88}
print(heapq.nlargest(3, scores, key=scores.get))
```

```python
from functools import cmp_to_key

def compare(a, b):
    ab, ba = a + b, b + a
    return (ab < ba) - (ab > ba)       # put a first if ab is the bigger number

nums = [3, 30, 34, 5, 9]
print("".join(sorted(map(str, nums), key=cmp_to_key(compare))))
```

## Warm-up

Python sorts strings by the numbers behind their characters: every capital letter comes before every lower-case letter, and digits are compared one character at a time, so `"10"` comes before `"2"`. Correct by the rules, and not what anyone wants. By the end of today you'll be able to sort these as `File1.txt, file1.txt, file2.txt, file10.txt`.

## Key functions

`sorted(items, key=f)` works out `f(item)` **once** for each item, sorts by those keys, and gives back the items themselves. You met this in Week 9; here's the full toolkit.

**Built-in functions as keys:** `len`, `abs`, `str.lower`, `str.casefold`, and any function you write. Pass the function itself, without brackets:

**Records:** `operator.itemgetter(i)` makes a function that returns item `i` (or key `i` of a dictionary). With several arguments it returns a tuple, which gives several levels at once:

`itemgetter(2)` does the same job as `lambda r: r[2]`, a little faster, and reads well. Use whichever is clearer.

**Sorting a dictionary:** a dictionary has no order to sort in place, but its items can be sorted into a list:

`key=stock.get` works because `stock.get` is a function from a name to its quantity.

## Custom orders

Some things have an order that isn't alphabetical or numeric: clothing sizes, days of the week, priority levels. Write the order down once, as a dictionary from each value to its **rank**, and use the rank as the key:

Build the rank dictionary from a list with `enumerate`, so the order lives in one place. Values missing from the dictionary need a decision: `SIZES.get` returns `None` for an unknown size, which can't be compared with numbers and crashes. A default such as `lambda s: SIZES.get(s, 99)` puts unknown values last.

## Sorting text the way people expect

**Case:** `key=str.lower` ignores capitals. `str.casefold` is a stronger version for other languages: the German "ß" casefolds to "ss", so "Straße" and "STRASSE" compare equal.

**Accents:** `"é"` has a much higher character number than `"z"`, so `"émile"` sorts after `"zoe"`. For proper language-aware sorting, the `locale` module or the `PyICU` library is the real answer; a simple version strips accents with `unicodedata`:

**Numbers inside text ("natural sort"):** split the text into runs of digits and non-digits, and turn the digit runs into numbers. Then `"file10"` becomes `["file", 10]` and `"file2"` becomes `["file", 2]`, and lists compare item by item:

`re.split` with the pattern in brackets keeps the digit runs in the result (you'll learn regular expressions properly in Week 23). Because the split always alternates text, number, text, the parts at matching positions are always the same type, so they compare safely.

## Missing values and many levels

**Missing values:** `None` can't be compared with numbers or strings, so a list with gaps crashes `sorted`. Put a flag first in the key that says whether the value is missing: `False` sorts before `True`, so present values come first.

**Many levels in mixed directions:** negate numbers to reverse them inside one tuple key. For strings, which can't be negated, use repeated stable sorts (Week 18), least important level first:

**Dates as text:** a date written `"2026-10-31"` (year, month, day, with leading zeros) sorts correctly as plain text, which is one reason that format is the international standard (ISO 8601). `"31/10/2026"` doesn't: turn it into a tuple `(year, month, day)` or a `date` object for the key.

## Beyond sorted

**`min`, `max` take a key too**, and they're O(n) where sorting is O(n log n):

**The top k:** `heapq.nlargest(k, items, key=...)` and `heapq.nsmallest` find the k biggest or smallest in about O(n log k), better than sorting everything when k is small:

**When no key works:** occasionally the order depends on comparing two items **together**. The classic example: arrange numbers to make the biggest possible number when written side by side. Is `9` before `34`? Compare `"934"` with `"349"`. That's a comparison, not a key, so use `cmp_to_key`:

Prefer a key when you can: it's called n times, while a comparison function is called about n log n times.
