## Warm-up

You've been using functions since your first lesson: `print`, `len`, `input`, `int`, `range`. Someone else wrote them, gave them names, and you call them. Today you'll write your own.

## Why functions?

Last week's Goldbach challenge needed the prime test twice, so you wrote the same seven lines twice. That's a problem in three ways:

- **It's long.** The program is twice as hard to read as it needs to be.
- **It's fragile.** If you find a bug in the prime test, you must remember to fix it in both places.
- **It hides the idea.** The program is about "is p prime and is q prime", but you have to read fourteen lines to see that.

A **function** solves all three. You write the code once, give it a name, and from then on the name stands for the code.

## def: defining a function

The definition has four parts:

```python
def greet():
    print("Hello!")
```

1. The keyword `def`, short for *define*.
2. The function's **name**, following the same rules as variable names. Use a verb or a verb phrase: `greet`, `print_menu`, `check_answer`.
3. Brackets `()` and a colon. (The brackets will soon hold parameters.)
4. An indented **body**: the code the function runs.

Then, to run the body, you **call** the function by writing its name followed by brackets: `greet()`. Each call runs the whole body once.

## Defining is not running

This trips everyone up at first, so slow down. Predict the output before you run it:

Python reads the program top to bottom as always. When it reaches `def show():`, it doesn't run the body: it just remembers it under the name `show`. Only when it reaches `show()` does it jump into the body, run it, and then jump **back** to where the call was and carry on.

Press X-ray on this one and step slowly. Watch three things:

- At the `def` line, a new name `show` appears, pointing at a **function** object.
- At each call, a new box appears in the names panel: a **frame** for the function, where its own names will live.
- When the body finishes, the frame disappears and the yellow line returns to just after the call.

### Define before you call

Because Python runs top to bottom, a function must be defined **before** the line that calls it runs:

`NameError: name 'hello' is not defined`: at the moment of the call, Python hasn't reached the `def` yet. The usual layout is: all the function definitions first, then the main program that calls them.

### Forgetting the brackets

`greet` on its own is just the name of the function, like a label on a box you haven't opened. Without `()`, nothing runs. Printing it shows what the label points at: a function object. (You'll do clever things with that in Week 7.)

## A first parameter

A function that does exactly the same thing every time is useful, but limited. A **parameter** is a name in the brackets that receives a value each time the function is called:

When you call `greet("Ada")`, Python makes the parameter `name` point at `"Ada"` for the duration of that call. The value in the call, `"Ada"`, is called an **argument**. The next call gives `name` a different value. In X-ray, watch `name` appear inside each new frame.

Tomorrow is all about parameters: several of them, defaults, and naming arguments in the call.

## Functions calling functions

A function's body can call other functions, including ones you wrote. This is how big programs are built: small functions combined into bigger ones.

Step through it in X-ray. When `triangle` calls `stars`, a second frame appears **on top of** the first. Python keeps a stack of frames, one for each call in progress. When `stars` finishes, its frame is removed and Python returns to exactly where it was in `triangle`. This stack of calls is called the **call stack**, and you'll meet it again in error messages, which show you the stack at the moment something went wrong.
