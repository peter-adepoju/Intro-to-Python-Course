## Warm-up

`sorted(x)` gave you a new list and left `x` alone. `x.sort()` changed `x` itself. Today's question for every method: which of those two kinds is it?

Here's the whole toolbox. Run any operation on the list, and keep chaining:

## Adding items

You know `append`, which adds **one** item at the end. Two more:

- `extend(other)` adds **every item** of another sequence to the end.
- `insert(i, value)` puts a value **at position** `i`, shifting the rest right.

### append or extend?

The difference catches everyone once:

`append` adds its argument as **one** item, even if that item is a list, so `a` ends up with a list inside it. `extend` adds the argument's **items** one by one. Use `extend` (or `+=`) to join one list onto another.

## Removing items

- `pop()` removes the **last** item and **returns** it. `pop(i)` removes and returns the item at position `i`. Popping from an empty list is an `IndexError`.
- `remove(value)` removes the **first** item equal to `value`. If there isn't one, it's a `ValueError`.
- `clear()` empties the list.

`pop` is the only one of these that gives you back what it removed, which makes it perfect for taking things out of a list one at a time.

## Finding items

- `index(value)` gives the position of the **first** match. An optional second argument says where to start looking. If the value isn't there: `ValueError`.
- `count(value)` gives how many items equal `value`, and is 0 when there are none.

`index` on a list is like `.index` on a string. Unlike strings, lists have no `find` that returns -1, so check with `in` first when the value might be missing.

## Ordering: sort and reverse

`x.sort()` sorts the list **in place**. It takes the same `key=` and `reverse=` as `sorted`:

`reverse()` turns the list round in place. (`x[::-1]` gives a reversed **copy**.)

### Changes the list, or returns an answer?

| Changes the list, returns `None` | Leaves the list alone, returns an answer |
|---|---|
| `append`, `extend`, `insert` | `index`, `count` |
| `remove`, `clear` | `len(x)`, `sum(x)`, `min(x)`, `max(x)` |
| `sort`, `reverse` | `sorted(x)`, `x[::-1]`, `x[a:b]`, `x + y` |
| `pop` (also returns the item) | `x.copy()` (a new list with the same items) |

Python's rule: methods that change a list return `None`, so that you can't mistake them for ones that give you something new.

### Speed: the ends are not equal

`append` and `pop()` work at the end of the list and are fast however long the list is. `insert(0, ...)` and `pop(0)` work at the front, so every other item has to shift along one place: on a list of a million items, that's a million moves each time. For a few hundred items you'll never notice. You'll learn faster structures for the front, and why, in Phase 4.
