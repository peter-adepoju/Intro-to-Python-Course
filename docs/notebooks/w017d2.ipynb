> **Key idea.** Big-O: keep the fastest-growing term of the step count and drop its constant. O(f(n)) promises that, for large n, the steps grow no faster than f(n).

````python
import time

def common_list(a, b):
    return [x for x in a if x in b]          # `in b` is O(len(b)) each time

def common_set(a, b):
    lookup = set(b)
    return [x for x in a if x in lookup]     # O(1) each time

a = list(range(0, 20000, 2))
b = list(range(0, 20000, 3))
for f in (common_list, common_set):
    start = time.perf_counter()
    result = f(a, b)
    print(f.__name__, len(result), f"{time.perf_counter() - start:.4f} s")
````

````python
def summary(items):
    n = len(items)                 # O(1)
    total = sum(items)             # O(n)
    biggest = max(items)           # O(n)
    ordered = sorted(items)        # O(n log n)
    middle = ordered[n // 2]       # O(1)
    return total, biggest, middle

print(summary([5, 1, 9, 3, 7]))
````

````python
def dedupe(items):
    result = []
    for x in items:                # runs n times
        if x not in result:        # O(len(result)), up to O(n)
            result.append(x)       # O(1)
    return result

print(dedupe([3, 1, 3, 2, 1, 4]))
````

> **Common mistake.** Big-O isn't a measure of time. "O(1)" doesn't mean "fast": a dictionary lookup is O(1), and so is a function that always does a billion steps. It means the cost **doesn't grow** with n.

`unique_words` is correct, but O(n²) because of one hidden cost: on a book of 100,000 words it takes minutes. Make it O(n) while still returning the words in the order they first appear.

````python
def unique_words(words):
    result = []
    for w in words:
        if w not in result:
            result.append(w)
    return result

print(unique_words("the cat and the hat and the bat".split()))
````

## Warm-up

The last column is the exact count divided by n². As n grows it settles at 3: for big n, the `50n + 200` part hardly matters. The count behaves like "3 times n²", and the 3 depends on how you count steps. What's left is the **shape**: n².

## Keep the biggest term

To find the Big-O of a step count:

1. Keep only the **fastest-growing** term: n² beats n, n beats log n, log n beats a constant.
2. Drop its **constant factor**.

| Exact count | Biggest term | Big-O |
|---|---|---|
| 3n + 5 | 3n | O(n) |
| n(n − 1)/2 = n²/2 − n/2 | n²/2 | O(n²) |
| 7 | 7 | O(1) |
| 2 log₂ n + 10 | 2 log₂ n | O(log n) |
| 4n² + 100n log n + n | 4n² | O(n²) |
| 2ⁿ + n³ | 2ⁿ | O(2ⁿ) |

Read O(n²) as "big O of n squared", or "order n squared". O(1) means the steps don't grow with n at all: **constant time**.

Why drop constants? Because they depend on details that change with the language, the computer, and how you choose to count, while the shape doesn't. An O(n) function and an O(n²) function can be compared on any computer, and for large enough n the O(n) one will win every time.

The base of a logarithm doesn't matter either: log₂ n and log₁₀ n differ only by a constant factor (about 3.3), so both are just O(log n).

## What Big-O promises

Saying a function is O(f(n)) is a promise: **for large enough n, its steps are at most a constant times f(n)**. It's an upper bound on growth.

From n = 54 on, `steps(n)` stays below `4n²` for ever, and that's all the promise needs. The early sizes where it doesn't hold don't count: Big-O only cares about what happens as n grows without limit.

Two consequences:

- Big-O tells you about **growth**, not speed at a particular size. It won't tell you whether a function takes 1 ms or 10 ms for n = 50.
- It's normally used for the **worst case**: a search that sometimes stops early is still O(n), because sometimes it doesn't.

Strictly, a function that is O(n) is also O(n²) (n² is a bigger upper bound), but that's an unhelpful thing to say. In practice, give the **tightest** Big-O you can.

## Hidden costs in Python

A single line of Python isn't always a single step. Many built-in operations loop over their data internally, so their cost grows with its size:

| Operation | Big-O | Why |
|---|---|---|
| `items[i]`, `items[i] = x`, `len(items)` | O(1) | a list knows where each item is, and its length |
| `items.append(x)`, `items.pop()` | O(1) | at the end, nothing moves |
| `items.insert(0, x)`, `items.pop(0)` | O(n) | every other item shifts along |
| `x in items`, `items.index(x)`, `items.count(x)` | O(n) | checks the items one by one |
| `items[a:b]`, `items.copy()`, `list(items)` | O(length copied) | builds a new list |
| `sum(items)`, `max(items)`, `min(items)` | O(n) | looks at every item |
| `sorted(items)`, `items.sort()` | O(n log n) | the sort (Week 19) |
| `x in s`, `s.add(x)` for a set; `d[k]`, `k in d` for a dict | O(1) on average | hashing (Week 10) |
| `text + other`, `text.replace(...)`, `"".join(parts)` | O(length of the result) | builds a new string |

The one that catches people most is `in` on a **list** inside a loop:

One line, one loop, and yet `common_list` is O(n × m): the comprehension loops over `a`, and `x in b` loops over `b` every time. `common_set` is O(n + m).

## Reading Big-O off code

To find the Big-O of a function:

1. Work out the cost of each line, **including hidden costs**.
2. A loop costs (the number of times it runs) × (the cost of its body).
3. Steps one after another **add**; keep the biggest.

O(1) + O(n) + O(n) + O(n log n) + O(1): the biggest is **O(n log n)**, from the sort.

The loop runs n times and its body can cost up to O(n), so this is **O(n²)**. With a `seen` set beside the list, the body becomes O(1) and the whole thing O(n).

## Small inputs

Big-O describes what happens for **large** n. For small n, the constants you dropped can decide which is quicker:

Below n = 100, the O(n²) version does less work. Above it, the O(n) version wins, and wins by more and more. This is why Python's own sort uses a simple O(n²) method for short stretches of a list and a cleverer one for the whole (you'll see this in Week 19). For most programs, though, the advice is simple: when inputs might be large, pick the better Big-O.
