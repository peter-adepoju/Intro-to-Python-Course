> **Why the fewest options first?.** Try `solve` on the hard puzzle choosing the **first** empty square each time instead of the one with the fewest options, and count the calls. Choosing the most constrained square means that a square with only one option is filled without any guessing, and a square with none ends a hopeless path straight away, before any more guesses are piled on top of it. People solving Sudoku with a pencil do the same thing. This "most constrained first" rule speeds up many backtracking searches, from timetables to puzzles.

## Forge: a Sudoku solver

A Sudoku is a 9 by 9 grid split into nine 3 by 3 boxes. Some squares hold digits; the rest are empty. Fill every empty square with a digit from 1 to 9 so that every **row**, every **column** and every **box** contains each digit exactly once.

Here is a puzzle written as 9 lines of text, with `.` for an empty square:

```
53..7....
6..195...
.98....6.
8...6...3
4..8.3..1
7...2...6
.6....28.
...419..5
....8..79
```

A solver can work like a careful person with a pencil and an eraser: find an empty square, try each digit that doesn't clash with its row, column or box, and for each, try to solve the rest. If the rest can't be solved, rub the digit out and try the next. If no digit works, this square is a dead end, so report failure and let the call above try its next digit. That's backtracking.

In the board, empty squares are stored as `0`, so a board is a list of 9 lists of 9 ints.

## Forge bonus: splitting words

Website addresses, hashtags and old telegrams run words together: `"thisisagoodday"`. Splitting them back into words is a branching search: the first word could be `"this"` or `"thi"` (if that were a word) or `"t"`, and each choice leaves a shorter string to split. Without memoisation, a long string with many ways to split can take for ever; with it, each starting position is worked out once.
