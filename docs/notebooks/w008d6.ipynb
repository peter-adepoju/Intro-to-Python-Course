### Exercise: Forge bonus: the Fisher-Yates shuffle

How do you shuffle a list so that every possible order is **equally likely**? Many people's first idea (swap every item with a random position anywhere) is subtly unfair. The correct method, the **Fisher-Yates shuffle**, works backwards:

```
for i from the last position down to 1:
    pick a random j from 0 to i (including i)
    swap the items at i and j
```

1. Write `shuffle(items)` that shuffles the list **in place** this way, using `random.randint(0, i)` for each `j`, and returns `None`.
2. Write `shuffled(items)` that returns a shuffled **copy** and leaves the original alone, by copying and then calling `shuffle`.
