## Forge: Wordle

In **Wordle**, you guess a five-letter word. Each letter of your guess is marked **green** if it's in the answer in that position, **yellow** if it's in the answer but somewhere else, and **grey** if it isn't in the answer. That sounds simple until letters repeat. Try these by hand before reading on (we'll write green as `G`, yellow as `Y`, grey as `.`):

| Answer | Guess | Feedback |
|---|---|---|
| `crane` | `react` | `YYGY.` |
| `crane` | `nanny` | `.Y.G.` |
| `abbey` | `babes` | `YYGG.` |
| `speed` | `erase` | `Y..YY` |

Look at `nanny`: the answer has one `n`, and it's matched green by the fourth letter. So the first and third `n`s are grey, not yellow: a letter can only be marked as many times as it appears in the answer, and greens claim their letters **first**. That's the rule people get wrong.
