> **Common mistake.** If the string is empty, the body never runs and the loop variable is never created. Using it after the loop then raises a `NameError`. Don't rely on the loop variable after a loop unless you know the sequence wasn't empty.

````python
word = input("Word: ")
count = 0
for ch in word:
    if ch == "s":
        count += 1
print(count)
````

````python
text = input("Text: ")
letters = digits = spaces = other = 0
for ch in text:
    if ch.isalpha():
        letters += 1
    elif ch.isdigit():
        digits += 1
    elif ch.isspace():
        spaces += 1
    else:
        other += 1
print(f"{letters} letters, {digits} digits, {spaces} spaces, {other} other")
````

## Warm-up

Four lines of machinery to visit four letters: set up `i`, check it, use it, move it on. Get any of them wrong and you have an off-by-one bug or an infinite loop. Today that machinery disappears.

## The for loop

Read it aloud: "**for** each **letter in word**, print it in capitals". Python does all the work:

1. It takes the first character of `word` and makes the name `letter` point at it.
2. It runs the body.
3. It moves `letter` on to the next character and runs the body again.
4. When there are no characters left, the loop ends.

There's no condition to write and no counter to update, so there's nothing to get wrong. Press X-ray and step through: watch `letter` point at `'l'`, then `'o'`, then `'o'`, then `'p'`.

The body can be as long as you like and can contain anything, just like a `while` body:

The name after `for` is up to you. `ch`, `c`, `letter` and `char` are all common for characters. Choose one that says what it holds.

## The loop variable

The **loop variable** is an ordinary name. Each pass, Python points it at the next item. It's a label, exactly like the ones from Week 1, Day 3.

That explains something that surprises people:

`ch = ch.upper()` moves the label `ch` to a new string. The string `word` is untouched (and strings can't be changed anyway). To get a changed string, **build a new one**, as you did with accumulators last week.

After the loop, the loop variable still points at the last item:

They all work in `for` loops, exactly as in `while` loops:

The `else` on a `for` loop runs if the loop reached the end of the sequence without a `break`, which is perfect for searches.

## for or while?

| Use `for` when... | Use `while` when... |
|---|---|
| you're going through every item of a sequence | you don't know in advance how many passes you need |
| you know the loop must finish | the loop waits for something: valid input, a sentinel, a condition |
| | you need to move through a sequence in an unusual way |

A `for` loop over a string can't run forever, because the string has an end. That alone makes it the safer choice whenever it fits. Searching for the first prime, reading until the user types `done` and Euclid's algorithm are still `while` jobs.
