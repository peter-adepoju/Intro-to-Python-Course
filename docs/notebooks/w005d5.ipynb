```python
text = input("Message: ")
shift = int(input("Shift: "))
result = ""
for ch in text:
    if "a" <= ch <= "z":
        result += chr((ord(ch) - ord("a") + shift) % 26 + ord("a"))
    elif "A" <= ch <= "Z":
        result += chr((ord(ch) - ord("A") + shift) % 26 + ord("A"))
    else:
        result += ch
print(result)
```

> **Key idea.** Shift a letter with `chr((ord(ch) - base + shift) % 26 + base)`, where `base` is `ord("a")` or `ord("A")`. The `% 26` wraps the end of the alphabet round to the start.

```python
secret = "Wkh txlfn eurzq ira"
for shift in range(1, 26):
    plain = ""
    for ch in secret:
        if ch.isalpha():
            base = ord("a") if ch.islower() else ord("A")
            plain += chr((ord(ch) - base - shift) % 26 + base)
        else:
            plain += ch
    print(f"{shift:2}: {plain}")
```

```python
secret = "Mjqqt, ymnx nx f xjhwjy rjxxflj ymfy jsix ns ymj jfwqd jajsnsl."
best = ""
best_n = 0
for letter in "abcdefghijklmnopqrstuvwxyz":
    n = secret.lower().count(letter)
    if n > best_n:
        best, best_n = letter, n
shift = (ord(best) - ord("e")) % 26
print("Commonest letter:", best, "so the shift is probably", shift)
```

## Warm-up

You've used `ord` since Week 2, to see why capitals sort before lower-case letters. Today `ord` and its partner `chr` become tools you'll use to change text.

## Characters are numbers

`ord(ch)` gives a character's **code point**, the number that stands for it. `chr(n)` goes the other way:

Because the letters have consecutive codes, arithmetic on codes moves through the alphabet. `chr(ord("A") + 1)` is `"B"`. That one idea powers today's cipher.

A letter's **position in the alphabet** (0 for a, 25 for z) is its code minus the code of `a`:

### ASCII

The first 128 code points form **ASCII**, a standard from 1963: the English letters, digits, punctuation, and invisible control characters such as the new line (10) and the tab (9).

| Range | Characters |
|---|---|
| 0 to 31 | control characters (tab is 9, new line is 10) |
| 32 | space |
| 48 to 57 | `0` to `9` |
| 65 to 90 | `A` to `Z` |
| 97 to 122 | `a` to `z` |

Notice that each lower-case letter is exactly 32 more than its capital. That's how the earliest computers changed case: by flipping one bit.

## Unicode: every character in every language

ASCII has no room for `é`, `ọ`, `ñ`, `π`, Arabic, Chinese or emoji. **Unicode** fixes this by giving a number to every character in every writing system, more than 150,000 so far, while keeping ASCII's first 128 exactly as they were. Python strings are Unicode, so all of them just work:

You can write any character by its number with an **escape**: `\u` followed by four hex digits, or `\U` followed by eight. You can also use its official name:

### Case is more complicated than it looks

In ASCII, every letter has exactly one capital. In the world's languages it isn't that simple:

The German `ß` becomes `SS` in capitals. For comparing text without caring about case, `.casefold()` is the more thorough version of `.lower()`.

### UTF-8: storing characters as bytes

Code points are numbers, but files and networks store **bytes**, each holding a number from 0 to 255. **UTF-8** is the standard way to turn code points into bytes. ASCII characters take one byte each, so plain English text is unchanged. Other characters take two, three or four:

`.encode()` turns a string into bytes, and `.decode()` turns bytes back into a string. Nearly every web page and file you'll ever read is UTF-8. When you see garbled text like `cafÃ©`, it's almost always UTF-8 bytes being read with the wrong encoding.

## The Caesar cipher

Over 2,000 years ago, Julius Caesar protected military messages by shifting every letter three places along the alphabet: `A` became `D`, `B` became `E`, and at the end it wrapped round, so `X` became `A`. Play with the shift:

To shift one lower-case letter in Python:

1. Turn it into its position: `p = ord(ch) - ord("a")`, from 0 to 25.
2. Add the shift, wrapping round with `% 26`: `(p + shift) % 26`.
3. Turn it back into a letter: `chr(... + ord("a"))`.

The `% 26` is what makes the alphabet wrap: `x` is position 23, plus 3 is 26, and 26 % 26 is 0, which is `a`. Remember from Week 1 that `%` with a positive divisor never gives a negative answer, so the same line also works for **negative** shifts.

Now a whole message, keeping capitals as capitals and leaving everything else alone:

To **decrypt**, shift the other way: by `-shift`, or equivalently by `26 - shift`.

## Breaking it

A cipher is only as good as its resistance to being broken, and Caesar's is weak. There are only 25 possible shifts.

### By brute force

Try them all, and read which one makes sense:

Scan the list: shift 3 is the only one that reads as English. Last week's brute-force idea, applied to code-breaking.

### By letter frequencies

In English text, `e` is by far the most common letter. A shift doesn't change **how often** letters appear, only which letter stands for which. So the most common letter in a long secret message probably stands for `e`, and the distance between them is the shift. This idea, **frequency analysis**, was first written down by the Arab scholar al-Kindi in the ninth century, and it breaks every simple substitution cipher.

Frequency analysis needs a reasonably long message. On three words it can easily guess wrong, which is why brute force is still handy.
