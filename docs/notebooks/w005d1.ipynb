## What a method is

A **function** like `len` or `print` stands on its own: you hand it a value in brackets. A **method** belongs to a value. You write the value, a dot, and the method's name:

```python
len(word)         # a function: word goes inside the brackets
word.upper()      # a method: upper belongs to word
```

Every string has the same set of methods, because they come with the `str` type. You've already met a few: `.lower()`, `.upper()`, `.isdigit()`, `.isalpha()`. Today you'll meet the rest of the everyday ones. Python has about 47; nobody memorises them all. You learn the dozen you use most and look up the others.

Play with the methods here. Pick one, change the string or the arguments, and watch what comes back:

## Changing case

`.lower()` is the one you'll use most: comparing text without caring about capitals, as you did in Week 2.

## Trimming: strip

Text typed by people, or read from files, often has stray spaces at the ends. `.strip()` removes whitespace (spaces, tabs and new lines) from **both ends**, but never from the middle:

`lstrip` trims only the left end and `rstrip` only the right. Give `strip` an argument to remove other characters from the ends instead:

The second one surprises people. The argument is a **set of characters** to remove, not a word: `strip` keeps peeling off any of `w`, `.`, `m`, `o` and `c` from both ends until it meets a character that isn't in the set.

## Searching: find, index, count

- `.find(sub)` gives the position where `sub` **first** starts, or `-1` if it isn't there. An optional second argument says where to start looking.
- `.rfind(sub)` is the same but searches from the **right**: the last occurrence.
- `.count(sub)` counts how many times `sub` appears, without overlapping.
- `.index(sub)` is like `find`, but raises a `ValueError` if `sub` isn't there:

Use `find` when "not there" is normal and you'll check for `-1`. Use `index` when "not there" would be a bug that you want to hear about loudly.

## Replacing

`.replace(old, new)` gives a copy with every `old` replaced by `new`. An optional third argument limits how many are replaced:

Replacing with `""` deletes, which is handy for removing every space or every comma.

## Starts and ends with

These are clearer and safer than slicing: `filename[-4:] == ".pdf"` breaks if you change the extension to one of a different length.

## Testing what a string contains

These methods all return `True` or `False`, and all return `False` for an empty string:

| Method | True when every character is... |
|---|---|
| `.isalpha()` | a letter |
| `.isdigit()` | a digit 0 to 9 |
| `.isalnum()` | a letter or a digit |
| `.isspace()` | whitespace |
| `.isupper()` | upper case (and there's at least one letter) |
| `.islower()` | lower case (and there's at least one letter) |

`repr(s)` shows a string with its quotes, so you can see empty strings and spaces.

## Methods return new strings

Strings can **never** be changed. Every method that seems to change a string actually builds and returns a new one. If you don't store the result, it's lost:

`name = name.upper()` points the label `name` at the new string. The old one is thrown away.
