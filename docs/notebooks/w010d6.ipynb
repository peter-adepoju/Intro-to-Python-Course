> **One fact, one place.** Notice that the library never stores how many copies are **available**. It works that out from `copies` and `loans` whenever it's asked. If it stored the number as well, every borrow and return would have to update two things, and one day a bug would update only one of them. Storing each fact in exactly one place, and working everything else out from it, is one of the most useful rules in data design. Real databases are built around it.

### Exercise: Forge bonus: a spelling corrector

An **edit** of a word is one of:

- a **delete**: remove one letter (`"speling"` → `"seling"`);
- a **swap**: swap two neighbouring letters (`"teh"` → `"the"`);
- a **replace**: change one letter to another letter `a` to `z`;
- an **insert**: add one letter `a` to `z` anywhere, including at the ends.

## Forge bonus: a spelling corrector

This bonus is based on a famous short essay by Peter Norvig, "How to Write a Spelling Corrector". The idea: most typing mistakes are one small **edit** away from the word that was meant, and among the real words one edit away, the most likely is the most common one.

Train it on a whole book and it corrects most single typos. Norvig's full version also looks two edits away, and you could add that now: the edits of every edit.
