## Questions with yes or no answers

So far every program you've written does the same thing every time it runs. Real programs react: a cash machine checks your PIN, a game checks whether you've run out of lives, a form checks whether a password is long enough. Each check is a question whose answer is *yes* or *no*.

Python's word for yes is `True` and its word for no is `False`. They are values, like `7` or `"hi"`, and they have their own type:

The type is called `bool`, short for **Boolean**, after George Boole, a nineteenth-century mathematician who worked out the algebra of true and false. There are exactly two `bool` values, and nothing else is a `bool`.

## Asking questions: comparison operators

You rarely type `True` or `False` yourself. Usually you get them by **comparing** two values:

Each comparison is an expression, just like `5 + 3`. The difference is that its value is a `bool`. There are six comparison operators:

| Operator | Asks | `5 ? 3` | `3 ? 3` |
|---|---|---|---|
| `==` | equal to? | `False` | `True` |
| `!=` | not equal to? | `True` | `False` |
| `<` | less than? | `False` | `False` |
| `<=` | less than or equal to? | `False` | `True` |
| `>` | greater than? | `True` | `False` |
| `>=` | greater than or equal to? | `True` | `True` |

Play with them. Type any two values, or press one of the examples underneath:

### = and == are different

This is the most common slip in all of programming, so meet it now:

- `=` **assigns**: "make this name point at that value". It's an instruction.
- `==` **compares**: "are these two values equal?". It's a question, and its answer is `True` or `False`.

Reading them aloud helps: say `x = 4` as "x *is now* 4" and `x == 4` as "is x *equal to* 4?".

`print(x = 4)` looks like a question but uses `=`, so Python reads it as an instruction it doesn't understand there, and raises a `TypeError`.

### Comparisons come after arithmetic

In an expression with both, Python does all the arithmetic first and compares at the end. So `2 + 3 == 5` means `(2 + 3) == 5`. Step through it:

### Chaining comparisons

To ask "is `x` between 0 and 100?", maths writes `0 ≤ x ≤ 100`. Python lets you write the same thing:

`0 <= x <= 100` means "`0 <= x` **and** `x <= 100`": both must be true. Most languages don't allow this, but Python does, and it reads beautifully.

## Comparing strings

`==` works on strings too, and it's exact: every character must match, including capitals and spaces.

But what does it mean for one string to be **less than** another? Python compares them the way a dictionary orders words: character by character, from the left. The first pair of characters that differ decides the answer.

To compare two characters, Python uses their **code numbers**. Every character has one, and `ord()` tells you what it is:

So `"a" < "b"` because 97 < 98. And because every capital letter (65 to 90) has a smaller number than every lower-case letter (97 to 122), `"Zebra" < "apple"` is `True`. Try these in the comparison lab above: `'apple'` and `'apricot'`, then `'Zebra'` and `'apple'`.

The last one: when one string runs out and everything so far matched, the shorter one counts as smaller. That's also how dictionaries order "app" before "apple".

### Ignoring capitals

To compare two strings without caring about capitals, turn both into lower case first. `"Hello".lower()` gives `"hello"`. The dot means `lower` belongs to the string; you'll meet many of these **methods** in Week 5.

### Is one string inside another?

`in` asks whether one string appears somewhere inside another. Its answer is a `bool` too.

`not in` asks the opposite: `"x" not in "banana"` is `True`.

## Truthiness: turning values into bools

`bool()` converts any value to `True` or `False`, the same way `int()` converts to a whole number. The rule is short. These count as **false**:

- `0` and `0.0`
- the empty string `""`
- `False` itself

So far, everything else counts as **true**. (You'll meet a few more false values later, like empty lists.)
