### Exercise: Forge: the receipt maker

Ask for an order with the prompt `Order: `. An order is a list of items separated by `;`. Each item looks like `name x quantity @ price`, with any amount of space around the parts, for example:

```
Rice (5kg) x 2 @ 8500; bread x1@950 ;  Palm oil x 3 @ 1200.50
```

Print a receipt like this:

```
Item          Qty      Price      Total
---------------------------------------
Rice (5Kg)      2   8,500.00  17,000.00
Bread           1     950.00     950.00
Palm Oil        3   1,200.50   3,601.50
---------------------------------------
Subtotal                      21,551.50
VAT 7.5%                       1,616.36
TOTAL                         23,167.86
```

The rules:

- Item names are stripped and written in title case.
- The item column is as wide as the longest item name or the word `Item`, whichever is longer. Then a space, then `Qty` right-aligned in 4, `Price` and `Total` right-aligned in 11, all money with thousands separators and 2 decimal places.
- The dashed lines are as wide as the whole table.
- The `Subtotal`, `VAT 7.5%` and `TOTAL` labels are left-aligned in a column as wide as the table minus 11, then the amount right-aligned in 11.
- Empty items (for example from a `;` at the very end) are skipped.
- If any item can't be read (missing `x` or `@`, or a quantity that isn't a whole number), print `Can't read: ` and that item exactly as it appeared, stripped, and print **no** receipt.

### Exercise: Forge bonus: Morse code

Morse code writes each letter as dots and dashes. Here are the codes for `a` to `z`, in order, in one string:

```python
MORSE = ".- -... -.-. -.. . ..-. --. .... .. .--- -.- .-.. -- -. --- .--. --.- .-. ... - ..- ...- .-- -..- -.-- --.."
```

`MORSE.split()` turns it into a lookup table: piece number 0 is the code for `a`, piece number 1 is for `b`, and so on, so the code for a letter is `codes[ord(letter) - ord("a")]`.

Ask for a mode with the prompt `Mode (to/from): `, then a message with the prompt `Message: `.

- **to**: translate the message into Morse. Letters are separated by single spaces and words by ` / `. Ignore anything that isn't a letter or a space, and ignore case. `Hi there!` becomes `.... .. / - .... . .-. .`.
- **from**: translate Morse back into lower-case text. Codes are separated by spaces, and `/` separates words. An unknown code becomes `?`.

````python
MORSE = ".- -... -.-. -.. . ..-. --. .... .. .--- -.- .-.. -- -. --- .--. --.- .-. ... - ..- ...- .-- -..- -.-- --.."
codes = MORSE.split()
````
