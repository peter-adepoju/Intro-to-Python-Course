````python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
````

## Asking the user: input()

`input()` pauses the program, waits for the user to type something and press Enter, and gives back what they typed. Put a question (a **prompt**) in the brackets so the user knows what to type.

When you press **Run**, the console waits for you: type your answer at the cursor and press Enter. Now watch what happens in slow motion. This is the most important line pattern of the week:

### input() always gives a string

Whatever the user types, even `42`, `input()` gives back a **string**. To do arithmetic, convert it:

That crashes with a `TypeError` (you can't add a string and an int). Convert with `int()` or `float()`:

Python works from the inside out: `input(...)` runs first and gives a string; `int(...)` turns it into a number; then `age` points at that number.

What if the user types something that isn't a number? `int("hello")` raises a `ValueError`. For now, that's fine. In Week 3 you'll learn to ask again, and in Week 11 to catch the error.

## print() in depth

You've used `print()` with several items separated by commas. Two extra settings, **`sep`** and **`end`**, control what goes *between* items and what goes *at the end*:

- `sep` is put between items. The default is one space, `" "`.
- `end` is put after the last item. The default is a new line, `"\n"`, which is why each `print()` starts a new line.

## f-strings

An **f-string** is a string with an `f` before the quote. Inside it, anything in curly brackets `{ }` is worked out as Python and inserted into the text.

Compare with what you've done before:

```python
print(name, "is", age, "years old.")          # commas: spaces added for you
print(name + " is " + str(age) + " years old.")  # + : you manage spaces and str()
print(f"{name} is {age} years old.")           # f-string: clearest of all
```

f-strings are what professional Python programmers use almost all the time. Inside the brackets you can put any expression:

(Don't worry about that last line's details: it shows that even complicated expressions work inside `{ }`. You'll understand every part of it by Week 2.)

### Formatting numbers

After a colon inside the brackets, you can add a **format spec** saying exactly how to show the value. Play with all the settings:

The ones you'll use most:

| f-string | Result | Meaning |
|---|---|---|
| `f"{3.14159:.2f}"` | `3.14` | 2 decimal places |
| `f"{1234567:,}"` | `1,234,567` | thousands separators |
| `f"{0.256:.1%}"` | `25.6%` | as a percentage |
| `f"{'hi':>6}"` | `    hi` | right-align in 6 characters |
| `f"{'hi':<6}"` | `hi    ` | left-align in 6 |
| `f"{'hi':^6}"` | `  hi  ` | centre in 6 |
| `f"{7:03}"` | `007` | pad with zeros to width 3 |
