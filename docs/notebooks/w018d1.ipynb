```python
def last_index(items, target):
    for i in range(len(items) - 1, -1, -1):
        if items[i] == target:
            return i
    return -1

print(last_index([3, 8, 3, 5, 3, 9], 3))
```

```python
def find_first(items, test):
    for i, item in enumerate(items):
        if test(item):
            return i
    return -1

words = ["kiwi", "fig", "banana", "plum", "cherry"]
print(find_first(words, lambda w: len(w) > 5))
print(find_first(words, lambda w: w.startswith("p")))
```

```python
def find_all(text, sub):
    found = []
    i = text.find(sub)
    while i != -1:
        found.append(i)
        i = text.find(sub, i + 1)
    return found

print(find_all("the cat sat on the mat with the hat", "the"))
```

```python
books = [
    {"title": "Dune", "year": 1965, "copies": 2},
    {"title": "Emma", "year": 1815, "copies": 0},
    {"title": "Kindred", "year": 1979, "copies": 1},
]

def find_book(books, title):
    for book in books:
        if book["title"].lower() == title.lower():
            return book
    return None

print(find_book(books, "emma"))
print(find_book(books, "Ulysses"))
print(next((b["title"] for b in books if b["copies"] == 0), "all in stock"))
```

> **Key idea.** Linear search: check each item in turn, return as soon as you find a match, and report "not found" only after checking every item. O(n) worst case, O(1) extra space, and it works on any data.

## Warm-up

`in` answers yes or no; `index` and `find` say **where**, and both give the **first** match. They differ when there's no match: `find` returns -1, while `index` raises an error. You'll write your own version of each today.

## Searching item by item

**Linear search** (or sequential search) looks at each item in turn, from the start, until it finds what it's looking for or runs out of items:

Two details matter:

- **Return as soon as you find it.** There's no point looking further, and stopping early is what gives the best case of 1 comparison.
- **Only decide "not found" after the loop.** A common mistake is an `else: return -1` inside the loop, which gives up after the first item.

`enumerate` gives the position and the item together, which reads more naturally:

Why return -1? It can't be a real position, so it's a clear signal. Some code returns `None` instead; either is fine if you're consistent and the caller checks. (In Python, -1 is a valid **index**, the last item, so a careless `items[result]` won't crash; it'll quietly give the wrong item. Always check the result first.)

## Python's built-in searches

Python has linear searches built in. Each is O(n) in the worst case, because each one checks the items one at a time:

| Expression | Gives | When nothing matches |
|---|---|---|
| `x in items` | `True` or `False` | `False` |
| `items.index(x)` | the first position | raises `ValueError` |
| `text.find(sub)` | the first position | `-1` |
| `text.index(sub)` | the first position | raises `ValueError` |
| `items.count(x)` | how many matches (always checks every item) | `0` |
| `any(test for x in items)` | `True` if any item passes | `False` |
| `next((x for x in items if test), default)` | the first item that passes | `default` |

`next(...)` with a generator is the neat way to say "the first item that passes this test". The second argument is what it returns if none does; without it, `next` raises `StopIteration`. You'll learn exactly how generators work in Phase 4; for now, use the pattern.

## Searching records

Real data is usually records: a list of dictionaries or tuples. Search by one **field** of each record:

If the same list is searched by title many times, though, a linear search each time is wasteful: build a dictionary from title to record once (O(n)), and every lookup after that is O(1). That's the most common way to speed up search in practice.

## When linear search is right

| Case | Comparisons | Big-O |
|---|---|---|
| Best: the target is first | 1 | O(1) |
| Worst: the target is last or missing | n | O(n) |
| Average: present, anywhere equally likely | (n + 1)/2 | O(n) |

Linear search needs **no preparation**: the data can be in any order, of any type, and it only needs `==` (or your test) to work. That makes it the right choice when:

- the list is small (up to a few thousand items, it's instant);
- you search it only once or a few times;
- the data is unsorted and sorting it would cost more than the searches save;
- the test is complicated ("the first order over £50 shipped to Wales").

When the same large collection is searched many times, sort it and use binary search (tomorrow), or put it in a set or dictionary.
