````python
def total(*numbers):
    result = 0
    for n in numbers:
        result += n
    return result

print(total(1, 2, 3))
print(total(10, 20, 30, 40, 50))
print(total())
````

````python
def greet_all(greeting, *names):
    for name in names:
        print(f"{greeting}, {name}!")

greet_all("Hello", "Ada", "Grace", "Alan")
greet_all("Hi")
````

> **Key idea.** `*name` in a `def` collects every extra positional argument into a tuple called `name`. There can be at most one starred parameter.

````python
def show(**kwargs):
    print(kwargs)

show(colour="red", size=12)
show()
````

````python
def describe(thing, **details):
    print(thing)
    for key, value in details.items():
        print(f"  {key}: {value}")

describe("Lagos", population="15 million", country="Nigeria")
describe("Python", created=1991, creator="Guido van Rossum")
````

````python
def make_tag(name, **attrs):
    text = "<" + name
    for key, value in attrs.items():
        text += f' {key}="{value}"'
    return text + ">"

print(make_tag("a", href="https://example.com", target="_blank"))
print(make_tag("br"))
````

````python
def log(*messages, level="INFO", sep=" "):
    print(f"[{level}]", sep.join(messages))

log("disk", "almost", "full", level="WARN")
log("started")
````

````python
def full_name(first, middle, last):
    return f"{last.upper()}, {first} {middle}"

parts = "Ada Augusta Lovelace".split()
print(full_name(*parts))
````

````python
def largest(*numbers):
    best = numbers[0]
    for n in numbers[1:]:
        if n > best:
            best = n
    return best

pieces = input("Numbers: ").split()
print(largest(*pieces))
````

## Warm-up

The same function took one argument, three, none, and two plus a keyword. Its signature, from yesterday, is `print(*objects, sep=' ', end='\n', ...)`. The `*` in front of `objects` is what makes it accept any number.

## *args: any number of positional arguments

Put a `*` in front of a parameter name, and that parameter collects **all the extra positional arguments** into one value:

`args` is a **tuple**: an ordered, unchangeable sequence, written with round brackets. You'll learn all about tuples in Week 9. For now you only need what you already know from strings and `split` results: `len(args)`, `args[0]`, and `for x in args:` all work.

The name `args` is a convention, short for "arguments". You can choose a better one when the meaning is clearer:

### Required parameters, then the rest

Ordinary parameters can come before the starred one. They take the first arguments, and the star takes whatever's left:

## **kwargs: any number of keyword arguments

Two stars collect **all the extra keyword arguments**:

`kwargs` (short for "keyword arguments") is a **dictionary**: a collection of names, called **keys**, each paired with a value. Dictionaries are Week 10's topic. Today you need just two things:

- `for key, value in kwargs.items():` visits every name and its value;
- `"colour" in kwargs` checks whether a keyword was given, and `kwargs["colour"]` gets its value.

The keywords come out in the order they were given in the call.

That's how a web page's tags are written: a name, then any number of attributes. Functions in real web frameworks build HTML exactly like this.

## The full order of parameters

All the kinds of parameter can appear in one `def`, but only in this order:

```python
def f(pos_only, /, normal, default=1, *args, kw_only, kw_default=2, **kwargs):
```

1. positional-only parameters, then `/`;
2. ordinary parameters (with and without defaults);
3. `*args`, or a bare `*`;
4. keyword-only parameters;
5. `**kwargs`, always last.

Anything after `*args` is automatically keyword-only. That's exactly how `print(*objects, sep=' ', end='\n')` works: `sep` and `end` come after the star, so you must name them.

## Stars in a call: spreading out

The star also works the other way round. In a **call**, `*sequence` spreads the sequence out into separate positional arguments:

`full_name(*parts)` is the same as `full_name("Ada", "Augusta", "Lovelace")`. This is handy for passing on the pieces from a `split`:

That compares the pieces as **strings**. To compare them as numbers, you'd convert each one first, and you'll have a neat way to do that in Week 9.

## When to use the stars

Flexible signatures are powerful, but they have a cost: Python can no longer catch a wrong call. If `area(width, height)` is called with three arguments, Python complains. If `area(*sides)` is called with three, it doesn't.

Use `*args` when the arguments really are "a bunch of the same kind of thing", like numbers to add or names to greet. Use `**kwargs` when the set of options is genuinely open-ended, like attributes on a tag. When a function has a fixed set of inputs, list them by name.
