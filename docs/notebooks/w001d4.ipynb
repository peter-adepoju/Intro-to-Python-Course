`a // b` asks "how many whole `b`s fit into `a`?" and `a % b` asks "what's left over?" Pick some numbers and watch the sharing:

### What % is good for

The remainder is one of the most useful operations in programming. Four patterns you'll use for years:

**1. Even or odd.** A number is even when dividing by 2 leaves nothing: `n % 2` is `0` for even numbers and `1` for odd ones.

**2. Digits.** `n % 10` is the last digit, and `n // 10` removes it.

**3. Wrapping around.** Clocks, days of the week and anything circular use `%`. What time is it 50 hours after 9 o'clock?

`% 24` wraps any number back into the range 0 to 23. Day of the week, 100 days after a Monday (day 0)? `(0 + 100) % 7` is `2`, a Wednesday.

**4. Unit conversion.** Seconds into minutes and seconds, pence into pounds and pence, centimetres into metres and centimetres:

### Negative numbers

Python's `//` always rounds **down**, towards minus infinity, not towards zero. So `-7 // 2` is `-4`, not `-3`. And the remainder always has the **same sign as the divisor**: with a positive `b`, `a % b` is never negative.

Use the figure above with `-17, 5` to see why: Python jumps down in steps of 5 to `-20` (that's 4 jumps down, so `-17 // 5` is `-4`), then the remainder is the distance back up to `-17`: `3`.

This is a deliberate choice, and a useful one: it means `n % 7` is always a valid day of the week, even if `n` is negative. (Some other languages, like C and JavaScript, do it differently. If you ever translate code between languages, watch out.)

## What a float really is

Floats are stored in **binary** (base 2), using 64 bits. Some decimals can't be written exactly in binary, the same way 1/3 can't be written exactly in decimal (0.3333…). So Python stores the nearest value it can.

This isn't a Python bug: every language that uses standard floats behaves the same way. It matters in three situations:

1. **Comparing floats.** Never check whether two floats are exactly equal. Check whether they're *close*. (You'll learn `==` properly next week.)
2. **Money.** Count in whole units (kobo, cents, pence) using ints, or use Python's `decimal` module (Week 20).
3. **Adding many small floats.** Errors add up. Use `math.fsum` for long sums.

That first line, `import math`, is new: it loads Python's **math module**, a toolbox of extra functions. You'll learn exactly how `import` works in Week 20. For now, put `import math` at the top of your program, then use `math.` in front of the tool's name.

Floats also have limits. Ints can be as large as you like, but floats top out around 1.8 × 10³⁰⁸:

`inf` means infinity: the number got too big for a float.

## Rounding

There are four ways to turn a float into a whole number, and they disagree. Drag the slider:

| Function | Rule | `2.7` | `-2.7` | `2.5` |
|---|---|---|---|---|
| `round(x)` | nearest; exact halves go to the **even** number | `3` | `-3` | `2` |
| `int(x)` | chop off the decimals (towards zero) | `2` | `-2` | `2` |
| `math.floor(x)` | always down | `2` | `-3` | `2` |
| `math.ceil(x)` | always up | `3` | `-2` | `3` |

Which do you need? **Think about the problem.** How many buses for 130 people, 50 per bus? You need `math.ceil(130 / 50)` = 3, because rounding down would leave people behind.

`round()` can also round to decimal places: `round(3.14159, 2)` is `3.14`.

Rounding exact halves to the even number ("banker's rounding") avoids a slight upward bias when you round lots of numbers. But beware: because of how floats are stored, `round(2.675, 2)` gives `2.67`. Try 2.675 in the float lab above to see why.

## Converting between types

`int()`, `float()` and `str()` build a new value of that type from another value. Feed different values into the machine:

The rules worth memorising:

- `int("42")` works: the text looks like a whole number. `int(" 42 ")` works too: spaces at the ends are ignored.
- `int("3.7")` fails: `int()` won't read a decimal point from text. Go through float: `int(float("3.7"))` is `3`.
- `int(3.7)` works on a float and **chops** to `3`.
- `float("3.7")` is `3.7`; `float("1e3")` is `1000.0`.
- `str(anything)` always works: it gives the text you'd see if you printed it.

Why does this matter? Because text and numbers don't mix:

`+` between two strings joins them; between two numbers it adds. Between a string and a number it doesn't know what you mean, so it refuses. Convert first:

Both work. The second (commas in `print`) is simpler for printing; the first matters when you're building a string to store. On Day 6 you'll learn the best way of all: f-strings.
