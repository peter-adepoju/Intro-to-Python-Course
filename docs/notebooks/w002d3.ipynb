````python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
````

## Making a choice: if

Here is your first program that doesn't always do the same thing:

Run it with `25`, then again with `15`. With 15, the middle message never appears. Now press **X-ray** and step through both times. Watch the yellow line: with 15, it jumps straight from the `if` line to the last line.

The parts of an `if` statement:

```python
if age >= 18:
    print("You can vote.")
```

1. The word `if`.
2. A **condition**: any expression Python can treat as true or false. Everything you learned this week fits here.
3. A **colon** `:`. It means "here comes the part that depends on this".
4. An **indented block**: one or more lines, each pushed in by 4 spaces. These run only when the condition is true.

The line after the block, `print("Thanks for checking.")`, is back at the left edge. That's how Python knows the block has ended, so it runs whatever the answer was.

## Blocks can hold many lines

Every line indented under the `if` belongs to it:

Try `12000`, then `5000`. With 5000, all three indented lines are skipped together.

## else: one or the other

Often you want one thing when the answer is yes and something else when it's no. That's `else`:

`else` has no condition of its own: it means "if the condition above was false". It lines up with its `if`, ends with a colon, and has its own indented block. Exactly **one** of the two blocks runs, every time.

## Indentation is part of the language

In most languages, indentation is just for looks. In Python it's how you say which lines belong to which block. The standard is **4 spaces** per level; the editors on this site insert them for you when you press Tab or Enter after a colon.
