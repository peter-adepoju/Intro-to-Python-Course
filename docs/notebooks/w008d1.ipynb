> **Key idea.** `items[i]` gives one item. `items[a:b]` gives a **new list** of items from position `a` up to, but not including, `b`. Indexing past the end is an `IndexError`; slicing past the end just stops.

## What a list is

A **list** is an ordered collection of values. You write one with square brackets, the values separated by commas:

A list can hold any kind of value, even a mix (though most useful lists hold one kind). It can be empty. `len` gives the number of items.

The values are called **items** or **elements**. Each one is a separate object, and the list holds a reference to each, like a row of labelled hooks. Press X-ray on this one and look at the objects panel: the list is a row of slots, each with an arrow to its value.
