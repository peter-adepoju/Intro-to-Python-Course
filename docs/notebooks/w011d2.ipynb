> **Common mistake.** Opening the file you're reading with `"w"`, to "update" it, empties it before you've read it. Read everything first, close it, and then write.

```python
def save(scores, name):
    with open(name, "w") as f:
        for person, mark in scores.items():
            f.write(f"{person},{mark}\n")

def load(name):
    scores = {}
    with open(name) as f:
        for line in f:
            person, mark = line.strip().split(",")
            scores[person] = int(mark)
    return scores

save({"Cy": 78, "Di": 55}, "saved.txt")
print(load("saved.txt"))
```

## Modes

The mode decides what happens to a file that already exists:

| Mode | Meaning | If the file exists | If it doesn't |
|---|---|---|---|
| `"r"` | read (the default) | reads it | `FileNotFoundError` |
| `"w"` | write | **empties it first** | creates it |
| `"a"` | append | adds to the end | creates it |
| `"x"` | create | `FileExistsError` | creates it |

`"w"` is the dangerous one: the moment you open a file with it, the old contents are gone, before you've written anything.

## Paths with pathlib

A **path** says where a file is: `marks.txt`, `data/2026/marks.txt`, or `/home/ada/data/marks.txt`. Paths are written differently on different systems (Windows uses `\` between folders, macOS and Linux use `/`), so building them by gluing strings together causes bugs. The `pathlib` module gives you `Path` objects instead:

The `/` operator joins parts of a path, and uses the right separator for the computer it runs on. The parts:

- `name`: the last part, `marks.csv`;
- `stem`: the name without its suffix, `marks`;
- `suffix`: the extension, `.csv`;
- `parent`: the folder it's in, `data/2026`.

A `Path` can be passed to `open` anywhere a string can. And it has shortcuts for small files:

`read_text` and `write_text` open, read or write, and close in one step. They're perfect for small files; for big files, loop over the lines with `open` as before.
