## A string is a sequence of characters

A **string** (`str`) is a sequence of characters in a particular order. Each character sits at a numbered position, called its **index**. Python starts counting at **0**.

Spaces and punctuation are characters too: `"Hi there!"` has 9 characters.

## Writing strings

### Quotes

Single and double quotes make exactly the same kind of string. Choosing one lets you put the other inside without trouble:

**Triple quotes** (`"""` or `'''`) let a string run over several lines:

### Escape sequences

Some characters are hard to type inside a string: a new line, a tab, or the same quote you used to open the string. A backslash `\` starts an **escape sequence**: the backslash and the character after it together stand for one special character.

| You type | You get |
|---|---|
| `\n` | a new line |
| `\t` | a tab |
| `\\` | one backslash |
| `\'` | a single quote |
| `\"` | a double quote |

Type a string literal below and look inside it. Notice that `\n` takes two keystrokes but is stored as **one** character.

A **raw string**, written with an `r` before the quote, switches escapes off. It's handy for Windows paths and, later, regular expressions (Week 23): `r"C:\new\table"` keeps its backslashes.

## Length, joining and repeating

`""` is the **empty string**: a string with no characters at all. It's still a string.

## Indexing: one character

`s[i]` gives the character at index `i`. Remember: the first is at index 0, so the last is at `len(s) - 1`.

Counting from the right uses **negative** indices: `-1` is the last character, `-2` the one before.

## Slicing: cutting out pieces

A **slice** `s[start:stop]` gives a new string from index `start` up to **but not including** `stop`. Use the ruler: set start and stop, and watch which boxes light up.

Why "up to but not including"? Because then `stop - start` is exactly the length of the slice, and `s[:k] + s[k:]` is always the whole string. You'll come to like it.

Leave out `start` to begin at the start; leave out `stop` to go to the end:

### The step

A third number, the **step**, says how far to jump: `s[::2]` takes every second character. A **negative step** walks backwards, so `s[::-1]` is the whole string reversed.
