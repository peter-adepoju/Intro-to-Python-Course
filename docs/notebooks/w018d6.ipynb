````python
def steps(n):
    lo, hi, count = 0, n - 1, 0
    while lo <= hi:
        mid = (lo + hi) // 2
        count += 1
        lo = mid + 1          # pretend the target is always bigger
    return count

print(steps(1), steps(8), steps(1000))
````

## Forge bonus: searching a rotated list

A sorted list has been **rotated**: some number of items were moved from the front to the end, so `[1, 3, 5, 7, 9, 11]` might now be `[7, 9, 11, 1, 3, 5]`. It's no longer sorted, so ordinary binary search fails, and a linear search is O(n). But it's made of **two sorted pieces**, and that's enough for an O(log n) search.

The key fact: if you cut the part you're searching in the middle, **at least one half is properly sorted**. You can tell which by comparing its ends, and for a sorted half you can tell at once whether the target lies inside it.
