### Exercise: Forge: the text analyser

Ask for a paragraph with the prompt `Text: ` (one line) and print this report about it:

```
Characters: 63
Letters: 50
Words: 12
Vowels: 19 (38.0% of letters)
Longest word: paragraph
Most common letter: a (9 times)
Capitalised words: 2
```

The rules:

- **Characters** counts everything, including spaces and punctuation.
- **Letters** counts characters for which `.isalpha()` is `True`.
- **Words** are separated by one or more spaces.
- **Vowels** are a, e, i, o, u in either case. The percentage is of letters, with 1 decimal place. If there are no letters, print `Vowels: 0 (0.0% of letters)`.
- **Longest word** is made only of the letters in a word, so `paragraph,` counts as `paragraph`. If two words tie, keep the first. Print it exactly as written (capitals kept).
- **Most common letter** ignores case and is printed in lower case. On a tie, the letter earliest in the alphabet wins. If there are no letters, print `Most common letter: none`.
- **Capitalised words** are words whose first character is a capital letter.

For the example the text was `A paragraph is a group of sentences. This is a short paragraph.`

````python
text = input("Text: ")
````

````python
# Check: run the exercise cell above first, then this one.
check("def report(text):\n    letters = [c for c in text if c.isalpha()]\n    vowels = [c for c in letters if c.lower() in \"aeiou\"]\n    words = text.split()\n    longest = \"\"\n    for w in words:\n        core = \"\".join(c for c in w if c.isalpha())\n        if len(core) > len(longest):\n            longest = core\n    if letters:\n        best = max(\"abcdefghijklmnopqrstuvwxyz\", key=lambda L: (sum(1 for c in letters if c.lower() == L), -ord(L)))\n        n = sum(1 for c in letters if c.lower() == best)\n        common = f\"{best} ({n} times)\"\n        pct = len(vowels) / len(letters) * 100\n    else:\n        common = \"none\"\n        pct = 0.0\n    caps = sum(1 for w in words if w[0].isupper())\n    return \"\\n\".join([f\"Characters: {len(text)}\", f\"Letters: {len(letters)}\", f\"Words: {len(words)}\",\n                      f\"Vowels: {len(vowels)} ({pct:.1f}% of letters)\", f\"Longest word: {longest}\",\n                      f\"Most common letter: {common}\", f\"Capitalised words: {caps}\"])\ndef check(text):\n    expect_output(\"Text: \" + text + \"\\n\" + report(text), inputs=[text], echo=True)\ndef test_example():\n    \"the example\"\n    check(\"A paragraph is a group of sentences. This is a short paragraph.\")\ndef test_spaces():\n    \"extra spaces\"\n    check(\"  Hello   World  \")\ndef test_ties():\n    \"ties for longest word and most common letter\"\n    check(\"bat cab tab\")\ndef test_no_letters():\n    \"no letters at all\"\n    check(\"123 456!\")\ndef test_capitals():\n    \"capitals and punctuation\"\n    check(\"Python, Java and C. Which is Best?\")")
````

### Exercise: Forge bonus: the calendar
