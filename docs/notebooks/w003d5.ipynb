````python
n = int(input("Number: "))
slow = 0
d = 2
while d < n:
    slow += 1
    if n % d == 0:
        break
    d += 1
fast = 0
d = 2
while d * d <= n:
    fast += 1
    if n % d == 0:
        break
    d += 1
print("Divisions tried, slow test:", slow)
print("Divisions tried, fast test:", fast)
````

> **Key idea.** To test whether `n` is prime, try divisors `d` from 2 while `d * d <= n`. If none divides `n` (and `n > 1`), it's prime.

## Prime numbers

A **prime** is a whole number greater than 1 whose only divisors are 1 and itself: 2, 3, 5, 7, 11, 13... Primes are the building blocks of all whole numbers, and they protect almost every secure connection on the internet.

To test a number, search for a divisor between 2 and `n - 1`. Finding one proves it isn't prime, so `break`. If the search finds nothing, it is:

That's yesterday's `while ... else`: the `else` runs only if no divisor was found. Note the special case: 1 isn't prime, by definition, and neither is 0.

### Doing less work

For a big prime, that test tries every number below it. For 1,000,003 that's a million tries. But it doesn't need to.

Divisors come in **pairs**: if `d` divides `n`, then so does `n // d`. For 36 the pairs are 1 × 36, 2 × 18, 3 × 12, 4 × 9 and 6 × 6. In every pair, one number is **at most the square root** (6 for 36). So if `n` has any divisor besides 1 and itself, it has one no bigger than √n. You only need to search up to there.

You don't even need a square root function: `d * d <= n` says the same thing as `d <= √n`.

A million tries against a thousand. For numbers with 20 digits, the slow test would run for longer than you'll live, and the fast one finishes in seconds. Counting the steps an algorithm takes is how computer scientists compare algorithms, and you'll do it properly in Week 16.

### A loop inside a loop

To list the primes up to 50, run the prime test on every number. That puts one loop **inside** another: for each `n`, the inner loop runs completely.

Notice where `d = 2` is: inside the outer loop, so the search starts again for every new `n`. Next week is all about loops inside loops.

## Euclid's algorithm

The **greatest common divisor** (gcd) of two numbers is the largest number that divides both. gcd(84, 36) is 12. You need it to simplify fractions: 36/84 is 3/7, dividing top and bottom by 12.

You could try every number from the smaller one downwards. Around 300 BC, Euclid of Alexandria wrote down something far faster. It rests on one fact: **gcd(a, b) is the same as gcd(b, a % b)**. Replace the pair with that, again and again, until the second number is 0. Then the first number is the answer.

Three lines of loop, and it's one of the oldest algorithms still in daily use: it runs inside the encryption that secures your web browser. The figure shows why it works. Cut the biggest squares you can from an `a` by `b` rectangle, then do the same with the strip that's left over. The last square fits both sides exactly:

The **lowest common multiple** (lcm) is the smallest number both divide into. It comes straight from the gcd: `lcm(a, b) = a * b // gcd(a, b)`. For 4 and 6: 24 // 2 = 12.
