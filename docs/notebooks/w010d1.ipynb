> **Common mistake.** `{}` is an empty **dictionary**, not an empty set or list. You'll meet sets on Day 5; an empty set is written `set()`.

````python
prices = {"tea": 150, "coffee": 900}
item = "juice"
if item in prices:
    print(prices[item])
else:
    print("not on the menu")
````

````python
prices = {"tea": 150, "coffee": 900}
print(prices.get("tea"))
print(prices.get("juice"))
print(prices.get("juice", 0))
print(prices)
````

> **Tip.** Use `d[key]` when the key **must** be there, so that a mistake shows up as a `KeyError` straight away. Use `d.get(key, default)` when a missing key is normal and you know what it should mean.

## Keys and values

A **dictionary** stores **entries**, each a **key** and the **value** that goes with it. Write it in curly brackets, with a colon between each key and its value:

You look a value up by putting its **key** in square brackets, just as you put an index in square brackets for a list. The difference is that the key can be a name, a word, or anything else suitable, not just a position.

Think of a real dictionary: you look up a word (the key) and find its meaning (the value). Or a phone's contact list: a name and a number.

## Looping

Looping over a dictionary gives you its **keys**, in the order they were added:

Three methods give you views of the parts:

- `d.keys()`: the keys;
- `d.values()`: the values;
- `d.items()`: the entries, as `(key, value)` tuples.

## What can be a key

Values can be anything: numbers, strings, lists, even other dictionaries. Keys are stricter. A key must be **immutable**: strings, numbers, booleans and tuples (of immutable things) work; lists and dictionaries don't.

**Why?** The reason is also why lookups are fast. Python doesn't search through the keys one by one. It turns the key into a number called its **hash**, and uses that number to go straight to the place where the entry is stored, like going straight to the right page in a paper dictionary. If a key could change after it was stored, its hash would change, and Python would look in the wrong place. So keys must be values that can't change: Python calls them **hashable**.

(The numbers themselves change each time Python starts, for security, so we only compare them.)
