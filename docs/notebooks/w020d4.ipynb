> **Common mistake.** `"03/04/2026"` is 3 April in Britain and 4 March in the United States. Never guess: know which layout your data uses, and say so in your code with an explicit pattern. Store and exchange dates in ISO form (`2026-04-03`), which has only one meaning and sorts correctly as text.

````python
from datetime import date

def age_on(born, on):
    """Whole years old on the date `on`."""
    years = on.year - born.year
    if (on.month, on.day) < (born.month, born.day):    # birthday not reached yet this year
        years -= 1
    return years

def days_until_birthday(born, today):
    nxt = born.replace(year=today.year)
    if nxt < today:
        nxt = born.replace(year=today.year + 1)
    return (nxt - today).days

born = date(2000, 11, 3)
today = date(2026, 10, 31)
print(age_on(born, today), days_until_birthday(born, today))
````

## Dates, times and datetimes

The `datetime` module has a few types:

| Type | Holds | Made with |
|---|---|---|
| `date` | a calendar day | `date(2026, 10, 31)` |
| `time` | a time of day | `time(14, 30)` or `time(9, 5, 30)` |
| `datetime` | a day and a time together | `datetime(2026, 10, 31, 14, 30)` |
| `timedelta` | a length of time | `timedelta(days=3, hours=4)` |

The module and one of its types share the name `datetime`, which confuses everyone at first. The usual style is to import the types you need: `from datetime import date, datetime, timedelta`. Then `datetime` means the type.

**Now:** `date.today()` and `datetime.now()` give the current date and time on the computer running the code (in the browser, yours). Their values change every time, so the examples below mostly use fixed dates.

## Arithmetic with timedelta

A `timedelta` is a duration. Add one to a date to move forward; subtract two dates to get the one between them:

- `date - date` and `datetime - datetime` give a `timedelta`. Use `.days`, or `.total_seconds()` for the whole length in seconds.
- `timedelta` takes `weeks`, `days`, `hours`, `minutes`, `seconds` (and smaller); internally it stores only days, seconds and microseconds.
- There's no `timedelta(months=1)`, because months differ in length. To move by months, change the month number yourself (and decide what "one month after 31 January" should mean).

## Text to dates and back

**Dates to text.** `isoformat()` gives the international standard form, `YYYY-MM-DD` (and `T` before the time for a datetime). For any other layout, `strftime` (string **f**ormat time) takes a pattern of codes:

| Code | Means | Example |
|---|---|---|
| `%d` | day, two digits | `05` |
| `%m` | month, two digits | `03` |
| `%Y` | year, four digits | `2026` |
| `%y` | year, two digits | `26` |
| `%B` / `%b` | month name / short name | `October` / `Oct` |
| `%A` / `%a` | weekday / short weekday | `Saturday` / `Sat` |
| `%H:%M` | hours and minutes, 24-hour clock | `14:30` |
| `%I:%M %p` | 12-hour clock with AM/PM | `02:30 PM` |
| `%j` | day of the year | `304` |

**Text to dates.** `date.fromisoformat("2026-10-31")` reads the ISO form. For any other layout, `datetime.strptime(text, pattern)` (string **p**arse time) reads it with the same codes, and raises `ValueError` if the text doesn't match:

## Testable code with dates

A function that calls `date.today()` inside it gives a different answer every day, which makes it impossible to check properly. Pass "today" **in** as an argument, with the real today as the default:

```
def is_overdue(due, today=None):
    today = today or date.today()
    return today > due
```

Your program calls `is_overdue(due)`, and your tests call `is_overdue(due, date(2026, 10, 31))` and always get the same answer. Every exercise today is written that way.
