### Exercise: 5. The whole game

Put it together. The starter has model versions of the four functions. Write `play(secret)`, which takes the secret as a parameter (so the game can be checked with a known secret), asks `Guess: ` until the guess is valid (printing `Four different digits, please.` otherwise; invalid guesses don't count), prints `describe(...)` after each valid guess, and finishes with `Solved in N guesses` (or `1 guess`). The last line of the program is `play(make_secret(4))`.

````python
import random

def valid_guess(text, length):
    text = text.strip()
    return len(text) == length and text.isdigit() and len(set(text)) == length

def score(secret, guess):
    bulls = sum(1 for s, g in zip(secret, guess) if s == g)
    return bulls, len(set(secret) & set(guess)) - bulls

def make_secret(length):
    return "".join(random.sample("0123456789", length))

def describe(bulls, cows):
    if bulls == 4:
        return "Four bulls! You win."
    if bulls == 0 and cows == 0:
        return "Nothing"
    plural = lambda n, w: f"{n} {w}" + ("" if n == 1 else "s")
    return plural(bulls, "bull") + ", " + plural(cows, "cow")
````

### Exercise: Refactor: split the report

The function `report(text)` works, but it does four jobs in one. Split it into four functions that each do one job, and make `report` call them. Keep the output exactly the same. The checker calls each new function on its own:

- `sentences(text)`: the number of sentences (count of `.`, `!` and `?`, at least 1 if there's any text);
- `words(text)`: the list of words, lower-cased, with the punctuation `.,!?;:` stripped from their ends, leaving out any that become empty;
- `average_length(word_list)`: the average word length rounded to 1 place (0 for no words);
- `top_word(word_list)`: the most common word, the alphabetically first on a tie (`""` for no words).

````python
def report(text):
    # count sentences
    n_sentences = sum(text.count(p) for p in ".!?")
    if text.strip() and n_sentences == 0:
        n_sentences = 1
    # make the word list
    word_list = []
    for w in text.lower().split():
        w = w.strip(".,!?;:")
        if w:
            word_list.append(w)
    # average length
    avg = round(sum(len(w) for w in word_list) / len(word_list), 1) if word_list else 0
    # most common word
    counts = {}
    for w in word_list:
        counts[w] = counts.get(w, 0) + 1
    top = min(counts, key=lambda w: (-counts[w], w)) if counts else ""
    return f"{n_sentences} sentences, {len(word_list)} words, average length {avg}, most common: {top}"

print(report("The cat sat. The cat ran! Did the dog?"))
````

## Warm-up

Three functions of one or two lines each, every one obviously right on its own. Put together, they do something you'd have to think hard about if it were written as one block. That's decomposition.

## Splitting a problem

There are two directions to split from, and you'll usually use both.

**Top-down**: start from the whole problem and ask "what are the main steps?". Each step becomes a function, at first just a name, a docstring and a `pass`. Then do the same for each step. You met this in Week 7, and the gradebook project used it.

**Bottom-up**: start from the small, obviously useful pieces you know you'll need (checking a guess is valid, scoring one guess) and build them first, checking each, then combine them.

Here's a top-down plan for **Bulls and Cows**, a code-breaking game older than its board-game version, Mastermind. The computer picks a secret of four different digits. You guess four different digits, and it tells you how many **bulls** (right digit, right place) and **cows** (right digit, wrong place) you got. You keep guessing until you get four bulls.

Each box is a small problem with a precise question: "is this text four different digits?", "how many bulls and cows?". You'll write every one of them in the practice below.

## Good boundaries

Where you draw the lines between functions matters as much as drawing them at all. Good functions have these properties:

- **One job**, which you can describe in a sentence without "and". `score` scores a guess; it doesn't also print the result.
- **Everything it needs comes in as parameters**, and everything it produces goes out with `return`. No reading global variables that something else changes.
- **A small, clear interface**: few parameters, a predictable return value, behaviour at the edges decided and documented.
- **Checkable on its own**: you can call it with an example and check the answer, without setting up the rest of the program.

`report_tangled` can only be checked by running it and typing. `word_stats` can be checked in two lines, for any number of examples, in a fraction of a second.

## Logic in the middle, input and output at the edges

The pattern in that example is worth naming. Keep the **logic** (the deciding, calculating and transforming) in functions that take values and return values. Keep **input and output** (`input`, `print`, files) in a thin layer at the edges of the program that calls them.

```
main (the edge):      read input  ->  call the logic  ->  print the result
logic (the middle):   pure functions: values in, values out
```

The middle is where the bugs are, and it's the part you can check easily. The edges are short and boring, which is exactly what you want of the parts that are hard to check.

## When to split

A function probably needs splitting when:

- you can't describe what it does without "and", or its name is vague (`process`, `handle`, `do_stuff`);
- it's longer than you can see on one screen (roughly 20 to 30 lines);
- it has comments marking sections ("# now work out the totals"): each section wants to be a function, named after the comment;
- the same few lines appear twice, perhaps with small differences;
- part of it would be useful somewhere else.

Splitting a working function without changing what it does is called **refactoring**. Do it in small steps, and check after each one that the program still gives the same results. The checks you wrote earlier make that safe.
