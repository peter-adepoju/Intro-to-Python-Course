### Exercise: Forge: the data cleaner

Write a program, built from these functions, that cleans a payments file.

- `clean_row(row)` takes one row as a dictionary from `DictReader` and returns a new dictionary with:
  - `id`: an int;
  - `date`: the date as text in the form `YYYY-MM-DD`, which must be a real date in 2026;
  - `payer`: stripped, with each word capitalised (`"  bo musa "` gives `"Bo Musa"`);
  - `amount`: a float from 0.01 to 1,000,000, which may contain commas;
  - `method`: one of `card`, `cash`, `transfer`, `cheque`, in lower case.

  If any field is bad, it raises a `ValueError` whose message names the problem: `bad date '01/10/2026'`, `no amount`, `bad amount '-500'`, `bad method 'bitcoin'`, and so on (the exact wording of the messages is up to you, but the message must include the bad value in quotes, except for `no amount`).
- `clean_file(source)` reads the file and returns a tuple `(good, problems)`. `good` is the list of clean rows, in order, **leaving out duplicates**: a row with the same date, payer and amount as an earlier good row. `problems` is a list of `(id, message)` tuples, one per rejected row, with the message `duplicate of N` for a duplicate (where N is the earlier row's id). The `id` in a problem is the text from the file.
- `write_outputs(good, problems)` writes three files:
  - `clean.csv`: the good rows, with the header `id,date,payer,amount,method` and amounts written with 2 decimal places, like `12500.00`;
  - `problems.txt`: one line per problem, `row 3: bad date '01/10/2026'`;
  - `summary.txt`: exactly these lines:

```
Rows read: 12
Accepted: 6
Rejected: 6
Total accepted: 28,550.50
```

  (with the numbers for the file read, and the total of accepted amounts with comma separators and 2 places).
- The main program cleans `payments.csv`, writes the outputs, and prints the contents of `summary.txt`.

> **Never silently drop data.** A cleaner that quietly skipped bad rows would be dangerous: a payment would vanish and nobody would know. Yours rejects bad rows **loudly**, with a reason for each, in a file someone can check and fix. That's the difference between a script and a tool people can trust. Notice too that `clean_row` never prints or decides what to do with a bad row: it raises, and `clean_file` decides. Each function has one job.
