````python
def selection_counting(items):
    comparisons = swaps = 0
    for i in range(len(items) - 1):
        smallest = i
        for j in range(i + 1, len(items)):
            comparisons += 1
            if items[j] < items[smallest]:
                smallest = j
        if smallest != i:
            items[i], items[smallest] = items[smallest], items[i]
            swaps += 1
    return comparisons, swaps

for name, data in [("random", [5, 2, 9, 1, 7, 3, 8, 6]), ("sorted", list(range(8))), ("reversed", list(range(8, 0, -1)))]:
    print(name, selection_counting(data))
````

> **Common mistake.** Selection sort isn't **stable**: equal items can change their order. Sorting `[("b", 2), ("a", 2), ("c", 1)]` by the number swaps `("c", 1)` with `("b", 2)`, sending `("b", 2)` behind `("a", 2)`. Day 5 explains why stability matters.

````python
def smallest_k(items, k):
    items = items[:]                     # don't change the caller's list
    for i in range(min(k, len(items))):
        smallest = i
        for j in range(i + 1, len(items)):
            if items[j] < items[smallest]:
                smallest = j
        items[i], items[smallest] = items[smallest], items[i]
    return items[:k]

print(smallest_k([42, 7, 19, 3, 88, 25, 11], 3))
````

> **Key idea.** Selection sort: for each position from the left, find the smallest item in the rest of the list and swap it there. Always n(n − 1)/2 comparisons, so O(n²) in every case; at most n − 1 swaps; in place; not stable.

## Warm-up

One step of selection sort: the smallest item, 10, swaps places with whatever was first. Now position 0 is final, and the job is to sort the other four. Repeat that until there's nothing left, and the list is sorted.

## Sorting by hand

Give someone a row of cards and ask them to sort it, and many will do this: scan the row for the lowest card, move it to the left end, then scan the rest for the next lowest, and so on. That's **selection sort**: repeatedly **select** the smallest item that's left.

At any moment the list is in two parts:

- on the left, a **sorted** part, holding the smallest items in their final places;
- on the right, the **unsorted** part, everything else in any order.

Each round finds the smallest item in the unsorted part and swaps it to the front of that part, which grows the sorted part by one.

```
[29, 10, 14, 37, 13]     smallest of all is 10: swap with 29
[10 | 29, 14, 37, 13]    smallest of the rest is 13: swap with 29
[10, 13 | 14, 37, 29]    smallest of the rest is 14: already in place
[10, 13, 14 | 37, 29]    smallest of the rest is 29: swap with 37
[10, 13, 14, 29 | 37]    one item left: it must be the largest
```

## Analysis

**Comparisons.** Round 0 compares against n − 1 items, round 1 against n − 2, down to 1 in the last round:

(n − 1) + (n − 2) + ... + 1 = n(n − 1)/2

That doesn't depend on the data at all: the inner loop always scans the whole unsorted part, because it can't know an item is the smallest without looking at all the others. So selection sort is **O(n²) in the best, worst and average case**. Even an already sorted list takes n(n − 1)/2 comparisons.

**Swaps.** At most one per round: at most n − 1 swaps, which is O(n). That's selection sort's one real strength. When moving an item is much more expensive than comparing (large records on slow storage, or physical objects), few moves matter.

**Space.** O(1) extra: a couple of position variables.

For 8 items, always 28 comparisons. For 10,000 items, about 50 million, a few seconds in Python, where `sorted` takes a millisecond or two.

## Variations

**Descending order:** look for the largest instead of the smallest (flip `<` to `>`).

**By a key:** compare `key(items[j]) < key(items[smallest])`, just like `sorted(items, key=...)`.

**The top k only:** if you only want the k smallest items in order, stop after k rounds: k × n comparisons, O(kn), much less than a full sort when k is small.
