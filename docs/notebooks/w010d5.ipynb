````python
import time
n = 10_000
as_list = list(range(n))
as_set = set(as_list)
start = time.perf_counter()
for x in range(0, n, 10):
    x in as_list
middle = time.perf_counter()
for x in range(0, n, 10):
    x in as_set
end = time.perf_counter()
print(f"list: {middle - start:.4f} s")
print(f"set:  {end - middle:.6f} s")
````

> **Common mistake.** Don't use a list for frequent "is it in here?" checks on large data, and don't use a set when the order or the count matters. Turning a list into a set throws away both.

## Warm-up

That loop removes duplicates by checking each name against everything kept so far. It works, but `v not in unique` looks through the whole list each time. With a million visitors, that's very slow. A set does the same job in one step, however big it is.

## What a set is

A **set** is an unordered collection of **unique** elements, written in curly brackets without colons:

The second `"red"` vanished: a set holds each element once. And the order you see may differ from the order you typed. A set has **no order**, so there's no indexing and no slicing: `colours[0]` is a `TypeError`.

`set(...)` makes a set from any sequence, which is the quickest way to remove duplicates:

`sorted` gives a list back, in order, which is the usual way to display a set.

### The empty set

`{}` is an empty **dictionary**. An empty set has to be written `set()`:

### What can be in a set

Elements follow the same rule as dictionary keys: they must be **hashable** (strings, numbers, tuples, not lists or dictionaries). That isn't a coincidence. A set works like a dictionary's keys: Python stores each element at a place worked out from its hash, so checking whether something is there doesn't need a search.

## Changing a set

Sets are mutable. `add` adds one element (nothing happens if it's already there); `update` adds every element of a sequence:

There are two ways to remove an element. `remove` raises a `KeyError` if it isn't there; `discard` quietly does nothing:

`pop` removes and returns an arbitrary element, and `clear` empties the set.

### The "seen" pattern

The most common use of a set in a loop is remembering what you've already met:

`set(visitors)` loses the order; this keeps the first appearance of each, and stays fast because the check is against the set.

## Set operations

The real power of sets is combining them. Suppose `a` and `b` are the students taking two subjects:

- `a | b`, the **union**: students in either (or both);
- `a & b`, the **intersection**: students in both;
- `a - b`, the **difference**: in `a` but not in `b`;
- `a ^ b`, the **symmetric difference**: in exactly one of them.

Each operator makes a **new** set. There are method versions too, `a.union(b)`, `a.intersection(b)`, `a.difference(b)` and `a.symmetric_difference(b)`, which accept any sequence, not just a set:

And there are versions that change a set in place: `a |= b`, `a &= b`, `a -= b`.

## Comparing sets

Two sets are equal when they hold the same elements, whatever order you wrote them in:

`a <= b` asks whether `a` is a **subset** of `b` (every element of `a` is in `b`); `a >= b` asks whether it's a **superset**. `a.isdisjoint(b)` asks whether they have **nothing** in common.

### Frozen sets

A set can't be an element of another set, or a dictionary key, because it's mutable. A `frozenset` is a set that can't be changed, so it can:

The two frozensets are equal (order doesn't matter), so the outer set keeps one. That's a neat way to treat `("Ada", "Bo")` and `("Bo", "Ada")` as the same pair.

## Choosing a collection

You now know the four built-in collections. Choose by what you need to do:

| Collection | Ordered? | Duplicates? | Changeable? | Use it for |
|---|---|---|---|---|
| list `[...]` | yes | yes | yes | a sequence of things, in order |
| tuple `(...)` | yes | yes | no | a fixed group of values, a record, a key |
| dict `{k: v}` | by insertion | keys unique | yes | looking up a value by a key |
| set `{...}` | no | no | yes | membership, removing duplicates, comparing groups |
