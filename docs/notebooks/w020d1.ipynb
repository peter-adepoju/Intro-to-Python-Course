## Warm-up

Three different ways of writing an import, all reaching the same `sqrt`. And `math is m` is `True`: there's only one `math` module object, which both names refer to. By the end of today you'll know why.

## A module is a file

A **module** is just a Python file. The file `helpers.py` is the module `helpers`. Everything defined at the top level of the file (functions, variables, classes) becomes a name inside the module, which you reach with a dot.

This page has a file called `greetings.py`. Open it to see what's inside:

Any program on this page can import it:

`greetings.greet` reads as "the `greet` that lives in `greetings`". The dot keeps names from different modules apart: `greetings.greet` and a `greet` of your own can't clash.

Modules come from three places:

| Kind | Examples | Where from |
|---|---|---|
| Built into the language | `sys`, `math`, `builtins` | part of Python itself |
| The **standard library** | `random`, `datetime`, `json`, `collections`, `pathlib` | installed with every copy of Python |
| **Third-party** packages | `requests`, `numpy`, `pandas` | installed separately with `pip` (Week 22) |
| Your own | `greetings`, `helpers` | files you write |

A **package** is a folder of modules, imported with dots: `os.path`, `email.mime.text`. You'll make one tomorrow.

## Four ways to import

**`import module`**: the whole module, under its own name. The safest and clearest: every use says where the name came from.

```
import random
random.choice(["a", "b"])
```

**`import module as alias`**: the same, under a shorter or clearer name. Some aliases are conventions everyone uses: `import numpy as np`, `import pandas as pd`.

**`from module import name1, name2`**: copy particular names into your program, so you can use them without the dot. Good for a few names you use often:

```
from math import sqrt, pi
from collections import Counter
```

**`from module import *`**: copy **every** public name in. Avoid it in programs: you can't tell where a name came from, and it can silently replace your own names.

You've seen a `from` import go wrong here too: `from math import sqrt` copies the name `sqrt`, but if you later do `sqrt = 5`, only your copy changes; `math.sqrt` is untouched.

## What happens on import

When Python meets `import greetings` for the first time, it:

1. **Finds** the file: it looks through the folders listed in `sys.path`, in order. The first is the folder of the program being run, so your own files are found first.
2. **Runs** the file, top to bottom, in a fresh namespace. Every `def` creates a function, every assignment creates a variable, and any `print` actually prints.
3. **Remembers** the result as a module object in `sys.modules`, a dictionary from module names to modules.
4. **Binds** a name in your program to the module object.

The second time anything imports the same module, step 1 to 3 are skipped: Python finds it in `sys.modules` and hands back the same object. A module's top-level code runs **once** per program, however many times it's imported. That's why `math is m` was `True` in the warm-up.

`noisy` and `n` are two names for the **same** module, so they share `COUNT`.

## Exploring a module

A module is an object like any other. `dir` lists its names, `help` shows its documentation (from docstrings), and a few attributes say where it came from:

Names starting with `_` are private by convention: the module's own business. `from module import *` leaves them out.

For the standard library, `help(module)` works in any Python session, and the official documentation at docs.python.org describes every module, with examples. Learning to read it is part of this week's work.

## Classic mistakes

**Naming your file after a module.** If you save your own experiment as `random.py` and then `import random` in it, Python finds **your** file first (its folder is first in `sys.path`), and `random.randint` doesn't exist. The error is confusing: `AttributeError: module 'random' has no attribute 'randint'`. Never name your files `random.py`, `math.py`, `statistics.py`, `string.py`, `csv.py` and so on.

**Importing a name that isn't there:** `from math import squareroot` raises `ImportError: cannot import name 'squareroot' from 'math'`, which at least tells you exactly what's wrong.

**Expecting an edit to take effect without rerunning:** because modules are imported once, a running program keeps the old version of a module you've edited. Run the program again (or, in an interactive session, restart it).

**Circular imports:** if `a.py` imports `b` and `b.py` imports `a`, one of them runs while the other is only half finished, and names may be missing. The fix is to reorganise: move what both need into a third module.
