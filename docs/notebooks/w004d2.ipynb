## Warm-up

A counting `while` loop: start at 1, keep going while `n <= 5`, add 1 each time. That's three parts in three places. `range` puts all three in one spot.

## range: a sequence of numbers

`range(5)` is a sequence of the numbers 0, 1, 2, 3, 4. A `for` loop walks through it just as it walks through a string:

Five numbers, starting at 0 and stopping **before** 5. That's the most important fact about `range`: **the stop value is never included**.

### Three forms

| You write | You get | Read as |
|---|---|---|
| `range(stop)` | 0, 1, ..., stop − 1 | "stop numbers, from 0" |
| `range(start, stop)` | start, ..., stop − 1 | "from start, up to but not including stop" |
| `range(start, stop, step)` | start, start + step, ... | "from start, in steps of step, before stop" |

### Why stop isn't included

It seems odd at first, but it makes everyday sums tidy:

- `range(n)` has exactly `n` numbers. `range(a, b)` has exactly `b - a`.
- `range(len(s))` gives exactly the valid positions of a string `s`, from 0 to `len(s) - 1`.
- Two ranges join without overlap: `range(0, 5)` then `range(5, 10)` covers 0 to 9 with nothing repeated.

It matches slicing, too: `s[2:5]` stops before position 5, just as `range(2, 5)` stops before 5.
