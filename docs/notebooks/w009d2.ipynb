> **Key idea.** A key that returns `(first, second)` sorts by `first`, and by `second` among equals. Negate a number to sort it in the opposite direction.

```python
results = [["Grace", 88], ["Ada", 93], ["Alan", 88], ["Kemi", 93]]
results.sort(key=lambda r: r[0])
results.sort(key=lambda r: r[1], reverse=True)
for name, score in results:
    print(name, score)
```

> **Tip.** A key that returns a pair does the same job in one sort. Use whichever reads more clearly. Sorting several times is handy when the keys go in different directions and can't simply be negated, such as two text fields, one A to Z and the other Z to A.

```python
files = ["photo10.jpg", "photo9.jpg", "photo100.jpg", "photo1.jpg"]
print(sorted(files))

def number_in(name):
    digits = ""
    for ch in name:
        if ch.isdigit():
            digits += ch
    return int(digits) if digits else 0

print(sorted(files, key=number_in))
```

## Warm-up

The numbers sort by value, but the same digits as **strings** sort letter by letter, so `"100"` comes before `"9"`. And `"Apple"` comes first only because capitals come before lower case. Sorting does exactly what you ask, which isn't always what you mean. Today you'll learn to say what you mean.

## sort, sorted and reverse

A reminder of the two tools:

- `items.sort()` sorts the list **in place** and returns `None`.
- `sorted(items)` returns a **new** sorted list and works on any sequence: lists, strings, ranges, the result of `split`.

Both take `reverse=True` for descending order:

## Key functions

A **key function** turns each item into the value that should be compared. Python works out the key once for each item, sorts by the keys, and gives you back the original items in that order.

## Several keys at once

To sort by one thing, and then by another to break ties, the key can return **a pair of values** in round brackets. Python compares pairs the way it compares words: first items first, and only if those are equal, the second items.

Words are sorted by length, and words of the same length alphabetically. (These bracketed pairs are **tuples**, tomorrow's topic.)

### Mixing ascending and descending

`reverse=True` reverses the whole order. To sort one key **downwards** and another **upwards**, negate the numeric key:

`-r[1]` turns 93 into -93, which sorts before -88, so higher scores come first, while names still sort A to Z.

## Stability

Python's sort is **stable**: items whose keys are equal stay in the order they had before sorting. That gives you a second way to sort by several things. Sort by the **least** important key first, then by the **most** important:

The second sort puts the scores in order, and because it's stable, people with equal scores keep the alphabetical order from the first sort. This is exactly what spreadsheets do when you click one column header and then another.
