### Bug hunt: B7: Three bugs

`top_words(text, n)` should return the `n` most common words of at least 3 letters (ignoring case and the punctuation `.,!?`), as a list of `(word, count)` pairs, most common first, ties in alphabetical order. It has three bugs. Find and fix them.

````python
def top_words(text, n):
    counts = {}
    for word in text.lower().split():
        word = word.strip(".,!?")
        if len(word) > 3:
            counts[word] = counts.get(word, 0) + 1
    pairs = sorted(counts.items(), key=lambda kv: (kv[1], kv[0]))
    return pairs[:n - 1]
````
