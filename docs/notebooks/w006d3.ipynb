> **Tip.** Most functions should **return** their results and let the caller decide what to do with them: print them, store them, test them. Keep `print` for functions whose whole job is showing something, like a menu or a report.

````python
def safe_divide(a, b):
    if b == 0:
        return 0
    return a / b

print(safe_divide(10, 4))
print(safe_divide(10, 0))
````

````python
def is_prime(n):
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True

for k in range(1, 30):
    if is_prime(k):
        print(k, end=" ")
````

````python
def first_vowel(word):
    for i in range(len(word)):
        if word[i] in "aeiou":
            return i
    return -1

print(first_vowel("python"), first_vowel("rhythm"))
````

````python
def is_even(n):
    return n % 2 == 0

def is_leap(year):
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

if is_leap(2024) and is_even(2024):
    print("2024 is an even leap year")
````

## Warm-up

`len` doesn't print anything. It works out a number and **gives it back**, so you can store it, do sums with it, or pass it to another function. Your functions can do the same.

## return: handing a value back

`return width * height` does two things: it works out the value, and it ends the function, handing that value back to wherever the function was called. The call `area(7, 4)` **becomes** that value, just as `len("hello")` becomes 5.

Press X-ray and watch the moment `return` runs: the frame disappears and the call is replaced by the returned value, which is then stored in `a`.

Because calls are expressions, functions combine naturally:

`add(square(3), square(4))` works from the inside out, just like `int(input())`: both `square` calls run first, then `add` runs with their results.

## print is not return

This is the most important idea of the day. `print` and `return` can look as if they do the same thing, because in a small test both make a number appear. They're completely different:

- `print` shows a value **to a person**, on the screen. The program can't use what was printed.
- `return` hands a value **to the program**. Nothing appears on screen unless the program decides to print it.

Here are two functions that look almost identical:

Line by line: `double_print(5)` shows 10. `double_return(5)` shows nothing, because the returned 10 isn't used. Then `a = double_print(5)` shows 10 **again**, and stores... what? The last line reveals it: `None`.

### None

A function without a `return` (or with a bare `return`) gives back a special value, **`None`**, which means "no value". It has its own type, `NoneType`, and it's falsy.

To check for `None`, write `x is None`. (`is` asks whether two names point at the very same object, and there is only ever one `None` object.)

The classic bug: using a printing function as if it returned something.

`add_vat(1000)` prints 1075.0 and returns `None`. Then Python tries `None + None` and raises `TypeError: unsupported operand type(s) for +: 'NoneType' and 'NoneType'`. Whenever you see `NoneType` in an error, look for a function that printed when it should have returned.

## return ends the function

The moment `return` runs, the function is over. Any lines after it are skipped. That makes `return` great for handling special cases first:

A function can have several `return` statements, one per branch:

Notice there's no `elif` and no `else`: once a `return` runs, nothing below it runs, so each `if` only gets checked if all the earlier ones were false.

### return from inside a loop

`return` doesn't just leave the loop like `break`: it leaves the whole function. That makes search functions wonderfully simple. Compare the prime test from Week 3, with its flag and its `while ... else`, with this:

As soon as a divisor is found, the answer is known: `return False`. If the loop finishes without finding one, the last line says so. No flag, no `break`, no `else`.

## Functions that answer yes or no

Functions that return `True` or `False` are called **predicates**, and they're usually named like questions: `is_prime`, `is_even`, `has_digit`, `can_vote`. They read beautifully in conditions:

Notice `return n % 2 == 0`: the comparison already **is** `True` or `False`, so there's no need for `if ... return True else return False`.
