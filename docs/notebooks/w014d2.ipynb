> **Key idea.** Bugs cluster at the edges: empty and single inputs, repeats and ties, zero and negatives, the first and last positions, and exact boundaries. Test there on purpose.

```python
def longest_word(sentence):
    """The longest word in sentence (the first one if several tie).
    Words are separated by spaces; punctuation counts as part of a word.
    Returns "" if there are no words."""
    longest = ""
    for word in sentence.split():
        if len(word) > len(longest):
            longest = word
    return longest

assert longest_word("the quick brown fox") == "quick"
assert longest_word("") == ""
assert longest_word("   ") == ""
assert longest_word("cat dog") == "cat"
assert longest_word("  hi   there ") == "there"
assert longest_word("Hello, world!") == "Hello,"
print("all edge cases pass")
```

```python
def ticket_price(age):
    """Children under 12 pay 500, people 65 or over pay 700, everyone else 1000."""
    if age < 12:
        return 500
    if age >= 65:
        return 700
    return 1000

for age in [11, 12, 13, 64, 65, 66, 0]:
    print(age, ticket_price(age))
```

### Bug hunt: Most common

`most_common(items)` should return the most common item, the one that appears **first** in the list if several tie, or `None` for an empty list. It fails two edge cases. Find them with your own examples, then fix it.

```python
def most_common(items):
    counts = {}
    for x in items:
        counts[x] = counts.get(x, 0) + 1
    best = items[0]
    for x in counts:
        if counts[x] >= counts[best]:
            best = x
    return best
```

### Bug hunt: Words that end the text

`count_word(text, word)` should count how many times `word` appears as a whole word, ignoring case and the punctuation `.,!?`. It misses some. Find the edge it fails on, and fix it.

```python
def count_word(text, word):
    count = 0
    for w in text.lower().split(" "):
        if w.strip(".,!?") == word.lower():
            count += 1
    return count
```

## Warm-up

The function is right for every list you'd think of first. Then one empty list arrives and the whole program stops. That single input is an **edge case**: an input at the edge of what the problem allows, where the usual reasoning stops working.

## Where bugs live

Here's a function that should return the second-largest **different** value in a list, or `None` if there isn't one. Four people wrote it. Every version passes the first two examples, the ones anybody would try first:

Each wrong version has exactly the kind of bug that lives at an edge:

- `sort_only` forgets that values can **repeat**: the second-largest of `[9, 9, 4]` is 4, not 9.
- `start_at_zero` assumes values are **positive**: with all negatives, nothing ever beats the starting 0.
- `no_checks` forgets the lists that are **too short** to have an answer at all.

None of these would be found by `[3, 9, 4]` or `[10, 20, 30]`. Ordinary examples check that the idea is right. Edge cases check that the code is right.

## A checklist

You won't think of edge cases reliably by waiting for inspiration. Go through a checklist. Not every line applies to every problem, but asking each question takes a minute:

| Ask about | Examples |
|---|---|
| **Zero, one, many** | an empty list, string or file; exactly one item; two items (the smallest case with a "between") |
| **Same and ties** | all items equal; duplicates; two items tied for best |
| **Signs and zero** | negative numbers, zero, a mix |
| **First, last, middle** | the thing you're looking for is at the very start, the very end, or not there at all |
| **Boundaries** | exactly on a limit (a mark of exactly 50, a time of exactly 00:00), and one either side |
| **Order** | already sorted, sorted backwards, all shuffled |
| **Size** | very large values, very long inputs (is it still fast enough?) |
| **Text** | spaces at the ends, upper and lower case, punctuation, accents and other alphabets, an empty string |
| **Bad input** | the wrong type, a format that's nearly right, a missing file |

Apply it to a small problem: *"Return the longest word in a sentence."*

- **Zero, one, many**: `""` (no words: `""`? `None`? an error?), `"hello"`.
- **Ties**: `"cat dog"`: the first, the last, or both?
- **Text**: `"  hi   there  "` (extra spaces), `"Hello, world!"` (does the comma count as part of the word?).
- **First, last**: the longest word at the start, and at the end.

Four of those are **decisions** the one-sentence description didn't make. That's normal; Day 1 said how to handle it.

## Deciding what should happen

For each edge case there are only a few sensible behaviours:

1. **There's a natural answer**: the sum of an empty list is 0; the longest word of `""` could reasonably be `""`.
2. **Return a value meaning "none"**: `None`, as `second_largest` does. Simple, but every caller must remember to check for it.
3. **Raise an exception**: the average of nothing doesn't exist, so `raise ValueError(...)`. Clear, and it can't be ignored by accident.
4. **It can't happen**: you've checked earlier in the program that the list is never empty. Then say so in the docstring, and perhaps with an `assert`.

What's never acceptable is behaviour nobody decided: an `IndexError` from deep inside, or a confident wrong answer like `start_at_zero`'s `0`.

Notice where the decisions went: the docstring states them, and a check pins each one down. `>` (not `>=`) is what makes the **first** of several tied words win; a check makes sure nobody "tidies" it into `>=` later.

## Boundaries

A **boundary** is a value where the behaviour changes: a pass mark, a size limit, the last valid index. Off-by-one mistakes (`<` instead of `<=`, `range(n)` instead of `range(n + 1)`) are the commonest bugs of all, and they only show up **on** the boundary.

The rule is to test three values at every boundary: just below, exactly on, and just above.

If the description had said "children 12 and under", then 12 would be the one value that changes: only a check exactly **on** the boundary can tell the two rules apart.
