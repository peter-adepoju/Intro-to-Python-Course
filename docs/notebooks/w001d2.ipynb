## Values have types

A **value** is a single piece of data: `42`, `3.14`, `"hello"`. Every value has a **type**, which tells Python what kind of thing it is and what you can do with it. Remember the eight switches from yesterday: the same bits can be a number or a letter. The type decides.

Today's four basic types:

| Type | Name | Examples | What it's for |
|---|---|---|---|
| `int` | integer | `7`, `0`, `-12`, `1_000_000` | whole numbers |
| `float` | floating-point number | `3.14`, `-0.5`, `2.0`, `1e3` | numbers with a decimal point |
| `str` | string | `"hi"`, `'42'`, `""` | text |
| `bool` | Boolean | `True`, `False` | yes/no, on/off |

Try sorting these values into their types. Then ask Python to check. Some are traps, so look closely at quotes and decimal points.

The rule that catches most people: **anything in quotes is a string**, even if it looks like a number. `"42"` is text made of the characters `4` and `2`. And `2.0` is a float, not an int, because it has a decimal point.

### Asking Python: type()

The function `type()` tells you the type of any value.

`<class 'int'>` is Python's way of saying "the type int". (Types are called *classes* in Python; you'll make your own in Phase 3.)

Use the inspector below to look at any value. Try `2 ** 100`: a gigantic number. Python's ints can be as big as your memory allows.

### Writing numbers

- Long numbers can use underscores to make them readable: `1_000_000` is one million. The underscores are ignored.
- `1e3` means 1 × 10³ = `1000.0` (a float). `2.5e-3` is `0.0025`. This is *scientific notation*.
- Don't write commas inside numbers: `1,000` isn't one thousand in Python. It's two values, `1` and `0`, separated by a comma.

## The arithmetic operators

Python has seven arithmetic operators. Drag the sliders below and watch all seven results update at once.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide | `7 / 2` | `3.5` |
| `//` | floor divide (divide and round down) | `7 // 2` | `3` |
| `%` | remainder ("modulo") | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

`//` and `%` are a pair: `7 // 2` is how many whole 2s fit into 7 (three), and `7 % 2` is what's left over (one). On Day 4 you'll explore them properly. They're more useful than they look.

### int or float?

Python follows simple rules for the type of the answer:

1. `/` **always** gives a float, even when the division is exact: `6 / 3` is `2.0`.
2. If **either** side is a float, the answer is a float: `2 + 1.0` is `3.0`.
3. Otherwise (`int` with `int`, using `+ - * // % **`) the answer is an int.

## Order of operations

What is `2 + 3 * 4`? It depends on which you do first. Python follows fixed rules, the same ones you learnt in school, often remembered as **BODMAS** or **PEMDAS**:

1. **Brackets** first: `( )`
2. **Powers**: `**`
3. **Multiplying and dividing**: `* / // %` (left to right)
4. **Adding and subtracting**: `+ -` (left to right)

Don't just trust the rule: watch Python apply it. Press **Next step** to see exactly which part Python works out next. Then choose another example from the list, or type your own.

Two surprises worth looking at in the stepper:

- `10 - 4 - 3` is worked from **left to right**: `(10 - 4) - 3` = 3, not `10 - (4 - 3)` = 9.
- `2 ** 3 ** 2` is worked from **right to left**: `2 ** (3 ** 2)` = `2 ** 9` = 512. Powers are the exception.

And one more trap:

`**` is done before the minus sign, so `-3 ** 2` is `-(3 ** 2)` = `-9`. When in doubt, **add brackets**. Brackets never hurt, and they make your intention clear to the next person who reads your code.
