## How assignment works

```python
seconds = 10000
```

This is an **assignment statement**. Read it as "**seconds** is now **10000**", or "let seconds refer to 10000". It is *not* a maths equation: `=` in Python is an instruction, not a statement of fact.

What Python does, in this order:

1. Work out the value on the **right** of `=`.
2. Make the **name** on the left refer to that value.

Press **X-ray** and step through. Watch the Names column on the left and the Objects column on the right, and the arrows between them.

On the third line, Python first looks up `price` and `quantity`, multiplies them, and *then* creates the name `total`. Because the right side is always worked out first, a name must exist before you use it:

## Names point at objects

There's a common way to picture variables: as boxes with a name on the front, and a value inside. It's a nice picture, but it's not how Python works, and it will mislead you later (badly, in Week 8, when you meet lists).

In Python, **values are objects** that live in memory, and **a variable is a name that points at an object**, like a label tied to it with string. Assignment ties the label on. Step through both pictures side by side:

The key moment is `y = x`. Python does **not** copy anything. It ties a second label, `y`, onto the very same object `x` points at. Then `x = 7` moves the `x` label to a new object. `y` still points at 5.

## Changing a variable

A variable can point at a different value later. A very common pattern is to update a variable using its own current value:

`score = score + 10` makes no sense as maths, but perfect sense as an instruction: work out `score + 10` (using the *current* score), then make `score` point at the result. X-ray it and watch the arrow move each time.

### Shortcut operators

Updating a variable is so common that Python has shortcuts:

| Shortcut | Means |
|---|---|
| `x += 5` | `x = x + 5` |
| `x -= 2` | `x = x - 2` |
| `x *= 3` | `x = x * 3` |
| `x /= 4` | `x = x / 4` |
| `x //= 4` | `x = x // 4` |
| `x %= 4` | `x = x % 4` |
| `x **= 2` | `x = x ** 2` |

## Several names at once

You can give several names values in one line. The right side is worked out completely first, then the names are assigned left to right.

This makes **swapping** two variables easy. Without it, you'd need a temporary third variable:

X-ray the second version. On line 3, Python first builds the pair `("blue", "red")` from the right side, then points `a` and `b` at them. Nothing is lost.

You can also point several names at the same value: `x = y = 0`.

## Variables don't have types, values do

A variable can point at an int now and a string later. The *value* has a type; the name doesn't care.

This is called **dynamic typing**. It's convenient, but it means *you* must keep track of what each name refers to. Good names help enormously.

## Naming things well

Variable names have rules (Python enforces them) and conventions (programmers follow them).

**Rules:**

- Letters, digits and underscores only. No spaces, no hyphens.
- Can't start with a digit.
- Can't be a **keyword**: one of Python's reserved words like `if`, `for`, `class`, `True`.
- Capitals matter: `score`, `Score` and `SCORE` are three different names.

**Conventions:**

- Use `lower_case_with_underscores` (called *snake_case*): `total_price`, `num_students`.
- Say what it means: `seconds_per_day` beats `x` or `spd`.
- Don't reuse the names of built-in functions like `print`, `type`, `list`, `sum`, `max`. Python allows it, but then the built-in stops working.
