````python
words = ["apple", "bean", "avocado", "cherry", "banana", "apricot"]
groups = {}
for w in words:
    groups.setdefault(w[0], []).append(w)
print(groups)
print(groups["a"])
````

````python
marks = {"Ada": 93, "Bo": 41, "Cy": 78, "Di": 55, "Ed": 71}
by_grade = {}
for name, m in marks.items():
    grade = "A" if m >= 70 else "B" if m >= 50 else "F"
    by_grade.setdefault(grade, []).append(name)
print(by_grade)
````

````python
words = ["listen", "silent", "enlist", "google", "inlets", "banana"]
groups = {}
for w in words:
    groups.setdefault("".join(sorted(w)), []).append(w)
for key, group in groups.items():
    if len(group) > 1:
        print(group)
````

> **Key idea.** **Grouping:** `groups.setdefault(key, []).append(item)`. Use it whenever a problem says "for each X, list the Ys" or "group by".

````python
sales = [("Lagos", 500), ("Abuja", 300), ("Lagos", 200), ("Kano", 150), ("Abuja", 50)]
totals = {}
for city, amount in sales:
    totals[city] = totals.get(city, 0) + amount
print(totals)
````

````python
books = [
    ("978-0", "Things Fall Apart", 1958),
    ("978-1", "Half of a Yellow Sun", 2006),
    ("978-2", "The Famished Road", 1991),
]
by_isbn = {}
for book in books:
    by_isbn[book[0]] = book
print(by_isbn["978-2"])
print(by_isbn["978-1"][1])
````

````python
from collections import Counter
counts = Counter("mississippi")
print(counts)
print(counts["s"], counts["z"])
print(counts.most_common(2))
````

````python
from collections import defaultdict
groups = defaultdict(list)
for w in ["apple", "bean", "avocado", "banana"]:
    groups[w[0]].append(w)
print(dict(groups))
totals = defaultdict(int)
for city, amount in [("Lagos", 500), ("Kano", 150), ("Lagos", 200)]:
    totals[city] += amount
print(dict(totals))
````

> **Careful.** With a `defaultdict`, even **reading** a missing key adds it. `if groups["z"]:` quietly creates `groups["z"] = []`. Use `in` to check whether a key is there.

> **Tip.** Learn the hand-written patterns first; they work everywhere and show what's going on. Then use `Counter` and `defaultdict` in your own code: they're clearer, and other programmers expect them. In the exercises today, some ask for one and some for the other.

## Counting

The **counting pattern** keeps one entry per distinct item, holding how many times it has been seen:

```
counts = {}
for item in items:
    counts[item] = counts.get(item, 0) + 1
```

`counts.get(item, 0)` is the count so far, or 0 for an item seen for the first time. Step through it and watch the dictionary grow:

Anything hashable can be counted: words, letters, numbers, tuples.

## Grouping

The **grouping pattern** collects items into lists, one list per key:

```
groups = {}
for item in items:
    key = the property to group by
    groups.setdefault(key, []).append(item)
```

The key can be anything you can work out from the item: its length, its first letter, whether it passed, which band a number falls in.

## Counter and defaultdict

These patterns are so common that Python's standard library has tools for them, in the `collections` module.

### Counter

`Counter` counts everything in a sequence in one go. It's a dictionary, with extras:

Two things to notice. A missing key gives **0**, not a `KeyError`. And `most_common(n)` gives the top `n` as `(item, count)` pairs.

`from collections import Counter` brings in just the name `Counter`, so you can write `Counter(...)` instead of `collections.Counter(...)`. You'll learn more about imports in Phase 2.

### defaultdict

`defaultdict` is a dictionary that makes a value for a missing key automatically. You tell it **how** to make one by giving it a function to call: `list` makes `[]`, `int` makes `0`.

`groups[w[0]].append(w)` works even the first time, because the missing key gets a fresh empty list. Note `list`, not `list()`: you hand over the function, and `defaultdict` calls it when it needs a new value.

## Choosing a pattern

| The problem says… | Pattern | Result |
|---|---|---|
| how many of each, how often, most common | counting | key → count |
| for each X, the list of Ys; group by | grouping | key → list |
| total (or average) for each X | totals | key → number |
| find the record for an ID or name | index | key → record |
