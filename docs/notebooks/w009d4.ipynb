> **Key idea.** `for i, item in enumerate(items):` gives each item with its position. Use it instead of `for i in range(len(items)):` whenever you need both. Note the order: the **position comes first**.

````python
prices = [100, 250, 80]
for i, p in enumerate(prices):
    prices[i] = p * 2
print(prices)
````

````python
names = ["Ada", "Grace", "Alan"]
scores = [93, 88, 97]
for name, score in zip(names, scores):
    print(f"{name:<6}{score:>4}")
````

````python
days = ["Mon", "Tue", "Wed"]
highs = [31, 29, 33]
lows = [24, 23, 25]
for day, hi, lo in zip(days, highs, lows):
    print(day, "range", hi - lo)
````

> **Careful.** If two lists should line up but might not, `zip` hides the problem by stopping early. Check `len(a) == len(b)` first, or use `strict=True`.

````python
names = ["Ada", "Grace", "Alan"]
scores = [93, 88, 97]
for rank, (name, score) in enumerate(zip(names, scores), start=1):
    print(rank, name, score)
````

````python
temps = [21, 23, 22, 25, 30]
for today, tomorrow in zip(temps, temps[1:]):
    print(today, "->", tomorrow, "change", tomorrow - today)
````

````python
grid = [[1, 2, 3], [4, 5, 6]]
for column in zip(*grid):
    print(column)
transposed = []
for column in zip(*grid):
    transposed.append(list(column))
print(transposed)
````

### Bug hunt: A missing student

Five names but only four marks, and nobody notices: the last student just disappears. Make `report` return `"Lengths differ"` when the lists don't line up, and otherwise the lines joined by `"\n"`.

````python
def report(names, marks):
    lines = []
    for name, mark in zip(names, marks):
        lines.append(f"{name} {mark}")
    return "\n".join(lines)
````

### When the lengths differ

`zip` stops as soon as the **shortest** sequence runs out, and says nothing about it:

That's handy sometimes, and a silent bug at others, when you expected the lengths to match. In Python 3.10 and later you can write `zip(a, b, strict=True)` to make it raise an error instead. Otherwise, check the lengths yourself when they matter.

## Lazy, and used up

Look at what `enumerate` and `zip` give you on their own:

They don't make a list of all the pairs at once. They're **iterators**: objects that produce the next pair only when a loop (or `list`) asks for it. That saves memory when the sequences are long. But an iterator can be used only **once**: after you've gone through it, it's empty. The second `list(z)` gives `[]`.
