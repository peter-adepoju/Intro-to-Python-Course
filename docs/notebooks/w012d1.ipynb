````python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
````

## Values, types and strings

Week 1's ideas in brief: every value has a **type** (`int`, `float`, `str`, `bool`); operators behave differently for different types; a variable is a **name** for a value, and assignment moves the name; strings are sequences you can index and slice; `input` always gives a string.

## Decisions

Week 2: comparisons give booleans; `and`, `or` and `not` combine them; `if`, `elif` and `else` choose **one** branch, checked from the top.

## Loops

Weeks 3 and 4: `while` repeats while a condition is true; `for` goes through a sequence; `range(start, stop, step)` stops **before** `stop`; `break` leaves a loop, `continue` skips to the next round; nested loops run the inner loop completely for each round of the outer one.

## The classic mistakes

These are the mistakes that every beginner makes, and that still trip up experienced programmers when they're tired. For each, can you say **why** it goes wrong?

| Mistake | What happens | The fix |
|---|---|---|
| `age = input("Age: ")` then `age + 1` | `TypeError`: input gives a string | `int(input(...))` |
| `if x = 5:` | `SyntaxError`: `=` assigns | `==` compares |
| `if answer == "yes" or "y":` | always true: `"y"` on its own is truthy | `answer in ("yes", "y")` |
| `while n > 0:` with no change to `n` | runs for ever | change `n` in the body |
| `for i in range(1, 10)` to mean 1 to 10 | stops at 9 | `range(1, 11)` |
| `total = 0` inside the loop | resets every time | set it once, before the loop |
| `print` in a function instead of `return` | the caller gets `None` | `return` the value |
| `0.1 * 3 == 0.3` | `False`: floats are approximate | `round` or `abs(a - b) < 1e-9` |
