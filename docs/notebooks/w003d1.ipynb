## How while works

```python
while n <= 5:
    print(n)
    n = n + 1
```

A `while` loop looks like an `if`: a condition, a colon, and an indented block called the **body**. The difference is what happens after the body:

1. Check the condition.
2. If it's `False`, skip the body and carry on after the loop. The loop is over.
3. If it's `True`, run the body, then **go back to step 1**.

So the body keeps running *while* the condition stays true. Press X-ray on the loop above and step through. Watch the yellow line jump back up to the `while` line after every pass, and watch `n` grow.

Each pass through the body is called an **iteration**. The table below is what you'd write if you traced it by hand: one row per line Python runs, with every variable's value after that line. Press **Next row**, and predict each row before it appears.

Look at the last check of the condition: `n` is 4, `4 <= 3` is `False`, and only then does the loop end. The condition is checked one more time than the body runs.

## Counting loops: three parts

Almost every counting loop has the same three parts. Learn to see them:

```python
n = 1              # 1. start: give the counter its first value
while n <= 5:      # 2. condition: keep going while this is true
    print(n)
    n = n + 1      # 3. update: move the counter towards the end
```

Change any part and the loop changes:

The update doesn't have to be `+ 1`. It just has to move the counter towards making the condition false.

## How many times does it run?

Two loops that look almost the same:

The first runs 4 times, the second 5. A useful rule for counting up by 1:

- `while n < stop:` starting from `start` runs `stop - start` times;
- `while n <= stop:` runs `stop - start + 1` times.

Getting this wrong by one is called an **off-by-one error**: the loop runs once too often or once too few. It's the loop version of the boundary bugs you met last week, and you'll meet it for the rest of your programming life. The cure is the same: check the first and the last iteration by hand.

## Infinite loops

If the condition never becomes false, the loop never ends. This is an **infinite loop**:

```python
n = 1
while n <= 5:
    print(n)
```

The update is missing, so `n` stays 1 forever and Python prints 1 forever. Infinite loops are the most common loop bug. Their usual causes:

- **No update**: the counter never changes.
- **Update in the wrong direction**: `n -= 1` when the condition waits for `n` to get big.
- **Update skipped**: the update sits inside an `if` that doesn't always run.
- **A condition that can't fail**: `while n != 10` with `n += 3` from 0 jumps straight past 10.

Nothing is printed and nothing seems to happen. On this site, a program that runs for more than 10 seconds is stopped for you, with a message. In other places, you stop a running program by pressing **Ctrl** + **C**.

## Decisions inside loops

The body of a loop can hold anything, including an `if`. The `if` is checked on every pass:
