> **+= is mutation for lists.** For numbers, `n += 1` means `n = n + 1`. For lists, `x += [4]` **changes the list in place**, like `append`, rather than making a new list. So with two names on one list, `a += [4]` changes what `b` sees, but `a = a + [4]` doesn't.

````python
a = [1, 2, 3]
b = a
a += [4]
print(b)
````

> **Common mistake.** `for item in items: item = ...` never changes the list. Use `for i in range(len(items)): items[i] = ...`, or build a new list.

````python
nums = [1, 2, 2, 3]
for n in nums:
    if n == 2:
        nums.remove(n)
print(nums)
````

````python
def add_tax(prices):
    for i in range(len(prices)):
        prices[i] = round(prices[i] * 1.075, 2)

basket = [100, 200]
add_tax(basket)
print(basket)
````

````python
def scale(values, factor):
    """Multiply every item of values by factor, in place. Returns None."""
    for i in range(len(values)):
        values[i] *= factor

def scaled(values, factor):
    """Return a new list: every item of values times factor."""
    result = []
    for v in values:
        result.append(v * factor)
    return result

data = [1, 2, 3]
print(scaled(data, 10), data)
scale(data, 10)
print(data)
````

> **Tip.** Prefer functions that return a new list. They can't surprise anyone by changing data behind their back. Change in place when the list is large and copying would waste memory, or when changing it is the whole point, and say so in the docstring.

## Mutation versus rebinding

This is the most important idea of the day. There are two very different ways to "change" `x`:

- **Mutation**: change the list object itself. `x[0] = 5`, `x.append(5)`, `del x[0]`.
- **Rebinding**: make a **new** object and move the name `x` to it. `x = x + [5]`, `x = []`.

When only one name points at the list, you can't see the difference. When two do, you can:

X-ray both. In the first, there's one list with two names on it, and appending changes that one list. In the second, `a + [4]` builds a **new** list, and only `a` is moved to it; `b` still points at the old one.

This is the labels picture from Week 1, Day 3 doing real work. With numbers and strings it never mattered, because they can't be changed. With lists it matters all the time. You'll spend Day 5 on it.

## Changing every item

To change every item, loop over the **indexes** and assign through them:

A tempting mistake is to change the loop variable instead:

The loop variable `p` is a **name** pointing at each item in turn. `p = p * 1.1` moves the name `p` to a new number; the list's slots still point at the old ones. Only assigning **through the list**, `prices[i] = ...`, changes a slot.

### Don't change a list's length while looping over it

Removing items from a list while a `for` loop walks over it makes the loop skip items, because everything after the removed one shifts left:

(`.remove(v)` deletes the first item equal to `v`; you'll meet it properly tomorrow.) One `2` survived. The safe way is to build a new list of the items you want to **keep** (the filter pattern), which you'll practise in Day 4.

## Functions that change lists

Because a parameter is just another name for the caller's object, a function can **change** a list it's given, and the caller will see the change:

That's very different from numbers and strings, where a function could never affect the caller's variable. So for every function that takes a list, decide which kind it is, and say so in its docstring:

- **It changes the list in place** and returns `None`, like `list.sort()` and `append`.
- **It leaves the list alone and returns a new one**, like `sorted()`.
