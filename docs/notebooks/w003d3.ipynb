> **Key idea.** `break` exits the **innermost** loop it's in, straight away. Whatever comes after `break` in the body is skipped.

````python
total = 0
while True:
    n = int(input("Number (0 to stop): "))
    if n == 0:
        break
    total += n
print("Total:", total)
````

````python
while True:
    choice = input("1 = hello, 2 = time, 3 = joke, q = quit: ")
    if choice == "q":
        print("Goodbye!")
        break
    elif choice == "1":
        print("Hello!")
    elif choice == "2":
        print("Time to practise.")
    elif choice == "3":
        print("Why do programmers mix up Halloween and Christmas? Because Oct 31 = Dec 25.")
    else:
        print("Unknown choice.")
````

````python
n = 0
while n < 10:
    n += 1
    if n % 3 == 0:
        continue
    print(n, end=" ")
````

````python
count = 0
while True:
    word = input("Word (done to stop): ")
    if word == "done":
        break
    if word == "":
        continue
    count += 1
print(count, "words")
````

> **Common mistake.** `continue` skips everything below it, **including the update**. Put the update before any `continue`, or avoid `continue` and use an `if` around the work instead.

## Warm-up

That loop ran until its condition failed. Today you'll learn two ways to control a loop from **inside** its body.

## break: leave the loop now

`break` ends the loop immediately. Python jumps to the first line after the loop, without checking the condition again and without finishing the current pass.

Without the `break`, the loop would keep going to 999, printing every multiple of 7 on the way. With it, the search stops at the first match. Press X-ray and watch: after `break`, the yellow line jumps straight past the loop.

## while True and the loop and a half

Yesterday's priming read needs the `input()` line written twice: once before the loop, once at the end of the body. There's another shape that writes it once:

`while True:` would loop forever, since `True` is always true. The only way out is the `break`. This shape is sometimes called a **loop and a half**: each pass reads, checks whether to stop, and only then does the work.

Compare it with the priming read version from yesterday. Both are correct. Which you use is mostly taste:

- The priming read puts the stopping rule in the `while` line, where readers look for it first.
- `while True` avoids repeating the input line, which matters when getting the next value takes several lines.

The joke: 31 written in base 8 (octal) is the same number as 25 in our usual base 10.

## continue: skip to the next pass

`continue` abandons the **rest of the current pass** and goes back to check the condition. The loop itself carries on.

When `n` is 3, 6 or 9, `continue` jumps back to the `while` line, so the `print` is skipped for those values only.

### The continue trap

Look at where `n += 1` is in the first example: at the **top** of the body, before the `continue`. If it were at the bottom, then the first time `continue` ran, the update would be skipped. `n` would stay the same, the same `continue` would run again, and the loop would never end:

```python
n = 0
while n < 10:
    if n % 3 == 0:
        continue        # when n is 0, this runs forever
    print(n)
    n += 1              # never reached once n is a multiple of 3
```

Often an `if` is clearer anyway:

Use `continue` when it saves a deep indent, typically for "skip the bad cases first, then do the real work" at the top of a long body.

### With while ... else

Python has a feature most languages don't: a loop can have an `else`. The `else` block runs if the loop ended **normally**, because its condition became false. It does **not** run if the loop ended with `break`.

Read the `else` as "if we never broke out", or "**no break**". It's made for searches: break when found, `else` for not found.
