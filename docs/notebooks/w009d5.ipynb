> **Key idea.** `[expression for item in sequence]` makes a new list holding `expression` worked out once for each item. It's exactly the loop `result = []`, `for item in sequence:`, `result.append(expression)`, written in one line.

> **Common mistake.** `[n for n in nums if n > 0 else 0]` is a `SyntaxError`. A filter at the end has no `else`. If you want an `else`, you want a choice, so the `if ... else` goes **before** the `for`.

````python
marks = [93, 41, -1, 50, 67]   # -1 means absent
print(["pass" if m >= 50 else "fail" for m in marks if m >= 0])
````

````python
grid = [[0] * 3 for _ in range(2)]
grid[0][0] = 5
print(grid)
````

````python
grid = [[1, 2, 3], [4, 5, 6]]
print([[cell * 10 for cell in row] for row in grid])
print([sum(row) for row in grid])
print([list(col) for col in zip(*grid)])
````

````python
grid = [[1, 2, 3], [4, 5, 6]]
flat = [cell for row in grid for cell in row]
print(flat)
````

> **Tip.** Read `[cell for row in grid for cell in row]` as the loop `for row in grid:` then `for cell in row:` then append `cell`. The `for`s keep their loop order. More than two `for`s, or two `for`s with conditions, is usually clearer as a plain loop.

````python
prices = [100, 250, 80]
qty = [2, 1, 5]
print(sum(p * q for p, q in zip(prices, qty)))
print(max(len(w) for w in ["tea", "biscuits", "jam"]))
````

````python
marks = [93, 41, 50, 67]
print(any(m < 50 for m in marks))
print(all(m >= 40 for m in marks))
print(all(c.isdigit() for c in "2026"))
print(any(c.isdigit() for c in "Python"))
````

## Transforming: one new item for each old one

A **list comprehension** builds the same list in one expression:

Read it from the middle out: *for each `n` in `range(1, 6)`, take `n * n`, and collect the results in a list.* The square brackets say that the result is a list.

### A filter or a choice?

There are two places an `if` can go, and they do different things.

- At the **end**, `if` is a **filter**: it decides whether an item goes in at all. The result can be shorter.
- In the **expression**, `a if condition else b` is a **conditional expression** (Week 2): every item goes in, but it chooses what goes in. The result is the same length, and the `else` is required.

You can use both together: filter first, then choose.

## sum, any and all

If you only need the list to pass it to `sum`, `max`, `min` or `len`, you can leave the square brackets out:

Without brackets this is a **generator expression**. Like `zip`, it produces its values one at a time rather than building the whole list, which saves memory when there are many of them.

In Week 8 you gave `any` and `all` a list of `True`/`False` values that took a loop to build. Now the loop fits inside the call. `any(...)` is `True` if **at least one** value is true; `all(...)` is `True` if **every** value is:

They also replace the "flag" loops from Week 4, where a variable starts as `False` and changes when something is found. Both stop as soon as the answer is known: `any` at the first true value, `all` at the first false one.

As in Week 8: an empty sequence has nothing that fails, so `all` gives `True`, and nothing true, so `any` gives `False`.

## When not to use one

A comprehension is for **building a list**. Use a plain loop when:

- You're doing something for its **effect**, such as printing. `[print(x) for x in items]` works, but it builds a list of `None`s nobody wants. Write a `for` loop.
- The logic needs several steps, an `elif`, a `break`, or a running total that depends on earlier items.
- It no longer fits comfortably on one line, or you have to read it twice.
