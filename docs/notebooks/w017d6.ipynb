> **The same three tricks.** Look at what made each function fast. Three ideas did nearly all the work: **a set or dictionary** to replace a search through a list (`in`, `count`); **precomputing** once what was being recomputed every time (the counts, the running totals); and **sorting first** when order makes the answer easy to find. Most everyday speed-ups are one of these three. Each one spends O(n) memory to save time, a trade you'll make again and again.

## Forge: make it fast

A school has asked for help. Its data scripts work on a class of 30, but take hours on the whole district's records. Here are the five slow functions, all correct:

```
def shared_students(club_a, club_b):
    """Names in both clubs, in the order of club_a."""
    return [name for name in club_a if name in club_b]

def first_unrepeated(text):
    """The first character that appears exactly once, or None."""
    for ch in text:
        if text.count(ch) == 1:
            return ch
    return None

def has_gap(scores, d):
    """Do any two scores differ by exactly d (d > 0)?"""
    for a in scores:
        for b in scores:
            if b - a == d:
                return True
    return False

def range_totals(values, queries):
    """For each (start, end) in queries, the total of values[start:end]."""
    return [sum(values[s:e]) for s, e in queries]

def most_common_pair(names):
    """The pair of neighbouring names (a, b) that appears most often in
    the list, the alphabetically first if tied; None if fewer than 2 names."""
    pairs = [(names[i], names[i + 1]) for i in range(len(names) - 1)]
    if not pairs:
        return None
    best = None
    for p in pairs:
        if best is None or pairs.count(p) > pairs.count(best) or \
           (pairs.count(p) == pairs.count(best) and p < best):
            best = p
    return best
```

## Forge bonus: counting inversions

An **inversion** in a list is a pair of positions `i < j` where `items[i] > items[j]`: two items in the wrong order. A sorted list has 0; a list sorted backwards has the most, n(n − 1)/2. Counting inversions measures how far from sorted a list is, and it's used to compare rankings: two people's top-10 films, say, or a search engine's results against what users clicked.

Checking every pair is O(n²). Divide and conquer does it in O(n log n): split the list in half, count the inversions inside each half recursively, and while **merging** the two sorted halves, count the inversions that cross between them. When the merge takes an item from the right half, it jumps ahead of every item still waiting in the left half, and each of those is one inversion.
