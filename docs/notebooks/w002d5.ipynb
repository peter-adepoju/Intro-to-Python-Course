```python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
```

## Design first, then code

So far the rules have been handed to you, neatly. Real rules arrive as sentences, often vague ones. Here's one, from a shop:

> Delivery is free on orders of 20,000 or more. Smaller orders pay 2,500, except that members pay half that. Orders below 1,000 can't be delivered.

Before writing a single `if`, answer three questions:

1. **What are the inputs?** The order total, and whether the customer is a member.
2. **What are the possible outcomes?** Free, 2,500, 1,250, or "can't deliver".
3. **Which inputs lead to which outcome?** Write it as a table.

A **decision table** lists the conditions and what happens in each case:

| Order total | Member? | Delivery |
|---|---|---|
| below 1,000 | either | can't deliver |
| 1,000 to 19,999 | no | 2,500 |
| 1,000 to 19,999 | yes | 1,250 |
| 20,000 or more | either | free |

Writing the table forces you to settle the questions the sentence left open. Is 20,000 itself free? ("or more" says yes.) Is 1,000 deliverable? ("below 1,000 can't" says yes.) Is the member discount on free delivery, half of nothing? (It doesn't matter.) When the rule is unclear, ask whoever wrote it. When you can't ask, write your decision down.

## Testing: which inputs should you try?

You can't try every possible input. You *can* choose a small set that's likely to catch mistakes. Three kinds matter.

**1. One typical value for every branch.** If a branch never runs during your testing, you have no idea whether it works. For the delivery rule: 500 (can't deliver), 12,000 as a non-member, 12,000 as a member, 30,000.

**2. The boundaries.** Most decision bugs are at the edges: `<` written where `<=` was meant, or 2000 typed for 20000. For every threshold, test **just below, exactly on, and just above** it. For 20,000: 19,999, 20,000 and 20,001.

**3. The strange cases.** Zero. Negative numbers. Empty input. Text where a number was expected. The largest value that makes sense.

Here is a buggy version of the grading chain from yesterday. Somewhere one boundary is wrong. Drag the slider to the edges of each band, or watch the strip for a band that's one mark too wide:

A mark of exactly 70 should be an A, but `mark > 70` sends it to B. Tests at 71 and 65 would never find this; only a test *at* 70 does. That's why boundaries come first.

### Writing a test plan

A **test plan** is a list of inputs, each with the output you *expect*, written before you run anything. Then you run the program with each input and compare. For the delivery program:

| Order | Member | Expected | Why |
|---|---|---|---|
| 500 | n | Can't deliver | first branch |
| 999 | n | Can't deliver | just below 1,000 |
| 1000 | n | 2500 | exactly on 1,000 |
| 12000 | y | 1250 | member branch |
| 19999 | n | 2500 | just below 20,000 |
| 20000 | y | free | exactly on 20,000 |
| 0 | n | Can't deliver | strange: zero |

## Checking input before you trust it

What happens if the user types `ten` where you expect a number?

The program crashes with a `ValueError` before your decision code even runs. You'll learn the full way to handle this, `try` and `except`, in Week 11. Until then, there's a simple check: a string's `.isdigit()` is `True` when the string is not empty and **every** character is a digit from 0 to 9.

Notice the last two: a minus sign and a decimal point aren't digits, so `.isdigit()` only accepts whole numbers of zero or more. That's often exactly what you want for quantities and ages.

## Simplifying conditions

Clear conditions are easier to test, and they hide fewer bugs. Here are three things to tidy.

### Repeating what the chain already knows

When Python reaches the first `elif`, it already knows `mark >= 70` was false, so `mark < 70` is always true there. The extra checks do nothing, except give you more places to make a mistake. Write just `elif mark >= 60:`.

### Setting a bool with if and else

### Comparing a bool with True

`if is_adult == True:` works, but `is_adult` is already a bool. Write `if is_adult:`, and `if not is_adult:` for the opposite.

### Branches that do the same thing

If two branches contain identical code, the conditions can usually be joined with `or`, or the shared lines moved out of the decision entirely.
