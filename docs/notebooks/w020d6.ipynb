### Exercise: Forge: a spending tracker

Write `build()` that creates the package with the starter's `write_file` helper:

1. `spend/records.py`: a namedtuple `Purchase` with fields `day` (a `date`), `category`, `amount` (a float) and `note`, and `parse(text)` returning a list of `Purchase`s (skip blank lines).
2. `spend/totals.py` (no printing): `by_category(purchases)` returning a `Counter` of totals per category, each rounded to 2 decimal places; `by_month(purchases)` returning a plain dict from `"YYYY-MM"` to the total, rounded to 2 places; `biggest(purchases, k)` returning the k largest purchases, largest first.
3. `spend/summary.py`: `summary(text)` returning a string report (exact layout below), using the other modules through **relative** imports and `statistics.mean` for the average purchase.
4. `spend/__init__.py`: makes `summary` importable as `from spend import summary`.

For the sample data, `summary` returns exactly:

```
8 purchases, 119.15 in total, 14.89 on average
food: 52.95
bills: 45.0
fun: 15.0
transport: 6.2
2026-10: 91.45
2026-11: 27.7
Biggest: bills 45.0 (phone)
```

Categories are listed by total, highest first (ties alphabetically); months in order.

````python
import os

def write_file(name, text):
    """Write text to a file, making its folder first if needed."""
    os.makedirs(os.path.dirname(name) or ".", exist_ok=True)
    with open(name, "w") as f:
        f.write(text)

def build():
    write_file("spend/__init__.py", """___""")
    write_file("spend/records.py", """___""")
    write_file("spend/totals.py", """___""")
    write_file("spend/summary.py", """___""")
````

````python
# Check: run the exercise cell above first, then this one.
check("TEXT = \"\"\"2026-10-02,food,12.40,market\n2026-10-02,transport,3.10,bus\n2026-10-05,food,7.85,lunch\n2026-10-14,bills,45.00,phone\n2026-10-20,food,23.10,market\n\n2026-11-01,transport,3.10,bus\n2026-11-03,fun,15.00,cinema\n2026-11-04,food,9.60,lunch\n\"\"\"\nREPORT = \"\"\"8 purchases, 119.15 in total, 14.89 on average\nfood: 52.95\nbills: 45.0\nfun: 15.0\ntransport: 6.2\n2026-10: 91.45\n2026-11: 27.7\nBiggest: bills 45.0 (phone)\"\"\"\ndef fresh():\n    import sys, importlib\n    for name in list(sys.modules):\n        if name == \"spend\" or name.startswith(\"spend.\"):\n            del sys.modules[name]\n    build()\n    importlib.invalidate_caches()\ndef test_records():\n    \"records\"\n    fresh()\n    from spend.records import parse, Purchase\n    from datetime import date\n    ps = parse(TEXT)\n    assert len(ps) == 8 and ps[0] == Purchase(date(2026, 10, 2), \"food\", 12.4, \"market\")\n    assert ps[0].day.month == 10 and isinstance(ps[0].amount, float)\ndef test_totals():\n    \"totals\"\n    fresh()\n    from spend.records import parse\n    from spend.totals import by_category, by_month, biggest\n    ps = parse(TEXT)\n    c = by_category(ps)\n    assert c[\"food\"] == 52.95 and c[\"transport\"] == 6.2 and c[\"missing\"] == 0\n    assert by_month(ps) == {\"2026-10\": 91.45, \"2026-11\": 27.7}\n    assert [p.note for p in biggest(ps, 2)] == [\"phone\", \"market\"]\n    assert \"print(\" not in read_file(\"spend/totals.py\")\ndef test_summary():\n    \"summary\"\n    fresh()\n    from spend import summary\n    got = summary(TEXT)\n    assert got == REPORT, \"Expected:\\n\" + REPORT + \"\\n\\nGot:\\n\" + got\n    assert \"from .\" in read_file(\"spend/summary.py\")")
````

> **One job per module.** Look at what each module knows about. `records` knows the text format and nothing about totals. `totals` knows how to add things up and nothing about text or printing. `summary` knows the report's layout and nothing about the file format. If the data moved from text lines to a spreadsheet, only `records` would change; if the report needed a new section, only `summary`. That independence is the whole point of splitting a program up, and it's what makes each part easy to test on its own.
