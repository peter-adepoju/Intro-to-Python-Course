> **Key idea.** Positional arguments are matched in order. Keyword arguments are matched by name and must come after positional ones. Parameters with defaults can be left out of the call, and must come after parameters without defaults.

> **Careful.** Use only simple, unchangeable values as defaults: numbers, strings, `True`, `False`. In Week 8 you'll see why a default like an empty list causes a famous bug.

````python
def add_tax(price):
    price = price * 1.075
    print("Inside:", price)

cost = 1000
add_tax(cost)
print("Outside:", cost)
````

> **Note.** Numbers and strings can't be changed, so this is the whole story for them. Lists can be changed in place, and then a function **can** affect the caller's data through a parameter. You'll see that, and how to control it, in Week 8.

## Warm-up

`sep` and `end` are **parameters** of `print`, with default values (a space and a new line) that you override by naming them. By the end of today you'll be able to write functions that work exactly like this.

## Several parameters

A function can have as many parameters as it needs, separated by commas. The arguments in a call are matched to the parameters **by position**: the first argument to the first parameter, the second to the second, and so on.

Because matching is by position, **order matters**. Get it wrong and the function silently does the wrong thing:

The second call runs without an error, but it's nonsense. Python can't know you meant the other order.

### The wrong number of arguments

If the number of arguments doesn't match, Python refuses before the body runs, and the error says exactly what's wrong:

`missing 1 required positional argument: 'height'` and `takes 2 positional arguments but 3 were given`. Read these messages carefully: they name the function and the parameter.

## Keyword arguments

You can also name the parameter in the call: `parameter=value`. These are **keyword arguments**, and with them the order doesn't matter:

Keyword arguments make calls easier to read, especially when a function has several parameters of the same type:

```python
book_room("Lagos", 3, 2)                          # what do 3 and 2 mean?
book_room(city="Lagos", nights=3, guests=2)       # clear
```

That's a `SyntaxError`: once you've named one argument, Python can't tell which parameter a following unnamed one is for.

## Default values

A parameter can have a **default value**, written `parameter=value` in the `def`. If the call doesn't give that argument, the default is used:

Defaults make a function easy to use in the common case, and flexible in the unusual one. That's exactly what `print` does with `sep` and `end`.

### Defaults must come last

Parameters with defaults must come **after** those without:

Think about why: in `greet("Ada")`, would `"Ada"` be the greeting or the name? Python avoids the question by not allowing it.

## Parameters are local labels

When a function is called, each parameter is a new name, inside the function's frame, pointing at the argument's object. Exactly as in Week 1, Day 3: names are labels.

So what happens if the function points its parameter at something else?

Step through it in X-ray. At the call, `price` in the function's frame points at the same object as `cost`: the number 1000. Then `price = price * 1.075` moves the label `price` to a **new** object. `cost`, in the main frame, still points at 1000. Reassigning a parameter never changes the caller's variable.

To get a value back out of a function, you need `return`, which is tomorrow's lesson.
