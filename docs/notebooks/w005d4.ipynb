## The format spec, part by part

Inside an f-string, `{value:spec}` formats `value` according to `spec`. A spec can have up to seven parts, and they always come **in this order**:

```
{value:  fill  align  sign  width  grouping  .precision  type}
         *     >      +     10     ,         .2          f
```

Every part is optional. Here's that whole example, and some of its parts on their own:

The parts:

| Part | Options | Meaning |
|---|---|---|
| fill | any one character | what pads the empty space (only with an align) |
| align | `<` `>` `^` `=` | left, right, centre; `=` pads after the sign |
| sign | `+` `-` space | show `+` on positives; only `-` (the default); a space for positives |
| width | a number | the minimum total width |
| grouping | `,` or `_` | thousands separators |
| precision | `.` then a number | digits after the point for `f`; maximum length for strings |
| type | `f` `e` `%` `d` `b` `o` `x` `X` `s` ... | how to write the value |

## Numbers: signs, bases and scientific

### Signs

`+` is perfect for changes, scores and temperatures, where a plus sign carries meaning.

### Percentages and scientific notation

`%` multiplies by 100 and adds a percent sign. `e` writes numbers in scientific notation, which is how scientists write very large and very small numbers:

`2.998e+08` means 2.998 × 10⁸. The `g` type ("general") chooses between plain and scientific automatically.

### Binary, octal and hexadecimal

In Week 1 you saw that computers store everything in binary. Format codes write integers in other bases: `b` for binary (base 2), `o` for octal (base 8), `x` and `X` for hexadecimal (base 16, using the digits 0 to 9 and a to f):

The `#` adds the prefix Python uses for that base (`0b`, `0o`, `0x`). `08b` pads with zeros to 8 binary digits, which is one **byte**.

Hexadecimal is everywhere in computing because each hex digit is exactly 4 bits. Web colours are written as three hex bytes, for red, green and blue:

That's the yellow this site uses to show the line that's about to run.
