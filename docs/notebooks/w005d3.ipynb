````python
digits = input("Digits: ")
best = -1
best_at = 0
for i in range(len(digits) - 1):
    s = int(digits[i]) + int(digits[i + 1])
    if s > best:
        best = s
        best_at = i
print(digits[best_at:best_at + 2], "add up to", best)
````

## Searching for a pattern yourself

`in`, `.find()` and `.count()` all search for a pattern. Underneath, the simplest way to search is this: put the pattern at every position in the text, and check whether the slice there matches.

The range stops at `len(text) - len(pattern) + 1` because the pattern has to fit: the last place it can start is `len(text) - len(pattern)`. Watch the search work character by character:

## Sliding windows

The search above looks at a **window** of the text, `len(pattern)` characters wide, at every position. The window idea is useful on its own:

Windows are how spell checkers suggest corrections, how DNA sequences are searched, and how you spot the hottest week in a year of temperature readings.

## Line by line

Text often has several **lines**. Inside a string, a line break is the single character `\n`. A string written with three quotes can span lines directly:

`.splitlines()` splits at every line break and returns the lines without the `\n` characters. Each line is an ordinary string: you can split it into words, strip it, search it.

Next week's files, and every log file and data file you'll ever read, are handled this way: one line at a time.

### Anagrams

Two words are **anagrams** if they use exactly the same letters the same number of times: `listen` and `silent`. With letter counts, the check is simple: the words are anagrams when every letter has the same count in both.
