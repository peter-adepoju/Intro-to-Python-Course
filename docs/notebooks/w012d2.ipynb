## Warm-up

If you got that right first time, you understand the most important idea of Weeks 8 to 10: a function receives the **same** list, so changing it in place is visible outside, but rebinding the parameter to a new list isn't. If not, keep this one on your list.

## Strings

Strings are immutable, so every method returns a **new** string. `split` and `join` move between text and lists. f-strings format values.

## Functions

A function packages steps behind a name. Parameters receive arguments; `return` sends a value back (and a function without it returns `None`); names assigned inside are **local**.

## Lists, tuples and comprehensions

Lists are mutable sequences; methods like `append`, `sort` and `reverse` change the list and return `None`. Assignment makes **aliases**, not copies. Tuples are immutable, and unpacking pulls values out. Comprehensions build lists in one line.

## Dictionaries and sets

Dictionaries map keys to values, with fast lookup; keys must be hashable. Sets hold unique items, with fast membership and set operations.

## Files and errors

Open files with `with`; read with `read`, `readline`, `readlines` or a `for` loop; write with mode `"w"` (replaces) or `"a"` (adds). Catch specific exceptions; raise your own with clear messages.

## The classic mistakes, part two

| Mistake | What happens | The fix |
|---|---|---|
| `items = items.sort()` | `items` becomes `None` | `items.sort()` alone, or `items = sorted(items)` |
| `def f(x, seen=[]):` | the list is shared between calls | `seen=None`, then `if seen is None: seen = []` |
| `copy = original` | an alias, not a copy | `original.copy()`, or `copy.deepcopy` for nested data |
| removing from a list while looping over it | items are skipped | loop over a copy, or build a new list |
| `counts[word] += 1` on a new word | `KeyError` | `counts.get(word, 0) + 1`, or `Counter` |
| `{}` for an empty set | it's a dictionary | `set()` |
| opening a file with `"w"` to update it | it's emptied before you read it | read everything first, then write |
| `except:` around everything | real bugs are hidden | catch the specific exception |
