- if it starts with a vowel (a, e, i, o, u), add `way` to the end: `apple` → `appleway`;
- otherwise, move all the consonants **before the first vowel** to the end and add `ay`: `string` → `ingstray`, `pig` → `igpay`;

## split: one string into many

`.split()` cuts the string wherever there's whitespace and gives back the pieces. The result, written with square brackets, is a **list**: an ordered collection of values. Lists get two whole weeks of their own (Weeks 8 and 9). For now you only need three things, and they all work the way they do for strings:

- `len(words)` is how many pieces there are;
- `words[0]` is the first piece, `words[-1]` the last;
- `for w in words:` visits each piece in turn.

### With no argument, or with a separator

## Strings can't be changed

You've seen hints of this all week. Now the rule: once a string exists, it can **never** be changed. Not one character of it.

`TypeError: 'str' object does not support item assignment`. To get `"hello"`, you build a **new** string and point the name at it:

This is called **immutability**, and it's deliberate. Remember from Week 1 that names are labels on objects. If two names point at the same string and strings could change, changing it through one name would silently change it for the other. Because strings are immutable, sharing them is always safe:

`a.upper()` made a new string, and `a = ...` moved the label `a` to it. `b` still points at the original. In Week 8 you'll meet lists, which **can** change, and this exact situation becomes a famous source of bugs.
