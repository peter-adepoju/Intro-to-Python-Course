## The accumulator pattern

An **accumulator** is a variable that collects a result as the loop runs. It always has the same shape:

```python
total = 0              # 1. start it before the loop, at a neutral value
n = 1
while n <= 4:
    total += n * n     # 2. add this pass's contribution
    n += 1
print(total)           # 3. use it after the loop
```

Watch it grow. Predict each row before you reveal it:

The **starting value** matters. For a sum, start at `0`, because adding 0 changes nothing. For a product, start at `1`, because multiplying by 1 changes nothing:

`5 × 4 × 3 × 2 × 1` is written **5!** and called "5 factorial". It counts the ways to put 5 things in order: 120.

## The best so far: maximum and minimum

To find the largest value, keep a variable holding **the largest seen so far**, and replace it whenever something bigger turns up:

Every value gets the same treatment: compare it with the record, and replace the record if it's beaten. With four separate variables you have to write that out four times. With values arriving one at a time in a loop, you write it once, as you'll see in a moment. (Lists, in Week 8, let you loop over stored values too.)

## Reading until the user says stop

Often you don't know how many values are coming. The user keeps typing, and a special value, called a **sentinel**, means "that's all". The cleanest shape uses a **priming read**: read the first value before the loop, and the next one at the end of the body.

Trace why this works:

- The first read happens before the condition is ever checked, so the condition has something to test.
- Inside the loop, the value is **used first**, then the next value is read.
- When the sentinel arrives, the condition is checked straight away, and the sentinel is never added.

If the user types `0` first, the body never runs and the total is 0. Press X-ray and try it.

An empty line is a good sentinel for text. Notice the guard after the loop: if nothing was entered, `total / count` would divide by zero.
