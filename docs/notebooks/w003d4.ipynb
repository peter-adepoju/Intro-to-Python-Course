## Why trace by hand?

X-ray shows you what a program does. Tracing by hand shows you what **you think** it does. A bug is exactly the place where the two disagree, so the skill is to write down your expectation first, then compare. Professional programmers trace in their heads all the time; they learned by doing it on paper.

## The trace table, pass by pass

On Day 1 of this week you saw a trace table with one row for every **line** Python runs. For loops, a more compact table is usually better: one row each time the **condition is checked**, showing every variable at that moment.

```python
total = 0
n = 5
while n > 0:
    total += n
    n -= 2
print(total)
```

| check | `n` | `total` | `n > 0`? |
|---|---|---|---|
| 1 | 5 | 0 | True |
| 2 | 3 | 5 | True |
| 3 | 1 | 8 | True |
| 4 | -1 | 9 | False |

Read down the table and every question answers itself:

- The body ran **3** times (the number of `True` rows).
- The final values are in the last row: `n` is -1, `total` is 9.
- It prints `9`.

Those are the **Fibonacci numbers**: each one is the sum of the two before. The pair `a, b` slides along the sequence one step per pass.

## What stays true: loop invariants

Look back at the table for the first loop. Something is true in **every** row: `total` plus all the numbers still to come equals 5 + 3 + 1. A statement that is true every time the condition is checked is called a **loop invariant**.

For the sum loop from Day 2:

```python
total = 0
i = 1
while i <= n:
    total += i
    i += 1
```

the invariant is: **`total` is the sum of every number before `i`**, that is 1 + 2 + ... + (i − 1).

- It's true at the start: `i` is 1, nothing has been added, and `total` is 0.
- If it's true at one check, the body keeps it true: it adds `i` and then moves `i` on.
- When the loop ends, `i` is `n + 1`, so the invariant says `total` is 1 + 2 + ... + n. That's exactly the answer you wanted.

You don't need to write invariants down for every loop. But asking "what does this variable **mean** at each check?" is the fastest way to understand a loop someone else wrote, and the best way to get your own loop right first time. You'll use invariants seriously in Phase 4, when you prove that algorithms work.

## A catalogue of loop bugs

Almost every loop bug is one of these five. Learn their symptoms, and you'll know where to look.

| Bug | Symptom | Where to look |
|---|---|---|
| **Off by one** | One pass too many or too few; first or last value missing | `<` versus `<=`; the counter's starting value |
| **Bad start** | Answer is wrong by a fixed amount, or always 0 | The accumulator's starting value (0 for sums, 1 for products) |
| **Reset inside** | Only the last pass counts | An accumulator set to its start value inside the body |
| **Never ends** | The program hangs | The update: missing, wrong direction, skipped by `continue` or an `if` |
| **Never starts** | Nothing happens, no error | The condition is false at the first check |
