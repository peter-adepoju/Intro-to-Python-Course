```python
import statistics as st

marks = [62, 71, 71, 58, 90, 45, 71, 83, 66]
print(st.mean(marks), st.median(marks), st.mode(marks))
print(round(st.stdev(marks), 2), round(st.pstdev(marks), 2))
print(st.quantiles(marks, n=4))
print(st.multimode([1, 1, 2, 2, 3]))
```

```python
import statistics as st

hours = [1, 2, 3, 4, 5, 6]
score = [52, 55, 61, 64, 70, 74]
print(round(st.correlation(hours, score), 3))
fit = st.linear_regression(hours, score)
print(round(fit.slope, 2), round(fit.intercept, 2))
print(round(fit.slope * 7 + fit.intercept, 1))     # predicted score for 7 hours
```

## Warm-up

Floats are stored in binary, so `0.1 + 0.2` is a tiny bit off 0.3, as you saw in Week 1. `math.isclose` is the safe way to compare them. And `round(2.5)` is 2: Python rounds halves to the nearest **even** number ("banker's rounding"), which avoids a bias upwards when rounding lots of values.

## math

`math` works with ordinary numbers (ints and floats). Here are the parts you'll reach for most, grouped by job.

**Rounding and parts of numbers:**

| Function | Gives | Example |
|---|---|---|
| `math.floor(x)` | the whole number at or below x | `floor(-2.5)` is -3 |
| `math.ceil(x)` | the whole number at or above x | `ceil(2.1)` is 3 |
| `math.trunc(x)` | x with the fraction chopped off | `trunc(-2.5)` is -2 |
| `math.isclose(a, b)` | whether a and b are equal, allowing for float error | `isclose(0.1 + 0.2, 0.3)` |
| `math.fsum(xs)` | an accurate sum of floats | better than `sum` for many floats |

**Powers, roots and logarithms:** `math.sqrt`, `math.isqrt` (whole-number square root, exact for huge numbers), `math.exp`, `math.log(x)` (natural), `math.log(x, base)`, `math.log2`, `math.log10`, and the constants `math.pi`, `math.e`, `math.inf`, `math.nan`.

**Counting and whole numbers:** `math.factorial(n)`, `math.comb(n, k)` (ways to choose k from n), `math.perm(n, k)` (ordered ways), `math.gcd(a, b, ...)`, `math.lcm(a, b, ...)`, `math.prod(xs)`.

**Geometry:** `math.hypot(x, y)`, `math.dist(p, q)`, and trigonometry with `math.sin`, `math.cos`, `math.tan`, which work in **radians**: convert with `math.radians(degrees)` and `math.degrees(radians)`.

A handy pattern: `math.inf` as a starting value for "the smallest so far", since every real number is smaller.

## random

`random` makes **pseudo-random** numbers: they come from a formula that looks random but is completely determined by a starting value, the **seed**. That's a feature: fix the seed and you get the same "random" results every time, which makes simulations and tests repeatable.

| Function | Gives |
|---|---|
| `random.random()` | a float from 0 up to (not including) 1 |
| `random.uniform(a, b)` | a float between a and b |
| `random.randint(a, b)` | a whole number from a to b, **both included** |
| `random.randrange(start, stop, step)` | like `range`, but one random value |
| `random.choice(seq)` | one item |
| `random.choices(seq, weights=..., k=n)` | n items, with replacement, optionally weighted |
| `random.sample(seq, k)` | k different items, without replacement |
| `random.shuffle(lst)` | shuffles a list in place (returns `None`) |
| `random.gauss(mu, sigma)` | a number from a bell curve (normal distribution) |

Prefer making your own `random.Random(seed)` object, as here, to calling `random.seed(...)`: the seed then affects only your generator, not any other code that uses `random`. The checks on this site use the same idea to make random exercises repeatable.

## statistics

`statistics` summarises data. Each function takes a list (or any sequence) of numbers:

- **mean**: the average. **median**: the middle value when sorted (the mean of the two middle values for an even count); it isn't pulled about by a few extreme values. **mode**: the most common value; `multimode` gives all of them when there's a tie.
- **stdev** and **variance**: how spread out the values are. Use `stdev` for a **sample** from a bigger population (it divides by n − 1), and `pstdev` when you have the **whole** population (divides by n).
- **quantiles(data, n=4)**: the cut points that split the data into n equal groups; with n = 4 those are the quartiles.
- **correlation(x, y)** and **linear_regression(x, y)**: how strongly two lists move together, and the straight line that fits them best.

An empty list raises `statistics.StatisticsError`, and so do some functions with too few values (`stdev` needs at least two). For big numerical work, the third-party `numpy` and `pandas` are much faster, but for everyday summaries `statistics` is right there.
