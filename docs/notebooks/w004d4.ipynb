> **Key idea.** **All of** starts `True` and looks for a failure. **Any of** starts `False` and looks for a success. Neither ever sets the flag back the other way inside the loop.

> **The flag that flips back.** A common bug in an all-of check is an `else` that sets the flag back to `True`. Then only the **last** item decides the answer:
> 
> ```python
> ok = True
> for ch in "ab3d":
>     if ch.isalpha():
>         ok = True      # wrong: this undoes an earlier failure
>     else:
>         ok = False
> ```
> 
> Here `ok` ends up `True` because the last character, `d`, is a letter, even though `3` isn't.

## Warm-up

That loop does two jobs at once: it **accumulates** a total and keeps the **best so far**. Both are patterns you've already used. Today they get names, along with eight more.

## Why name patterns?

When you read "count the words that start with a capital", you could stare at a blank screen. Or you could think: *count* (a counter that goes up), and *filter* (only some items). Two patterns you know, joined together. Naming patterns turns a new problem into a combination of old ones.

Each pattern below has three parts: what you set up **before** the loop, what you do **inside**, and what you have **after**.

## 1. Accumulate

Combine every item into one result.

Start with 0 for sums, 1 for products, `""` for strings.

## 2. Count

Count the items that pass a test. It's accumulate, adding 1 at a time.

## 3. Filter

Keep only the items that pass a test, building a new result.

## 4. Transform

Change **every** item, building a new result. No test, or a test that chooses between two changes.

## 5. Find first

Stop at the first item that passes a test. Afterwards, you need to know whether one was found.

## 6. All of

Is **every** item OK? Start by assuming yes, and look for one that isn't. One failure is enough to decide, so you can `break`.

## 7. Any of

Is **at least one** item OK? Start by assuming no, and look for one that is. One success is enough to decide.

## 8. Best so far

Find the largest or smallest. Start with the **first** item, and replace the record whenever it's beaten. Often you also want **where** the best was, so keep its position too:

Should a tie replace the record? With `>`, the **first** of several equal bests is kept. With `>=`, the **last** one is. Decide on purpose.

## 9. Running total

Print (or keep) the total **so far** at every step, not just the final one. Bank statements and scoreboards do this.

## 10. Compare with the previous

Each item is compared with the one **before** it: is it bigger, the same, part of a run? Keep the previous item in a variable, and update it at the **end** of each pass.

When the items are in a string, you can use positions instead, comparing `s[i]` with `s[i - 1]` for `i` from 1, as you did yesterday with neighbours.

## Choosing and combining

Here are some problems and the patterns they need:

| Problem | Patterns |
|---|---|
| Average of the readings | accumulate + count |
| Is this word a valid code (only capitals and digits)? | all of |
| Where is the first space? | find first |
| How long is the longest run of the same letter? | compare with previous + best so far |
| Remove the vowels and count what's removed | filter + count |
| Does the password contain a symbol? | any of |
