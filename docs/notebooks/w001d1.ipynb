````python
# Run this cell once. It lets you check exercise answers inside this notebook.
# Workflow: run your exercise cell, then immediately run the "check" cell below it.
import io, builtins

def check(tests_src):
    """Run an exercise's tests against the cell you ran just before this one."""
    ip = get_ipython()
    code = ip.user_ns["In"][-2]
    def run(feed=(), echo=False):
        out = io.StringIO(); q = [str(x) for x in feed]; real_in, real_print = builtins.input, builtins.print
        def fake_input(p=""):
            if not q:
                raise AssertionError("The program asked for more input than the test gave it.")
            v = q.pop(0); out.write(str(p) + (v + "\n" if echo else "")); return v
        def fake_print(*a, **k):
            k.setdefault("file", out); real_print(*a, **k)
        builtins.input, builtins.print = fake_input, fake_print
        try:
            exec(code, {"__name__": "__main__"})
        finally:
            builtins.input, builtins.print = real_in, real_print
        return out.getvalue()
    norm = lambda t: "\n".join(l.rstrip() for l in str(t).rstrip().splitlines())
    def expect_output(want, inputs=(), echo=False):
        got = run(inputs, echo)
        assert norm(got) == norm(want), f"expected {norm(want)!r} but got {norm(got)!r}"
    import re
    uses = lambda *names: all(re.search(r"\b" + re.escape(n) + r"\b", code) for n in names)
    def run_with(subs, feed=(), echo=False):
        nonlocal code
        saved = code
        for old, new in subs.items():
            code = code.replace(old, new, 1)
        try:
            return run(feed, echo)
        finally:
            code = saved
    env = dict(ip.user_ns, run=run, norm=norm, source=code, expect_output=expect_output, uses=uses, run_with=run_with)
    try:
        env["output"] = run()
    except AssertionError:
        env["output"] = ""
    exec(tests_src, env)
    tests = [(k, v) for k, v in env.items() if k.startswith("test_") and callable(v)]
    ok = 0
    for name, fn in tests:
        label = (fn.__doc__ or name[5:].replace("_", " ")).strip()
        try:
            fn(); ok += 1; print("✓", label)
        except AssertionError as e:
            print("✗", label + ":", e or "not quite yet")
    print(f"{ok} of {len(tests)} checks passed.")
````

## What is a computer?

A computer is a machine that follows instructions very fast and very exactly. It doesn't understand anything. It doesn't guess what you meant. It does exactly what it's told, billions of times a second.

Inside, there are two parts you need to know about today:

- **Memory** (also called RAM): a long row of numbered cells. Each cell holds a number. Your programs and all their data live here while they run.
- **The processor** (CPU): the part that does the work. It repeats one cycle forever: **fetch** the next instruction from memory, **decode** what it means, **execute** it. Then on to the next.

A special counter inside the processor, the **program counter**, remembers which cell holds the next instruction. Step through this toy computer and watch it work. The yellow arrow is the program counter.

Real processors work the same way, just with thousands of kinds of instruction and billions of steps per second. When you write Python, you never write `LOAD` or `STORE` yourself. Python turns your code into instructions like these for you.

## Everything is numbers

Memory cells only hold numbers. So how does a computer store text, pictures and music? It stores **numbers that stand for them**, and it agrees on a code for what each number means.

Inside the machine, each number is made of **bits**: tiny switches that are either off (0) or on (1). Eight bits make a **byte**. Each switch is worth double the one to its right: 1, 2, 4, 8, 16, 32, 64, 128. Add up the switches that are on and you get the number.

The same eight switches can mean the number 65 or the letter `A`. What decides? The **type** of the data: whether the program treats it as a number or as text. You'll meet types properly tomorrow. For now, remember this: *the computer only stores numbers; types tell it how to read them.*

## What is a program?

A **program** is a list of instructions for a computer, written in a language the computer (with some help) can follow. A **programming language** is a language designed for writing those instructions precisely, with no room for misunderstanding.

**Python** is a programming language. It was created by Guido van Rossum and first released in 1991. It's designed to be easy to read, which makes it a great first language, and it's also one of the most used languages in the world: for websites, science, data, artificial intelligence, automation, games and more. Learning Python well opens a huge number of doors.

When you press **Run**, a program called the **Python interpreter** reads your code and carries it out. You write text; the interpreter does the rest. Here's what it actually does with your text, step by step, this figure is calculated live by real Python:

You don't need to understand the syntax tree or the bytecode yet. We'll come back to them in Phase 5, when you study how Python works inside. Today, take away two ideas:

1. Python first **reads** all your code. If any of it breaks the grammar rules, it stops before running anything and reports a `SyntaxError`.
2. Then it **runs** your code, one instruction at a time, from top to bottom.

## Printing several things at once

## Comments

A **comment** is a note for humans. It starts with `#`, and Python ignores everything from the `#` to the end of the line.

Use comments to explain *why* your code does something. Programmers also "comment out" lines to switch them off for a moment while testing, as in the last line above.

## Code runs from top to bottom

Python runs your program one line at a time, from the first line to the last, in order. Use **X-ray** on this program and watch the yellow pointer move down. Watch the "Printed so far" box grow as each line runs.

This sounds obvious. But "Python does exactly one thing at a time, in order" is the most important idea in all of programming, and every confusing bug you'll ever meet is easier once you trace it line by line like this.

## Errors are clues, not failures

You *will* see errors. Every programmer sees them every day, experts included. An error message is Python telling you exactly what went wrong and where. Learning to read errors is one of the most valuable skills you'll build.
