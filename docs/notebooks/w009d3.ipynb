## Warm-up

You've used tuples since Week 1 without knowing their name. `1, 2` is a tuple; swapping builds one; a function returning two values returns one. The last line shows it, in round brackets.

## What a tuple is

A **tuple** is an ordered sequence, like a list, written with round brackets:

Indexing, slicing, `len`, `in`, `for`, `+` and `*` all work as they do for lists and strings.

Actually, it's the **commas** that make a tuple, not the brackets. The brackets are only needed when the tuple would otherwise be unclear:

### One item and no items

Because the comma makes the tuple, a tuple with one item needs a comma after it:

`(5)` is just the number 5 in brackets, like `(2 + 3)`. `(5,)` is a tuple holding 5.

### Tuples can't be changed

A tuple is **immutable**, like a string. Once it's made, its items can't be replaced, added or removed:

So tuples have only two methods, `count` and `index`, the ones that don't change anything:

To "change" a tuple, make a new one, just as with strings:

## Tuples or lists?

Both hold sequences. Choose by what the data **is**:

| Use a tuple when... | Use a list when... |
|---|---|
| the group has a **fixed** size and meaning: `(x, y)`, `(year, month, day)`, `(name, score)` | you'll **add or remove** items |
| the values shouldn't change | items are all the same kind of thing, in any number |
| you're returning several values from a function | you'll sort it in place, or change items |

A good rule of thumb: a **list** is like a column in a table (many values of the same kind), and a **tuple** is like a row (one record, with a fixed meaning for each position).

Immutability is also a guarantee. If you give a tuple to a function, you know for certain it comes back unchanged: none of last week's aliasing surprises can happen. And in Week 10 you'll see that tuples can be used as dictionary keys, which lists can't.
