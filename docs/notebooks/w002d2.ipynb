## Warm-up

## and: both must be true

`and` puts two conditions together. The whole thing is `True` only when **both** sides are `True`.

A cinema lets you in only if you're old enough *and* you have a ticket. If either part fails, you're not getting in.

## or: at least one must be true

`or` is `True` when **at least one** side is `True`. It's only `False` when both sides are `False`.

## not: flip it

`not` takes one condition and turns `True` into `False` and `False` into `True`.

## Truth tables

A **truth table** lists every combination of inputs and the answer for each. With two conditions there are only four combinations, so you can know the full behaviour of `and` and `or` by heart:

| `a` | `b` | `a and b` | `a or b` | `not a` |
|---|---|---|---|---|
| True | True | **True** | **True** | False |
| True | False | False | **True** | False |
| False | True | False | **True** | **True** |
| False | False | False | False | **True** |

Read the `and` column as "only the top row", and the `or` column as "all but the bottom row".

## Order of evaluation

When a condition mixes several operators, Python works in this order:

1. Arithmetic (`*`, `+` and the rest)
2. Comparisons (`==`, `<`, `in` and the rest)
3. `not`
4. `and`
5. `or`

So `not` binds tightest of the three, and `or` happens last. Step through this one:

`a or b and c` means `a or (b and c)`, because `and` goes first. If you have to stop and think about the order, so will anyone reading your code. Add brackets:

## Short-circuit evaluation

Python is lazy in a useful way. It works from left to right and stops as soon as it knows the answer:

- `False and anything` is always `False`, so Python never looks at `anything`.
- `True or anything` is always `True`, so again Python skips the right side.

This is called **short-circuit evaluation**. You can see it happen. In both lines below, the right side would crash if Python ever ran it:

## and and or return one of their sides

So far `and` and `or` have given `True` or `False`, because both sides were bools. Strictly, they give back **one of the two values**:

- `a or b` gives `a` if `a` is truthy, otherwise `b`.
- `a and b` gives `a` if `a` is falsy, otherwise `b`.

That explains why `x == 1 or 2` printed `2`. It also gives a neat way to supply a **default** when the user types nothing:

If the user types nothing, `input()` returns `""`, which is falsy, so `or` gives back `"friend"`.

## De Morgan's laws

Sometimes you need the opposite of a combined condition. Augustus De Morgan, another nineteenth-century mathematician, noticed two rules:

- `not (a and b)` is the same as `not a or not b`
- `not (a or b)` is the same as `not a and not b`

In words: when you push a `not` inside brackets, it flips each part, and `and` and `or` swap places. Check it yourself. The figure compares two conditions row by row:

An everyday example: "not (rainy and cold)" means "not rainy, or not cold": at least one of the two is fine.

Both are correct. The second, with no `not`, is usually easier to read.
